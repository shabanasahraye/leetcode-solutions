# <center> 332. Reconstruct Itinerary </center>


## Problem Description
[Click here](https://leetcode.com/problems/reconstruct-itinerary/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given flight tickets (directed edges) between airports (nodes). We need to reconstruct the itinerary starting from "JFK" and use all tickets exactly once. If multiple itineraries are possible, return the one with the smallest lexical (alphabetical) order.

Treat the tickets as a directed graph. Sort the tickets in reverse lexical order so that during DFS we can always remove the smallest destination first. Use DFS to visit every ticket exactly once and build the itinerary in reverse, then reverse the result at the end.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = a list for itinerary
- create a hashmap  to represent the graph (adjacency list)
    - *key = node*
    - *value = list of neighbors*
- Sort the tickets in reverse lexical order so that during DFS we can always pop the smallest destination first
- traverse the tickets and add edges to the graph
    - for each source node, and destination node in the sorted edges (tickets)
        - add an edge from the source node to the target node i.e add the destination node to the neighbors' list of the source node
            - *(the tickets are sorted in reverse, so the neighbors will be added in descending order)*
- define a helper function for DFS <br>
*dfs(node)*
    - loop while the current node has outgoing edges
        - remove the last neighbor (the smallest lexical destination) and continue DFS
            - *because the neighbors are stored in reverse lexical order, popping from the end always gives the smallest destination*
    - when a node has no outgoing edges left, add it to the result list
- start DFS from the JFK
- reverse the result list because we have constructed the itinerary in reverse and return it


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sorting tickets + graph creation + DFS + reversing result) → O(ElogE + E + E + E) → O(ElogE) → O(nlogn)
    - *V = total vertices/nodes = airports*
    - *E = total directed edges = tickets = n*
    - *each ticket is visited exactly once during DFS*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sorted tickets + graph + DFS recursion stack + result list) → O(E + (V + E) + E + E) → O(V + E) → O(E) → O(n)
    - *since every airport appears in at least one ticket, V ≤ E + 1, so O(V + E) simplifies to O(E)*


## Code

```python
class Solution:

    def findItinerary(self, tickets: List[List[str]]) -> List[str]:
        res = []
        graph = defaultdict(list)
        for src, dst in sorted(tickets, reverse=True):
            graph[src].append(dst)
        
        def dfs(node: str) -> None:
            while graph[node]:
                dfs(graph[node].pop())
            res.append(node)
        
        dfs('JFK')
        return res[::-1]
        
```
