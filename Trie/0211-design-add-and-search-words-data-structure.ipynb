## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Use a Trie data structure.

A Trie, also called a prefix tree, stores words character by character. Each node represents a character, allowing efficient insertion and search of words or prefixes.

Since the search word may contain `.` which can match any letter, a Trie allows us to explore only the possible paths that match the search word instead of checking every stored word.


## Approach
<!-- Describe your approach to solving the problem. -->
### Class TrieNode

**init()**
- set children = a hashmap to store the child nodes
    - *key = characters*
    - *value = child nodes*
- set isEnd = false to track if a node (character) is the last node (character) of a word i.e end of a word

### Class WordDictionary

**init()**
- create a root node

**insert()**
- *start from the root and create a node for each character if it doesn't already exist. Mark the last node as the end of the word*
    - set cur = root node to track current node
    - for each character in the word
        - if the character is not a child node of the current node
            - create a new node
        - set the current node to the child node of the character to move to the next node
    - at the end of the loop, the current node will point to the last node (character) of the word. So, set the isEnd attribute to true

**search()**
- *searching for a word in a Trie is like finding a path from the root to a word-ending node using DFS. If the search word contains `.`, explore all child nodes because `.` can match any letter*
    - define a helper function for DFS <br>
    *dfs(current node, current character index in the word)*
        - if the current index equals the length of the word
            - check if the current node is the last node (character) of the word
                - return the result
        - get the character at the current index in the search word
        - if the character is not `.`
            - if the character is a child node
            - search for the character in the child nodes
            - else return false
        - else the character is a `.`
        - continue DFS with each child node to find a matching path
    - start DFS from the root node with index 0 and return the result


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - init(): O(1)
    - addWord(): O(word length) → O(n)
    - search(): O(backtracking DFS) → O(total recursive calls × work per call) → O(total states explored × work per call) → O($26^n$ × constant) → O($26^n$)
        - *branching factor = number of child nodes explored when the current character is '.'*
        - *branching factor is variable because different trie nodes can have different numbers of children (from 0 to 26 for lowercase English letters)*
        - *as the branching factor is variable, total recursive calls = total states explored*
        - *in the worst case, every character is `.` and each level can branch to up to 26 child nodes*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - init(): O(1)
    - addWord(): O(word length) → O(n) 
        - *in the worst case, a new node is created for each character in the word*
    - search(): O(recursion stack) → O(recursion depth) → O(n)
        - *recursion depth is n because we process one character per level*
    - the total Trie space after multiple insertions is O(total characters inserted)


## Code

```python
class TrieNode:

    def __init__(self):
        self.children = defaultdict(TrieNode)
        self.isEnd = False


class WordDictionary:

    def __init__(self):
        self.root = TrieNode()

    def addWord(self, word: str) -> None:
        cur = self.root
        for c in word:
            if c not in cur.children:
                cur.children[c] = TrieNode()
            cur = cur.children[c]
        cur.isEnd = True

    def search(self, word: str) -> bool:

        def dfs(node: TrieNode, index: int) -> bool:
            if index == len(word):
                return node.isEnd
            c = word[index]
            if c != '.':
                if c in node.children:
                    return dfs(node.children[c], index + 1)
                return False
            return any(dfs(child, index + 1) for child in node.children.values())

        return dfs(self.root, 0)
        
    
        


# Your WordDictionary object will be instantiated and called as such:
# obj = WordDictionary()
# obj.addWord(word)
# param_2 = obj.search(word)
```
