## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Singly Linked List

#### List Node Class
*class to create linked list nodes*

**init()**
- set the node value and next pointer 

#### LinkedList Class
*class to create a singly linked list*

**init()**
- set head = none to track the head node
- set size = 0 to track the linked list size

**get()**
- if the index is out of bounds
    - return -1
- set the current node pointer to the head node
- move the current node index times to reach the required node and return its value

**addAtHead()**
- insert a node at the start (index 0) with the specified value

**addAtTail()**
- insert a node at the end (index = linked list size) with the specified value

**addAtIndex()**
- if the index is out of bounds
    - return
- set the current node pointer to head
- if the index is 0
    - create a new node and set its next node as the current node
    - make the new node as the head node
- else:
    - move the current pointer index - 1 times to reach the previous node i.e node at index before the required index
    - create a new node and set its next node to the next of the previous node
    - set the previous node's next pointer to the new node 
- increment size because a new node is added

**deleteAtIndex()**
- if the index is out of bounds
    - return
- if the index is 0, we have to delete the head node
    - make the second node as the head node
- else:
    - set current node to head
    - move current node pointer index - 1 times to reach the previous node i.e node at index before the required index
    - delete the required index node by updating the previous node's next pointer
- decrement size because a node is removed


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - Singly Linked List Approach:
        - init(): O(1)
        - get(): O(index) → O(n)
        - addAtHead(): O(1)
        - addAtTail(): O(n)
        - addAtIndex(): O(index) → O(n)
        - deleteAtIndex(): O(index) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - Singly Linked List Approach:
        - init(): O(1)
        - get(): O(1)
        - addAtHead(): O(1)
        - addAtTail(): O(1)
        - addAtIndex(): O(1)
        - deleteAtIndex() :O(1)
        - overall space: O(n)
            - *linked list size will be n after n calls to addAtIndex()*


## Code

```python
class ListNode():

    def __init__(self, val, next= None):
        self.val = val
        self.next = next


class MyLinkedList:

    def __init__(self):
        self.head = None
        self.size = 0

    def get(self, index: int) -> int:
        if index >= self.size:
            return -1
        cur = self.head
        for _ in range(index):
            cur = cur.next
        return cur.val

    def addAtHead(self, val: int) -> None:
        self.addAtIndex(0, val)
        
    def addAtTail(self, val: int) -> None:
        self.addAtIndex(self.size, val)

    def addAtIndex(self, index: int, val: int) -> None:
        if index > self.size:
            return
        cur = self.head
        if index == 0:
            new_node = ListNode(val, next= cur)
            self.head = new_node
        else:
            for _ in range(index - 1):
                cur = cur.next
            new_node = ListNode(val, next= cur.next)
            cur.next = new_node
        self.size += 1

    def deleteAtIndex(self, index: int) -> None:
        if index >= self.size:
            return
        if index == 0:
            self.head = self.head.next
        else:
            cur = self.head
            for _ in range(index - 1):
                cur = cur.next
            cur.next = cur.next.next
        self.size -= 1



# Your MyLinkedList object will be instantiated and called as such:
# obj = MyLinkedList()
# param_1 = obj.get(index)
# obj.addAtHead(val)
# obj.addAtTail(val)
# obj.addAtIndex(index,val)
# obj.deleteAtIndex(index)
```
