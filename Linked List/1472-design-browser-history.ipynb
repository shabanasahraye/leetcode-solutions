## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
- **Approach 1: Doubly Linked List**
    - use a doubly linked list because it allows us to move both backward and forward using next and previous node pointers 

- **Approach 2: Stack**
    - use two stacks: one for back history, and one for forward history to make navigation simple

- **Approach 3: Array**
    - use an array to store and access urls by index in constant time


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Doubly Linked List

#### Node Class
*class to create linked list nodes*

**init()**
- set the url, previous pointer, and next pointer

#### BrowserHistory Class

**init()**
- create a node for the homepage
- set cur = homepage node

**visit()**
- create a new node for the url
- connect the new node after the current node to discard the forward history
- set cur to the new node

**back()**
- while there is a previous node and steps remain
  - move cur to the previous node
  - decrement steps
- return the current url

**forward()**
- while there is a next node and steps remain
  - move cur to the next node
  - decrement steps
- return the current url

### Approach 2: Stack

**init()**
- add the homepage to the back stack
- create an empty forward stack

**visit()**
- add the new url to the back stack
- clear the forward stack to discard all forward history
- add the new url to the back stack

**back()**
- while more than one url is in the back stack and steps remain
  - move the current url to the forward stack
  - decrement steps
- return the top url of the back stack

**forward()**
- while the forward stack is not empty and steps remain
  - move the top url to the back stack
  - decrement steps
- return the top url of the back stack

###  Approach 3: Array

**init()**
- initialize an array with the homepage to store visited website urls
- set cur = 0 to track the index of the current url
- set end = 0 to track the last visited url index

**visit()**
- increment cur because we are visiting a new url
- if we are at the end of the array, add url at the end 
- else update the value at the current index, replacing the old forward URL
- update end to remove the forward urls
    - *note: end doesn't literally delete elements from the array, it marks the last valid history position, so old forward urls are ignored/overwritten*

**back()**
- move k steps backward to get the url index and set cur to this index because it is now the current url index 
    - *the index can go out of bounds, so take the max to find the earliest url we can go back to*
- return the url at the current index

**forward()**
- move k steps forward to get the url index and set cur to the new index because it is now the current url index 
    - *the index can go out of bounds, so take the min to find the latest url we can go forward to*
- return the url at the current index


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Doubly Linked List Approach:
        - init(): O(1)
        - visit(): O(1)
        - back(): O(loop) → O(number of steps moved backward) → O(k)
        - forward(): O(loop) → O(number of steps moved forward) → O(k)
  - Stack Approach:
        - init(): O(1)
        - visit(): O(clearing forward stack) 
            - *clear() may remove n elements*
        - back(): O(loop) → O(number of steps moved backward) → O(k)
        - forward(): O(loop) → O(number of steps moved forward) → O(k)
    - Array Approach:
        - init(): O(1)
        - visit(): O(1)
        - back(): O(1)
        - forward(): O(1)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Doubly Linked List Approach:
        - init(): O(1)
        - visit(): O(1)
        - back(): O(1)
        - forward(): O(1)
        - overall space: O(total urls stored in the linked list) → O(n) 
            - *linked list size will be n after n calls to visit()*
    - Stack Approach:
        - init(): O(1)
        - visit(): O(1)
        - back(): O(1)
        - forward(): O(1)
        - overall space: O(total urls stored in both stacks) → O(n) 
            - *stack size can be n after n calls to visit()*
    - Array Approach:
        - init(): O(1)
        - visit(): O(1)
        - back(): O(1)
        - forward(): O(1)
        - overall space: O(total urls stored in the array) → O(n) 
            - *array size can be n after n calls to visit()*


## Code

```python
# Approach 1: Doubly Linked List

class Node:
    def __init__(self, url):
        self.url = url
        self.prev = None
        self.next = None

class BrowserHistory:

    def __init__(self, homepage: str):
        self.cur = Node(homepage)

    def visit(self, url: str) -> None:
        node = Node(url)
        node.prev = self.cur
        self.cur.next = node
        self.cur = node

    def back(self, steps: int) -> str:
        while self.cur.prev and steps:
            self.cur = self.cur.prev
            steps -= 1
        return self.cur.url

    def forward(self, steps: int) -> str:
        while self.cur.next and steps:
            self.cur = self.cur.next
            steps -= 1
        return self.cur.url


# Approach 2: Stack

class BrowserHistory:

    def __init__(self, homepage: str):
        self.back_stack = [homepage]
        self.forward_stack = []

    def visit(self, url: str) -> None:
        self.back_stack.append(url)
        self.forward_stack.clear()

    def back(self, steps: int) -> str:
        while len(self.back_stack) > 1 and steps:
            self.forward_stack.append(self.back_stack.pop())
            steps -= 1
        return self.back_stack[-1]

    def forward(self, steps: int) -> str:
        while self.forward_stack and steps:
            self.back_stack.append(self.forward_stack.pop())
            steps -= 1
        return self.back_stack[-1]


# Approach 3: Array

class BrowserHistory:

    def __init__(self, homepage: str):
        self.history = [homepage]
        self.cur = 0
        self.end = 0

    def visit(self, url: str) -> None:
        self.cur += 1
        if self.cur == len(self.history):
            self.history.append(url)
        else:
            self.history[self.cur] = url
        self.end = self.cur

    def back(self, steps: int) -> str:
        self.cur = max(0, self.cur - steps)
        return self.history[self.cur]

    def forward(self, steps: int) -> str:
        self.cur = min(self.end, self.cur + steps)
        return self.history[self.cur]
        


# Your BrowserHistory object will be instantiated and called as such:
# obj = BrowserHistory(homepage)
# obj.visit(url)
# param_2 = obj.back(steps)
# param_3 = obj.forward(steps)
```
