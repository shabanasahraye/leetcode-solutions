# <center> 23. Merge k Sorted Lists </center>


## Problem Description
[Click here](https://leetcode.com/problems/merge-k-sorted-lists/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
This is same as [Merge Two Sorted Lists](https://leetcode.com/problems/merge-two-sorted-lists/). 

Use the divide and conquer approach.

Each time take two sorted lists and merge them into one sorted list. Repeat this until all the lists are merged.


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if the input contains no lists i.e [] all lists are empty i.e [[], []]
        - return null
    - if there is only one list
        - return it
    - if there are only two lists
        - merge them 
- set k = number of lists to be merged
- loop until all lists are merged i.e until k becomes zero
    - set left pointer l = 0 and right pointer r = k
    - loop until l < r
        - take two sorted lists (left and right lists), merge them using [Merge Two Sorted Lists Solution](https://leetcode.com/problems/merge-two-sorted-lists/solutions/3604740/o-min-m-n-time-o-1-space-solution-explained/) and store the merged list at the left list's position
        - increment l and decrement r to get the next pair
    - divide k by 2 because each time we merge k lists into k/2 lists i.e k/2 lists are left to be merged
- return the first list because it contains the final merged lis


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(outer loop × inner loop × merging two sorted lists) → O(logk × decreasing merges × n) → O(logk × n) → O(nlogk)
    - *the outer loop runs logk times because k is divided by 2 in each iteration
    - *for example, if k = 8, the outer loop runs log(8) = 3 times and the inner loop performs 4, 2, and 1 merges*
    - *the number of merges decreases but each round processes all n nodes*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(total new nodes created) → O(1)
    - *the existing linked-list nodes are reused, only constant extra pointers and a dummy node are used*


## Code

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next


class Solution:

    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        if not lists or not any(lists):
            return None
        if len(lists) == 1:
            return lists[0]
        if len(lists) == 2:
            return self.merge_sorted_lists(lists[0], lists[1])
        k = len(lists) - 1
        while k:
            l, r = 0, k
            while l < r:
                lists[l] = self.merge_sorted_lists(lists[l], lists[r])
                l += 1
                r -= 1
            k //= 2
        return lists[0]

    def merge_sorted_lists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        if not list1:
            return list2
        if not list2:
            return list1
        head = ListNode()
        cur = head
        while list1 and list2:
            if list1.val < list2.val:
                cur.next = list1
                list1 = list1.next
            else:
                cur.next = list2
                list2 = list2.next
            cur = cur.next
        cur.next = list1 or list2
        return head.next
        
```
