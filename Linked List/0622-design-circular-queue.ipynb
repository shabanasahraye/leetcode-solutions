## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Implement the circular queue using a linked list.

A circular queue is a normal queue where the tail connects back to the head


## Approach
<!-- Describe your approach to solving the problem. -->

#### ListNode Class
*class for linked list nodes*

**init()**
- set the node value and next node pointer

#### MyCircularQueue Class
*class to implement circular queue using linked list*

**init()**
- set head and tail node pointers to null
    - *head → front of queue (where you remove)*
    - *tail → end of queue (where you add)*
- set the queue capacity to k
- set size = 0 to track the current number of elements in the queue

**enQueue()**
- if the queue is full, insertion is not possible
    - return false
- create a node for the value to be inserted
- if the queue is empty
    - set the head and tail of the linked list to the node
- else add the node after tail and move tail to the new node i.e set the tail and tail next to the node
- connect tail back to head
    - *this is to maintain circular link after every operation*
- increment size
- return true

**deQueue()**
- if the queue is empty, deletion is not possible
    - return false
- if there is only 1 element in the queue
    - delete it by setting head and tail to null
        - *no need to connect tail back to head because queue is empty*
- else 
    - delete an element from the queue front (linked list start) by updating head to head.next
    - connect tail back to head
- decrement size
- return true

**Front()**
- if the queue is empty, return -1, else return the head value (first node value in the linked list)

**Rear()**
- if the queue is empty, return -1, else return the tail value (last node value in the linked list)

**isEmpty()**
- check if size of the queue is 0 and return the result

**isFull()**
- check if size of the queue is equal to the capacity and return the result


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - init(): O(1)
    - enQueue(): O(1)
    - deQueue(): O(1)
    - Front(): O(1)
    - Rear(): O(1)
    - isEmpty(): O(1)
    - isFull(): O(1)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - init(): O(1)
    - enQueue(): O(1)
    - deQueue(): O(1)
    - Front(): O(1)
    - Rear(): O(1)
    - isEmpty(): O(1)
    - isFull(): O(1)
    - overall space: O(total elements stored in the queue) → O(k)
        - queue size will be k after k calls to enQueue() *


## Code

```python
class Node:

    def __init__(self, val: int):
        self.val = val
        self.next = None


class MyCircularQueue:

    def __init__(self, k: int):
        self.head = None
        self.tail = None
        self.capacity = k
        self.size = 0

    def enQueue(self, value: int) -> bool:
        if self.isFull():
            return False
        node = Node(value)
        if self.size == 0:
            self.head = node
            self.tail = node
        else:
            self.tail.next = node
            self.tail = node
        self.tail.next = self.head  # skip in normal queue
        self.size += 1
        return True

    def deQueue(self) -> bool:
        if self.isEmpty():
            return False
        if self.size == 1: 
            self.head = None
            self.tail = None # skip in normal queue
        else:
            self.head = self.head.next
            self.tail.next = self.head  # skip in normal queue
        self.size -= 1
        return True

    def Front(self) -> int:
        return -1 if self.isEmpty() else self.head.val

    def Rear(self) -> int:
        return -1 if self.isEmpty() else self.tail.val

    def isEmpty(self) -> bool:
        return self.size == 0

    def isFull(self) -> bool:
        return self.capacity == self.size
    


# Your MyCircularQueue object will be instantiated and called as such:
# obj = MyCircularQueue(k)
# param_1 = obj.enQueue(value)
# param_2 = obj.deQueue()
# param_3 = obj.Front()
# param_4 = obj.Rear()
# param_5 = obj.isEmpty()
# param_6 = obj.isFull()

```
