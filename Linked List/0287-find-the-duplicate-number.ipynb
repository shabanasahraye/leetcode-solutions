# <center> 287. Find the Duplicate Number </center>


## Problem Description
[Click here](https://leetcode.com/problems/find-the-duplicate-number/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
A duplicate number creates a cycle in a list. To find cycle in a list, use slow-fast pointer algorithm. We need to find the duplicate number i.e the cycle entrance point.

Use slow-fast pointers approach: 
    - move the slow pointer one step and the fast pointer two steps at a time
    - they'll meet in the cycle, but the intersection point is not necessarily the cycle entrance
    - to find the cycle entrance, slow down the fast,  fast will start from the starting position and slow from the intersection point 
    - move both pointers one step at a time, this time they'll meet at the cycle entrance

*Note: There are n + 1 numbers in range 1 to n i.e 1, 2, 3, 4, ...* <br> 
*Each new element in the sequence points to the next element using its value as an index: x → nums[x] → nums[nums[x]] → nums[nums[nums[x]]] → ...*


## Approach
<!-- Describe your approach to solving the problem. -->
- set slow and fast at the start of the list (first element)
- loop until slow pointer meets the fast (intersection point)
    - move slow one step forward
    - move fast two steps forward
    - break when slow = fast
- reset fast to the starting position
- loop until the slow pointer meets the fast again (entrance point)
    - move slow one step forward
    - move fast one step forward
- return slow


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(finding intersection point + finding entrance point) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def findDuplicate(self, nums: List[int]) -> int:
        slow, fast = nums[0], nums[0]
        while True:
            slow = nums[slow]
            fast = nums[nums[fast]]
            if slow == fast:
                break
        fast = nums[0]
        while slow != fast:
            slow = nums[slow]
            fast = nums[fast]
        return slow
        
```
