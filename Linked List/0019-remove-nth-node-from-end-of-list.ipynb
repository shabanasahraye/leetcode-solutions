# <center> 19. Remove Nth Node From End of List </center>


## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
To remove the nth node from end, we need to update the next pointer of its previous node i.e (n+1)th node from the end. In a singly linked list, we can't move in backward direction (end to start) to find the nth node from end.

- **Approach 1:**
    - the nth node from end is the same as (total nodes - n)th node from start
    - for this, we need to find the list size (total nodes), which requires an extra pass/loop
- **Approach 2: Two-Pointers**
    - use the slow-fast pointer approach
    - give the fast pointer an earlier start by moving it to n node from the start and then move both pointers at the same pace, by the time fast pointer reaches the end, the slow pointer will be at (n+1)th node from end


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: Two-Pointers
- set slow and fast pointers at the start of the list
- loop n times to move the fast pointer (n+1)th node from the start 
    - *the range is n but due to next pointer, the fast pointer will point to (n+1)th node*
- if fast becomes null, the first node is the target node
    - delete it by returning the second node as start of the list i.e head.next 
- loop until the fast pointer reaches the end
    - move both pointers one node at a time
- after the loop, the slow pointer will be at (n+1)th node from the end. update its next pointer to remove the next node i.e nth node
- return head


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Two-Pointers Approach: O(moving fast pointer n nodes + moving both pointers to the end) → O(n + (m - n)) → O(m)
        - *m = total number of nodes, and m ≥ n*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Two-Pointers Approach: O(total new nodes created) → O(0) → O(1)


## Code

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next


class Solution:
    
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:

        # Approach 2: Two-Pointers
        slow = fast = head
        for _ in range(n):
            fast = fast.next
        if not fast:
            return head.next
        while fast.next:
            slow = slow.next
            fast = fast.next
        slow.next = slow.next.next
        return head
```
