## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Other stack operations take O(1) time. Finding the minimum element takes O(n) in a stack. To make it O(1), use a variable to track the minimum value.


## Approach
<!-- Describe your approach to solving the problem. -->
**init()**
- initialize an empty list to represent stack
- initialize a variable min = largest possible value to track the minimum value in the stack

**push()**
- *store the previous minimum below the new minimum*
    - if the new element <= current minimum
        - the new element becomes the minimum and the current minimum becomes the previous minimum, add the previous minimum first and then the new minimum
        - update min
    - else only add the new element

**pop()**
- get the stack top element 
- remove the stack top (last element)
- if the removed element is the minimum, we need to update min
    - update min to the previous minimum stored below
    - remove stack top again to remove the stored minimum

**top()**
- return the last/rightmost element in the stack 

**getMin()**
- return min 


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - init(): O(1)
    - push(): O(1)
    - pop(): O(1)
    - top(): O(1)
    - getMin(): O(1)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - init(): O(1)
    - push(): O(1)
    - pop()L O(1)
    - top(): O(1)
    - getMin(): O(1)
    - overall space: O(total elements stored in the min stack) → O(n + 1) → O(n)
        - *stack size will be n after n calls to push()*


## Code

```python
class MinStack:

    def __init__(self):
        self.stack = []
        self.min = float('inf')

    def push(self, val: int) -> None:
        if val <= self.min:
            self.stack.append(self.min)
            self.min = val
        self.stack.append(val)
        
    def pop(self) -> None:
        top = self.stack[-1]
        self.stack.pop()
        if top == self.min:
            self.min = self.stack[-1]
            self.stack.pop()

    def top(self) -> int:
        return self.stack[-1]
        
    def getMin(self) -> int:
        return self.min
        

# Your MinStack object will be instantiated and called as such:
# obj = MinStack()
# obj.push(val)
# obj.pop()
# param_3 = obj.top()
# param_4 = obj.getMin()
```
