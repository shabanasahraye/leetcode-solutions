## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
It is the same [Longest Repeating Character Replacement](https://leetcode.com/problems/longest-repeating-character-replacement/description/) problem, but here we are dealing with numbers instead of characters.

We need to find the longest window (subarray) where the number of operations required to make all elements (numbers) the same is less than or equal to k.

One way to reduce the number of operations is to make all values equal to the max value in the window

*no. of operations = desired sum - current sum = subarray size * subarray max - current sum*

Use a variable-size sliding window:
- expand the window by adding elements to the right
- when the window becomes invalid (operations > k), shrink it by removing elements from the left until it becomes valid
- keep track of the largest valid window
    -  use a running sum to calculate the number of operations
    - sort the array so the rightmost element is always the maximum value in the window

*Note: in a variable-size window, expand or shrink the window based on the problem's validity condition. Here, the window expands until invalid, then shrinks until valid.*

Example: nums = [1, 9, 2, 6], k = 5
- sort → [1, 2, 6, 9]
- [1] → max = 1, window size = 1, current sum = 1, desired sum = 1 × 1 = 1, operations = 1 - 1 = 0 → 0 < k → valid
- [1, 2] → max = 2, window size = 2, current sum = 3, desired sum = 2 × 2 = 4, operations = 4 - 3 = 1 → 1 < k → valid
- [1, 2, 6] → max = 6, window size = 3, current sum = 9, desired sum = 3 × 6 = 18, operations = 18 - 9 = 9 → 9 > k → invalid
- shrink → [2, 6] → max = 6, window size = 2, current sum = 8, desired sum = 2 × 6 = 12, operations = 12 - 8 = 4 → 4 < k → valid
- maximum valid window size = 2


## Approach
<!-- Describe your approach to solving the problem. -->
- sort the array
- set res = 0 to track the length of the largest valid window
- set window_sum = 0 to track the current window sum
- set left pointer l = 0 and right pointer r = 0 to track the window left and right sides
- loop while r < len(nums) to traverse the array
    - *expand the window by adding the current element*
        - add current element to window_sum 
    - while the no. of operations is greater than k, the window is invalid
        - *shrink the window by removing the left element*
            - subtracting the left element from window_sum
            - move left pointer
    - move right pointer at the end
        - *this keeps r at the current/rightmost element for calculating the max value*
    - update res with the length of the longest valid window (max of the current window size and previous maximum)


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort + traversal + window shrinking) → O(nlogn + n + n) → O(nlogn)
    - *the costs are added because the inner loop runs n in total, not n for each outer iteration*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort) → O(n)


## Code

```python
class Solution:

    def maxFrequency(self, nums: List[int], k: int) -> int:
        nums.sort()
        res, window_sum = 0, 0
        l, r = 0, 0
        while r < len(nums):
            # expand the window
            window_sum += nums[r]

            # shrink the window
            while (r - l + 1) * nums[r] - window_sum > k:
                window_sum -= nums[l]
                l += 1

            # update result
            res = max(res, r - l + 1)

            # move right pointer
            r += 1

        return  res
        
```
