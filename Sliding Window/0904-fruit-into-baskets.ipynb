## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the longest window (subarray) with at most two distinct elements. This is similar to
- [Longest substring With At Most Two Distinct Characters](https://leetcode.com/problems/longest-substring-with-at-most-two-distinct-characters/description/)
- [Longest Substring With At Most K Distinct Characters](https://leetcode.com/problems/longest-substring-with-at-most-k-distinct-characters/)

Use a variable-size sliding window:
- expand the window by adding elements to the right
- when the window becomes invalid (more than two distinct elements), shrink it by removing elements from the left until it becomes valid
- keep track of the largest valid window
    - use a hashmap to track the frequency of each element in the window

*Note: in a variable-size window, expand or shrink the window based on the problem's validity condition. Here, the window expands until invalid, then shrinks until valid to find the largest valid window.*


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the length of the largest valid window
- set count = a hashmap to store the frequency of each element in the current window
    - *key = element*
    - *value = frequency of the element*
- set left pointer l = 0 and right pointer r = 0 to track the window left and right sides
- loop while r < len(fruits) to traverse the array
    - *expand the window by adding the current element*
        - increment the count of fruit at position r
        - move right pointer
    - while the current window is invalid, i.e. the hashmap has more than 2 keys
        - *shrink the window by removing the left element*
            - decrement the count of fruit at position l
            - if its count is 0
                - delete its entry from the hashmap 
            - move left pointer
    - update res with the largest valid window (max of the current window size and previous maximum)
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(2) → O(1)
    - *2 because at most two distinct elements are stored in the hashmap*


## Code

```python
class Solution:

    def totalFruit(self, fruits: List[int]) -> int:
        res = 0
        count = defaultdict(int)
        l, r = 0, 0
        while r < len(fruits):
            # expand window
            count[fruits[r]] += 1
            r += 1

            while len(count) > 2:
                # shrink window
                count[fruits[l]] -= 1
                if not count[fruits[l]]:
                    del count[fruits[l]]
                l += 1
        
            # update result
            res = max(res, r - l)
        return res
        
```
