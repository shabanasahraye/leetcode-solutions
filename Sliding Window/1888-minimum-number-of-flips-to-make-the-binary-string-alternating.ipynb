## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the min flips required to make the binary string alternating. 

We can perform two operations: rotate the string and flip a character. If the string length is:
- even → rotating does not change the number of flips needed
- odd → rotating may reduce the number of flips

An alternating binary string can start with:
- **0 → 0101...**
    - 0's are required at the even indices and 1's at the odd indices*
- **1 → 1010...**
    - 0's are required at the odd indices and 1's at the even indices

There are two approaches to find the minimum number of flips.
- **Approach 1: Index Mismatch**
    - for each alternating pattern, count the positions where the current character does not match the expected character.
    - for even-length string, return the min number of flips
    - for odd-length string, check every rotation because it changes which characters are at even and odd indices 
        - instead of physically rotating the string, update the index counts because rotation changes the even/odd index positions in an odd-length string.
- **Approach 2: Sliding Window**
    - we can treat each rotation as a window of size n in the conceptual string s + s.
    - use a fixed-size sliding window:
        - expand the window by adding the current element
        - when the window becomes larger than n, slide the window by removing the left element's contribution
        - when the window reaches size n, calculate the flips needed
            - count mismatches with the `0101...` pattern
            - mismatches with the `1010...` pattern are `n - flips`
        - keep track of the minimum number of flips


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Index Mismatch
- set n = size of input string
- create an even hashmap to count the frequency of 0's and 1's at even indices
- create an odd hashmap to count the frequency of 0's and 1's at odd indices
- traverse the string and count the even-odd index flips <br>
*for each index i in the range 0 to n*
    - if current index is odd
        - increment the count of current character in the odd hashmap
            - *if character is 1, it needs to be flipped for case II because 1 is required at even indices*
            - *if character is 0, it needs to be flipped for case I because 0 is required at even indices*
    - else current index is even
        - increment the count of current character in the even hashmap
            - *if character is 1, it needs to be flipped for case I because 1 is required at odd indices*
            - *if character is 0, it needs to be flipped for case II because 0 is required at odd indices*
- set res = minimum of case i flips and case ii flips to track the minimum flips
- if n is odd, we need to perform the rotate operation also
    - traverse string s
        - decrement the flip count of current char in even hashmap
        - increment the flip count of current char in odd hashmap
        - swap even odd hashmaps
        - find min flips of the two cases and update res 
- return res

### Approach 2: Sliding Window
- set n = input string size
- set res = n to track the min number of flips
- set flips = 0 to track mismatches with the `0101...` pattern
- set left pointer l = 0 and right pointer r = 0 to track the window left and right sides
- loop while r < 2n to traverse the conceptual string `s + s`
    - *expand the window by adding the current element*
        - if the current char does not match the expected `0101...` pattern
            - increment flips
        - move right pointer
    - if the window becomes larger than n, slide the window by removing the left element's contribution
        - if the left char was a mismatch
            - decrement flips
        - move left pointer
    - if the window reaches size n
        - update res with the minimum of flips needed for both cases
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - Index Mismatch Approach: O(string traversal + rotations) → O(n + n) → O(n)
    - Sliding Window Approach: O(conceptual string traversal) → O(2n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Index Mismatch Approach: O(hashmaps) → O(2 + 2) → O(1)
        - *2 because each hashmap has two keys i.e 0 and 1*
    - Sliding Window Approach: O(1)

```python
class Solution:
    
    def minFlips(self, s: str) -> int:

        # Approach 1: Index Mismatch
        # n = len(s)
        # even, odd = {'0': 0, '1': 0}, {'0': 0, '1': 0}
        # for i in range(n):
        #     if i & 1:
        #         odd[s[i]] += 1
        #     else: 
        #         even[s[i]] += 1
        # res = min(odd['0'] + even['1'], odd['1'] + even['0'])
        # if n & 1: 
        #     for c in s:
        #         even[c] -= 1
        #         odd[c] += 1
        #         even, odd = odd, even
        #         res = min(res, odd['0'] + even['1'], odd['1'] + even['0'])
        # return res


        # Approach 2: Sliding Window
        n = len(s)
        res = n
        flips = 0
        l, r = 0, 0
        while r < (2 * n):
            # expand window
            if s[r % n] != str(r % 2):
                flips += 1
            r += 1

            # slide window
            if r - l > n:
                if s[l % n] != str(l % 2):
                    flips -= 1
                l += 1

            # update result
            if r - l == n:
                res = min(res, flips, n - flips)

        return res

        
```
