# <center> 1343. Number of Sub-arrays of Size K and Average Greater than or Equal to Threshold </center>


## Problem Description
[Click here](https://leetcode.com/problems/number-of-sub-arrays-of-size-k-and-average-greater-than-or-equal-to-threshold/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find all windows (subarrays) of size k with average >= threshold.

Use a fixed-size sliding window:
- expand the window by adding elements to the right until its size becomes k
- once the window reaches size k, check whether it is valid (average >= threshold)
- slide the window by moving the left pointer and adding the next element from the right
- keep track of the number of valid windows

*Note: in a fixed-size window, we slide the window by moving the left pointer and adding the next right element to keep the window size fixed.*


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the total number of valid windows
- set window_sum = 0 to track the current window sum
- find the sum of the first k - 1 elements
    - *this is to start with a window of size k - 1*
- set left pointer l = 0 to track the left side of the window
- use the right pointer r = k - 1 to traverse the array and track the right side of the window
- traverse the array using the right pointer
    - *expand the window by adding the current element*
        - add the right-side element to the window sum to make the window size equal to k
        - *move the right pointer, which will auto-update in the for-loop*
    - if current window is valid i.e its average is >= threshold
        - increment res
    - slide the window to maintain a fixed size of k
        - subtract left element from the window sum
        - move the left pointer 
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
Time complexity: O(finding the sum of the first k - 1 elements + traversing the remaining elements) → O(k + n - k) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def numOfSubarrays(self, arr: List[int], k: int, threshold: int) -> int:
        res, window_sum = 0, 0
        for i in range(k - 1):
            window_sum += arr[i]
        l = 0
        for r in range(k - 1, len(arr)):
            # expand window
            window_sum += arr[r]

            # update result
            if window_sum / k >= threshold:
                res += 1
            
            # slide window
            window_sum -= arr[l]
            l += 1
            
        return res
        
```
