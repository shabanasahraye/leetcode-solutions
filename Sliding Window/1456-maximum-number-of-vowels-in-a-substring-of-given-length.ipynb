# <center> 1456. Maximum Number of Vowels in a Substring of Given Length </center>


## Problem Description
[Click here](https://leetcode.com/problems/maximum-number-of-vowels-in-a-substring-of-given-length/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the maximum number of vowels in any substring of size k

Use a fixed-size sliding window:
- expand the window by adding elements to the right
- once the window becomes larger than k, slide the window by removing the left element and adding the next right element
- keep track of the maximum number of vowels in the window
    - use a hashset to check whether a character is a vowel

*Note: in a fixed-size window, we slide the window by removing the left element and adding the next right element to keep the size fixed.*


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the max number of vowels
- set count = 0 to count vowels in the current window
- set vowels = a hashset containing all vowel letters 
- set left pointer l = 0 and right pointer r = 0 to track the window left and right sides
- loop while r < len(s) to traverse the string
    - *expand the window by adding the current element*
        - if current character is a vowel
            - increment its count to add it to the window
        - move right pointer
    - if the window becomes larger than k, slide the window
        - if the left character is a vowel
            - decrement the count
        - move left pointer
    - update res with the maximum vowel count
- return res 


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashset) → O(vowels) → O(5) → O(1)


## Code

```python
class Solution:

    def maxVowels(self, s: str, k: int) -> int:
        res = 0
        count = 0
        vowels = {'a', 'e', 'i', 'o', 'u'}
        l, r = 0, 0
        while r < len(s):
            # expand the window
            if s[r] in vowels:
                count += 1
            r += 1

            # slide the window
            if r - l > k:
                if s[l] in vowels:
                    count -= 1
                l += 1
            
             # update result
            res = max(res, count)
        return res
    
        
```
