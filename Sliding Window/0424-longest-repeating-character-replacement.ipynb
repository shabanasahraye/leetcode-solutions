# <center> 424. Longest Repeating Character Replacement </center>


## Problem Description
[Click here](https://leetcode.com/problems/longest-repeating-character-replacement/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the longest window (substring) where the number of replacements required to make all elements the same is less than or equal to k.

Number of replacements = window size - frequency of the most frequent element

Use a variable-size sliding window:
- expand the window by adding elements
- when the window becomes invalid (replacements > k), shrink it by removing elements from the left until it becomes valid
- keep track of the largest valid window
    - use a hashmap to track the element frequencies
    - use max_f to track the frequency of the most frequent element

*Note: in a variable-size window, expand or shrink the window based on the problem's validity condition. Here, expand the window until invalid, then shrink until valid to find the largest valid window.*


## Approach
<!-- Describe your approach to solving the problem. -->
- set count = a hashmap to store element frequencies in the current window
- set res = 0 to track the length of the max size valid window
- set max_f = 0 to track the frequency of the most frequent element
- set left pointer l = 0 to track the left side of the current window
- use the right pointer r = 0 to traverse the string and track the right side of the window
- traverse the string using the right pointer
    - *expand the window by adding the current element*
        - update the current character count
        - update max_f by taking max of current character freq and previous max
        - *move the right pointer, which will auto-update in the for-loop*
    - if no. of replacements is more than k, the window is invalid,
        - *shrink window by removing the left element*
            - decrement the left char count
            - move the left pointer
    - update res with the largest valid window
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(uppercase English letters) → O(26) → O(1)


## Code

```python
class Solution:

    def characterReplacement(self, s: str, k: int) -> int:
        count = defaultdict(int)
        res, max_f = 0, 0
        l = 0
        for r in range(len(s)):
            # expand window
            count[s[r]] += 1
            max_f = max(max_f, count[s[r]])
            
            if (r - l + 1) - max_f > k:
                # shrink window
                count[s[l]] -= 1
                l += 1

            # update result
            res = max(res, r - l + 1)
        return res
        
```
