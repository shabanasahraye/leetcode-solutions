## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
It is the same as [Find All Anagrams in a String](https://leetcode.com/problems/find-all-anagrams-in-a-string/description/). The difference is here we have to find just one anagram (permutation).

A permutation of a string must have the same character frequencies as string 1, so we need to find a window of size s1 in s2 with the same frequencies.

Use a fixed-size sliding window:
- expand the window by adding elements
- once the window reaches size s1, slide it by moving the left pointer and adding the next element from the right
- check if the window has the same character frequencies as s1
    - use a hashmap to track the character frequencies in the window

*Note: in a fixed-size window, we slide the window by moving the left pointer and adding the next right element to keep the window size fixed.*


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if s1 size is greater than s2, there's no way we can find a permutation of s1 in s2
        - return false
- set count = a hashmap to store the required character frequencies in s1
    - *key = char in s1*
    - *value = frequency of char in s1*
    - *either loop to fill the hashmap or use Counter()*
- set window = a hashmap to store the character frequencies in the current window
    - *key = char in the current window*
    - *value = frequency of char in the current window*
- set left pointer l = 0 to track the left side of the window
- use the right pointer r = 0 to traverse s2 and track the right side of the window
- traverse s2 using the right pointer
    - *expand the window by adding the current element*
        - increment the current element count in the window hashmap
        - *move the right pointer, which will auto-update in the for-loop*
    - if the window becomes larger than s1, slide the window by moving the left pointer
        - decrement the count of the left element in the window
        - move the left pointer
    - if the window is valid (has the same character frequencies as s1), we found a permutation
        - return True 
- if the loop completes without returning true, there is no permutation, return false


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(counting frequencies in string1 + string2 traversal) → O(m + n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(unique characters in string1) → O(lowercase english letters) → O(26) → O(1)


## Code

```python
class Solution:

    def checkInclusion(self, s1: str, s2: str) -> bool:
        if len(s1) > len(s2):
            return False
        count = Counter(s1)
        window = Counter()
        l = 0

        for r in range(len(s2)):
            # expand the window
            window[s2[r]] += 1

            # slide the window
            if r - l + 1 > len(s1):
                window[s2[l]] -= 1
                l += 1

            # check if window is valid
            if window == count:
                return True

        return False
        
```
