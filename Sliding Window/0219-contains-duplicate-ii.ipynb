## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find a duplicate within k positions, so we only need to check elements within a window of size k.

Use a fixed-size sliding window:
- expand the window by adding the current element
- once the window reaches size k, slide the window by moving the left pointer
    - use a hash set to track the elements in the current window and check for duplicates

*Note: in a fixed-size window, we slide the window by moving the left pointer and adding the next right element to keep the size at k + 1.*


## Approach
<!-- Describe your approach to solving the problem. -->
- set seen = a hash set to store the elements in the current window
- set left pointer l = 0 and right pointer r = 0 to track the window left and right sides
- loop while r < len(nums) to traverse the array
    - *expand the window by adding the current element*
        - if the current number is already in the set, it is a duplicate
            - return true
        - else add it to the set
        - move right pointer
    - if the window reaches size k, slide the window by moving the left pointer
        - remove the left element from seen
        - move left pointer
- if the loop completes without returning true, there's no duplicate, return false


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashset) → O(window size) → O(k)


## Code

```python
class Solution:

    def containsNearbyDuplicate(self, nums: List[int], k: int) -> bool:
        seen = set()
        l, r = 0, 0
        while r < len(nums):
            # expand the window
            if nums[r] in seen:
                return True
            seen.add(nums[r])
            r += 1
            
            # slide the window
            if r - l >= k:
                seen.remove(nums[l])
                l += 1
                
        return False
        
```
