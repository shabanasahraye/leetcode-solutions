# <center> 239. Sliding Window Maximum </center>


## Problem Description
[Click here](https://leetcode.com/problems/sliding-window-maximum/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the maximum element in every window (subarray) of size k.

Use a fixed-size sliding window:
- expand the window by adding elements
- once the window reaches size k, slide the window by moving the left pointer and adding the next element from the right
- keep track of the maximum element in the current window
    - to get the max element in a window of size k, we can use:
        - linear search O(k)
        - max heap O(logk)
        - double-ended queue O(1) time
            - Deque allows insertion and deletion at both ends in O(1) time
            - keep indices in decreasing order of their values, so the largest element is always at the front of the deque

*Note: in a fixed-size window, we slide the window by moving the left pointer and adding the next right element to keep the size fixed.*


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if k is 1
        - return the input array 
    - if k is equal to the length of the array
        - return the max of the input array
- set dq = a deque to store indices of elements in decreasing order of their values
- set res = a list to store the max value in each window
- set left pointer l = 0 to track the left side of the window
- use the right pointer r = 0 to traverse the array and track the right side of the window
- traverse the array using the right pointer
    - *expand the window by adding the current element*
        - while dq is not empty and the current element is greater than the element at the rear of dq
            - remove the rear element from dq
        - add the current index to dq
        - *move the right pointer, which will auto-update in the for-loop*
    - *if the window reaches size k, slide the window by removing the left pointer*
        - if the leftmost index in dq is outside the window
            - remove it from dq
        - move the left pointer
    - if the window is valid
        - add the maximum element nums[dq[0]] to res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(deque) → O(window size) → O(k)


## Code

```python
class Solution:

    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        dq = deque()
        res = []
        l = 0
        for r in range(len(nums)):
            # expand the window
            while dq and nums[dq[-1]] < nums[r]:
                dq.pop()
            dq.append(r)

            # slide the window
            if r - l + 1 > k:
                if dq[0] == l:
                    dq.popleft()
                l += 1

            # update result
            if r - l + 1 == k:
                res.append(nums[dq[0]])

        return res
        
```
