# <center> 438. Find All Anagrams in a String </center>


## Problem Description
[Click here](https://leetcode.com/problems/find-all-anagrams-in-a-string/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Two strings are anagrams if they have the same characters with the same frequencies.

Use a fixed-size sliding window:
- expand the window by adding elements
- once the window reaches size p, slide it by moving the left pointer and adding the next element from the right
- if the window has the same character frequencies as p, add its start index to the result
    - use a hashmap to track the required character frequencies and the current window


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if the p size is greater than s, there can't be an anagram of p in s
        - return an empty list
- set res = a list to store the start indices of the anagrams
- set count = a hashmap to store the required character frequencies in p
    - *key = character*
    - *value = frequency of character in p*
    - *either loop to fill the hashmap or use Counter()*
- set window = a hashmap to store the character frequencies in the current window
    - *key = character*
    - *value = frequency of character in the current window*
- set left pointer l = 0 to track the left side of the window
- use the right pointer r = 0 to traverse s and track the right side of the window
- traverse s using the right pointer
    - *expand the window by adding the current element*
        - increment the current element count in the window hashmap
        - *move the right pointer, which will auto-update in the for-loop*
    - if the window becomes larger than p
        - *slide the window by moving the left pointer*
        - decrement the count of the left element in the window
        - move the left pointer
    - if the window is valid (has the same character frequencies as p)
        - add the left pointer to res as the start index of the anagram
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(counting frequencies in string p + string s traversal) → O(m + n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(unique characters) → O(26) → O(1)
    - *26 because s and p consist of lowercase English letters*


## Code

```python
class Solution:

    def findAnagrams(self, s: str, p: str) -> List[int]:
    if len(p) > len(s):
        return []
    res = []
    count = Counter(p)
    window = Counter()
    l = 0

    for r in range(len(s)):
        # expand the window
        window[s[r]] += 1

        # slide the window
        if r - l + 1 > len(p):
            window[s[l]] -= 1
            l += 1

        # check if window is valid
        if window == count:
            res.append(l)

    return res
    
```
