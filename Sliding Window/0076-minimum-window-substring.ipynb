# <center> 76. Minimum Window Substring </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-window-substring/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the minimum-size window (substring) in string s that contains all characters of string t.

Use a variable-size sliding window:
- expand the window by adding elements until it becomes valid
- when the window is valid, shrink the window by removing elements from the left to find a smaller valid window
- keep track of the smallest valid window
    - use a hashmap for required character counts and a counter to check if the window is valid

*Note: in a variable-size window, expand or shrink the window based on the problem's validity condition. Here, expand the window until it becomes valid, then shrink while it remains valid to find a smaller valid window.*


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if the length of t is greater than s, there's no way we can find a substring
        - return an empty string
- set need = a hashmap to track the required count of each char i.e characters in t 
    - *key = character*
    - *value = required count*
    - *either loop or use counter() to fill the hashmap*
- set have = 0 to track the number of required chars currently in the window
- set start = -1, end = len(s) to track the boundaries of the minimum valid window
    - *invalid indices are used initially to detect if no valid window is found*
    - *these pointers avoid creating a substring each time we find a valid window*
- set left pointer l = 0 and right pointer r = 0 to track the window left and right sides
- loop while r < len(s) to traverse the string
    - *expand the window by adding the current element*
        - if current char count is greater than 0, it is a required character
            - increment have because we have found a required character
        - decrement its count in the hashmap 
            - *decrement because a char is added to the window*
        - move right pointer
    - while we have all the required characters, the current window is valid, search for a smaller window
        - if current window is smaller than previous
            - update start and end
        - *shrink the window by removing the left element*
            - increment the count of left character in the hashmap
                - *increment because a char is removed from the window*
            - if the left character count is greater than 0, it is a required character
                - *decrement have because a required char is removed*
            - move left pointer
- if start is still -1, return an empty string because we have not found the substring, else return the substring from start to end


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string t frequency count + string s traversal + window shrinking) → O(m + n + n) → O(m + n)
    - *m = string t size*
    - *n = string s size*
    - *each character is added to and removed from the window at most once*
    - *the costs are added because the inner while loop runs n in total, not n for each outer iteration*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(unique characters in t) → (lowercase and uppercase English letters) → O(52) → O(1)


## Code

```python
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        if len(t) > len(s):
            return ""
        need = Counter(t)
        have = 0
        start, end = -1, len(s)
        l, r = 0, 0
        while r < len(s):
            # expand the window
            if need[s[r]] > 0:
                have += 1
            need[s[r]] -= 1
            r += 1
            
            while have == len(t):
                if r - l < end - start:
                    start = l
                    end = r

                # shrink the window
                need[s[l]] += 1
                if need[s[l]] > 0:
                    have -= 1
                l += 1
                
        return '' if start == -1 else s[start:end]
        
```
