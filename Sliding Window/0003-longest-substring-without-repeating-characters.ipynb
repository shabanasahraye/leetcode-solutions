# <center> 3. Longest Substring Without Repeating Characters </center>


## Problem Description
[Click here](https://leetcode.com/problems/longest-substring-without-repeating-characters/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the longest window (substring) without repeating characters.

Use a variable-size sliding window:
- expand the window by adding elements
- when the window becomes invalid (contains a duplicate), shrink the window by removing elements from the left until it becomes valid
- keep track of the largest valid window
    - use a hash set to check if the window is valid (has no duplicate)

*Note: in a variable-size window, expand or shrink the window based on the problem's validity condition. Here, expand the window until it becomes invalid, then shrink until it becomes valid.*


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if string size is 0 or 1, the longest substring without duplicates is the string itself
        - return string size
- set res = 0 to track the largest valid window
- set seen = a hash set to check for duplicates in the current window
- set left pointer l = 0 to track the left side of the window
- use the right pointer r = 0 to traverse the string and track the right side of the window
- traverse the string using right pointer
    - while the window is invalid (current element is already in the seen)
        - *shrink the window*
            - removing the left element from seen
            - move the left pointer
    - *expand the window* 
        - add the current element to seen
        - *move the right pointer, which will auto-update in the for-loop*
    - update res with the largest valid window (max of the current window size and previous maximum)
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hash set) → O(unique characters) → O(256) → O(1)


## Code

```python
class Solution:

    def lengthOfLongestSubstring(self, s: str) -> int:
        res = 0
        seen = set()
        l = 0
        for r in range(len(s)):
            while s[r] in seen:
                # shrink the window
                seen.remove(s[l])
                l += 1
            # expand the window
            seen.add(s[r])
            # update result
            res = max(res, r - l + 1)
        return res
```
