# <center> 209. Minimum Size Subarray Sum </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-size-subarray-sum/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the smallest window (subarray) whose sum is >= target

Use a variable-size sliding window:
- expand the window by adding elements until it becomes valid
- when the window is valid, shrink the window by removing elements from the left to find a smaller valid window
- keep track of the smallest valid window
    - use a running sum to check if the window is valid

*Note: in a variable-size window, expand or shrink the window based on the problem's validity condition. Here, expand the window until it becomes valid, then shrink while it remains valid to find a smaller valid window.*


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = a large number to track the smallest window size
- set window_sum = 0 to track the current window sum
- set left pointer l = 0 and right pointer r = 0 to track the window left and right sides
- loop while r < len(nums) to traverse the array
    - *expand the window by adding the current element*
        - add the current element to window_sum
        - move right pointer
    - while the current window is valid
        - update res by taking min of current window size and the previous minimum
        - *shrink the window by removing the left element*
            - subtract left element from window_sum
            - move left pointer
- return res if a valid window was found, else 0


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal + window shrinking) → O(n + n) → O(n)
    - *the costs are added because the inner loop runs n in total, not n for each outer iteration*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        res = float('inf')
        window_sum = 0
        l, r = 0, 0
        while r < len(nums):
            # expand the window
            window_sum += nums[r]
            r += 1

            while window_sum >= target:
                # update result
                res = min(res, r - l)

                # shrink the window
                window_sum -= nums[l]
                l += 1

        return res if res != float('inf') else 0
        
```
