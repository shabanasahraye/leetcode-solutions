# <center> 1423. Maximum Points You Can Obtain from Cards </center>


## Problem Description
[Click here](https://leetcode.com/problems/maximum-points-you-can-obtain-from-cards/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to select k cards from either the left or right side with the maximum possible points.

This is a fixed-size sliding window problem because every selection must contain exactly k cards.

The possible selections are:
- start with the first k cards and replace cards from the left with cards from the right
- start with the last k cards and replace cards from the right with cards from the left

Start with the last k cards, then slide the window by moving the left and right pointers:
- add the next card from the left
- remove the current card from the right
- keep track of the maximum points

*Note: in a fixed-size window, we slide the window by moving the left and right pointers while keeping the window size fixed.*


## Approach
- check base case
    - if k equals the total number of cards, we have to choose all cards
        - return the sum of all cards
- set left pointer l = 0 to track the card added from the left
- set right pointer r = n - k to track the card removed from the right
- set total = sum of the last k cards to track the current points
- set res = total to track the maximum points found so far
- traverse the array starting from r
    - slide the window by adding a card from the left and removing a card from the right
        - *total = total + left card - right card*
    - update res with the max of the current points and previous max points
    - move the left and right pointers
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sum of last k cards + loop) → O(k + k) → O(k)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def maxScore(self, cardPoints: List[int], k: int) -> int:
        if k == len(cardPoints):
            return sum(cardPoints)
        l = 0
        r = len(cardPoints) - k
        total = sum(cardPoints[i] for i in range(r, len(cardPoints)))
        res = total

        for _ in range(k):
            # slide the window
            total += cardPoints[l] - cardPoints[r]
            res = max(res, total)
            l += 1
            r += 1

        return res
```
