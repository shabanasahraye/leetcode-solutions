# <center> 199. Binary Tree Right Side View </center>


## Problem Description
[Click here](https://leetcode.com/problems/binary-tree-right-side-view/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
The right view of the binary tree includes the last node of each level.

Do Level Order Traversal (BFS) and get the last node in each level. 


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if the tree is empty
        - return an empty array
- set res = an empty array to store the right view i.e last nodes of each level
- create a queue for BFS and add root node to the q
- loop until q is empty i.e until the end of the tree
    - add the last node of the current level to the result array
        - *at the start of each loop, the queue contains the current level, so the last node is the rightmost node*
    - loop to traverse current level nodes 
        - pop q to get current level node
        - if the node has a left or right child
            - push the child into the q to get the next level nodes
- return the result array


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(tree BFS) → O(total nodes visited) → O(n)
    - *each node is inserted into and removed from the queue exactly once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(queue + result array) → O(max width + total levels) → O(w + h) → O(n + n) → O(n)
    - *the result array stores one node for each level, so its size is the tree height*
    - *tree height (h) and width (w) depend on the tree structure*
        - *balanced tree: last level has the max nodes → w = $2^h$ = $2^{logn}$ = n*
        - *skewed tree: each level has 1 node → w = 1*


## Code

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right


class Solution:

    def rightSideView(self, root: Optional[TreeNode]) -> List[int]:
        if not root:
            return []
        res = []
        q = deque([root])
        while q:
            res.append(q[-1].val)
            for _ in range(len(q)):
                cur = q.popleft()
                if cur.left:
                    q.append(cur.left)
                if cur.right:
                    q.append(cur.right)
        return res
        
```
