## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Traverse the tree (DFS or BFS) and swap the left and right children at each node.


## Approach
<!-- Describe your approach to solving the problem. -->
### Recursive Approach
- check base cases
    - if there's no node in the tree
        - return null
- swap the left and right subtrees
- call invert for left subtree
- call invert for right subtree
- the recursion continues until we reach a null node
- return root

### Iterative Approach
- check base cases
    - if there's no node in the tree
        - return null
- create a queue and add the root node
- loop until the queue is empty
    - pop a node from queue and store it in a variable cur as the current node
    - swap the left and right children
    - if the current node has left child
        - add it to queue 
    - if the current node has right child
        - add it to queue 
- return root


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Recursive Approach: O(tree DFS) → O(total nodes visited) → O(n)
        - *each node is visited exactly once during DFS*
    - Iterative Approach: O(tree BFS) → O(total nodes visited) → O(n)
      - *each node is added to and removed from the queue exactly once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Recursive Approach: O(recursion stack) → O(h) → O(n)
        - *tree height (h) depends on the tree structure, balanced tree height is logn and skewed tree height is n*
    - Iterative Approach: O(queue) → O(max width) → O(w) → O(n)
        - *tree width (w) depends on the tree structure*
            - *balanced tree: last level has the max nodes → w = $2^h$ = $2^{logn}$ = n*
            - *skewed tree: each level has 1 node → w = 1*
