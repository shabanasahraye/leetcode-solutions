# <center> 662. Maximum Width of Binary Tree </center>


## Problem Description
[Click here](https://leetcode.com/problems/maximum-width-of-binary-tree/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Do level order traversal using BFS, find the width of each level, and return the max width.

Formula to find width of a level <br>
= end nodes position difference + 1 <br>
= rightmost node position - leftmost node position + 1 <br>
= 2 * parentnode position - 2 * parentnode position + 1 (using complete binary tree indexing)


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 1 to track the max width
    - *1 because the tree has at least one node*
- create a queue for BFS and push the root node and its position as a pair into q
    - *index 0 = node*
    - *index 1 = node position*
- loop until q is empty i.e until the end of the tree
    - store the position of the first node in the current level
    - loop in range q size to get current level nodes
        - pop q to get the current node and its position
        - update the position of the last node
        - add the left child with position `2 × pos` into q
        - add the right child with position `2 × pos + 1` into q
    - compute current level width and update the res
        - *width = last position − first position + 1*
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(tree BFS) → O(total nodes) → O(n)
    - *each node is inserted into and removed from the queue exactly once during BFS*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(queue) → O(max width) → O(w) → O(n)
    - *tree height (h) and width (w) depend on the tree structure*
        - *balanced tree: last level has the max nodes → w = $2^h$ = $2^{logn}$ = n*
        - *skewed tree: each level has 1 node → w = 1*


## Code

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right


class Solution:

    def widthOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        res = 1
        q = deque([(root, 1)])
        while q:
            first = q[0][1]
            last = first
            for _ in range(len(q)):
                node, pos = q.popleft()
                last = pos
                if node.left:
                    q.append((node.left, 2 * pos))
                if node.right:
                    q.append((node.right, 2 * pos + 1))
            res = max(res, last - first + 1)
        return res
        
```
