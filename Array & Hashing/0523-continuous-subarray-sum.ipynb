## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
A good subarray:
- has at least 2 elements
- its sum is multiple of k i.e divisible by k 

We need to find a subarray whose sum has a specific remainder i.e 0 when divided by k.

We can represent a subarray using prefix sum. A prefix sum represents a subarray from the start of the array up to the current index.

An array has a good subarray if two prefix sums have the same remainder when divided by k, and the subarray between them has at least two elements. Their difference is divisible by k, so the subarray sum is divisible by k. For example:
1. arr = [2, 2, 1, 2, 3], k = 6  
    - prefix sum = [0+2, 2+2, 2+2+1, 2+2+1+2, 2+2+1+2+3] = [2, 4, 5, 7, 10]  
    - prefix sum % k = [2/6, 4/6, 5/6, 7/6, 10/6] = [2, **4**, 5, 1, **4**]
    - the array has a good subarray because:
        - there are two prefix sums with the same remainder (4), one ending at index 1 and the other at index 4
        - the subarray between these indices (inclusive) contains 3 elements: [1, 2, 3]
        - *cross-check: (1 + 2 + 3) / 6 = 1, the subarray sum is divisible by k*

2. arr = [23, 2, 4, 6, 6], k = 7  
    - prefix sum = [23, 25, 29, 35, 41]  
    - prefix sum % k = [2, 4, 1, **0**, 6]
    - the array has a good subarray because:
        - there are two prefix sums with the same remainder (0): the empty prefix at index -1 and the prefix ending at index 3
        - the subarray between these indices contains 4 elements: [23, 2, 4, 6]
        - *cross-check: (23 + 2 + 4 + 6) / 7 = 5, the subarray sum is divisible by k*

3. arr = [23, 2, 6, 4, 7], k = 13  
    - prefix sum = [23, 25, 31, 35, 42]  
    - prefix sum % k = [10, 12, 5, 9, 3]
    - the array doesn't have a good subarray because there are no duplicate remainders

Use prefix sums to find subarrays and a hashmap to store each remainder and its earliest index for fast searching.


## Approach
<!-- Describe your approach to solving the problem. -->
- set remainders = a hashmap to map each remainder to its earliest index
    - *key = subarray (prefix sum) remainder*
    - *value = earliest index where the remainder occurs*
    - *we need the index to check the number of elements in the subarray*
- add remainder 0 with index -1
    - *this represents the empty subarray (prefix) before the array starts*
- set pre_sum = 0 to track the prefix sum
- traverse the array
    - add the current number to prefix sum
    - calculate the remainder of current subarray (prefix sum) when divided by k
    - if the remainder is in the hashmap, there is another subarray with same remainder
        - if there are at least two elements between the current and previous indices, (current index - previous index >= 2), there is a good subarray
            - return true
        - else continue
    - else add the remainder and its index to the hashmap
- if the loop completes without returning true, there's no good subarray, return false


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(number of remainders) → O(n)
    - *in the worst case, a different remainder is added in each iteration, so the hashmap contains at most n remainders*


## Code

```python
class Solution:

    def checkSubarraySum(self, nums: List[int], k: int) -> bool:
        remainders = {0: -1}
        pre_sum = 0
        for i, n in enumerate(nums):
            pre_sum += n
            mod = pre_sum % k
            if mod in remainders:
                if i - remainders[mod] >= 2:
                    return True
                continue
            remainders[mod] = i
        return False
        
```
