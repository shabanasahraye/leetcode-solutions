## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
- **Approach 1: Sorting**
    - sort the strings and find compare the two most dissimilar strings (first and last) to find th common prefix
    - time complexity = O(sorting strings array) → O(n × s × logn)

- **Approach 2: Scanning**
    - compare the first string with the remaining strings character by character to find the common prefix
    - time complexity = O(m × n)


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: Scanning
- check base cases
    - if there is only one string in the array
        - return it
    - if there's an empty string in the array, there can't be a common prefix
        - return an empty string
- traverse the first string <br>
*for i in range 0 to the length of the first string*
    - traverse the remaining strings starting from the se <br>
    *for j in range 1 to the length of the strings array*
        - if current string is shorter than the first string i.e i is equal to the size of current string or the current char in the first string != the current char in the current string
            - return the substring of the first string up to index i as the longest common prefix
- if the loop completes without returning anything, the first string is the longest common prefix, return it


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
  - Scanning Approach: O(empty string check + comparing strings) → O(n + first string × string array) → O(n + m × n) → O(m × n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
  - Scanning Approach: O(substring returned) → O(m)


## Code

```python
class Solution:

    def longestCommonPrefix(self, strs: List[str]) -> str:

        # Approach 2: Scanning
        if len(strs) == 1:
            return strs[0]
        if '' in strs:
            return ''
        for i in range(len(strs[0])):
            for j in range(1, len(strs)):
                if i == len(strs[j]) or strs[0][i] != strs[j][i]:
                    return strs[0][:i]
        return strs[0]
        
```
