# <center> 1461. Check If a String Contains All Binary Codes of Size K </center>


## Problem Description
[Click here](https://leetcode.com/problems/check-if-a-string-contains-all-binary-codes-of-size-k/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Check if the number of unique k-size substrings in the given binary string equals the total number of possible binary codes of size k.

Total possible binary codes = $2^k$


## Approach
<!-- Describe your approach to solving the problem. -->
- set m = total number of possible binary codes i.e $2^k$ 
- if the size of the string is less than m, return false
- set codes = a hashset to store the unique substrings (binary codes)
- traverse the string <br>
*loop until size of string - k + 1 to keep index in bounds*
    - get a k-size substring and add it to the set
- check if the number of substrings in the set is equal to the required substrings (m)


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop × substring creation) → O((n - k + 1) × k) → O(n × k)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashset) → O(total unique substrings × substring size) → O($2^k$ × k) → O(n × k)
    - *to contain all $2^k$ k-size binary codes, the string must have at least $2^k$ k-size substrings, so ${n - k + 1 ≥ 2^k}$ → ${n ≥ 2^k + k - 1}$* 


## Code

```python
class Solution:

    def hasAllCodes(self, s: str, k: int) -> bool:
        m = 1 << k       # 1 << k = 2**k
        if len(s) < m:
            return False
        codes = set()
        for i in range(len(s) - k + 1):
            codes.add(s[i: i + k])
        return len(codes) == m  
        
```
