# <center> 2017. Grid Game </center>


## Problem Description
[Click here](https://leetcode.com/problems/grid-game/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Robot 1 can take n possible paths, depending on where it moves down.

Robot 2 can move right on the top row and then down, or move down first and then right on the bottom row. It collects top-row points after Robot 1 moves down or bottom-row points before that column.

Robot 2 wants to maximize its points
- so for each possible down position, take the maximum of these two sums (top and bottom row sums)

Robot 1 wants to minimize Robot 2's points
- so take the minimum of these maximum values


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = a very large number to track the minimum points Robot 2 can get
- set bottom = 0 to track the bottom-row sum before the current column
- set top =  top row sum to track the top-row sum after the current column
- traverse the values in the first and second rows in parallel <br>
*the grid has two rows of size n, either loop in range 0 to n or use zip() for parallel traversal*
    - decrement top by the current top-row value
        - *this removes the current top-row value, leaving the top-row points after the current column*
    - update res
        - *take the max of bottom and top because Robot 2 chooses the row with more points*
        - *take the min of this result and the previous res because Robot 1 chooses the down position that minimizes Robot 2's points*
    - increment the bottom sum
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(min(row1 size, row2 size)) → O(min(n, n)) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def gridGame(self, grid: List[List[int]]) -> int:
        res = math.inf
        bottom, top = 0, sum(grid[0])
        for a, b in zip(grid[0], grid[1]):
            top -= a
            res = min(res, max(top, bottom))
            bottom += b
        return res
        
```
