# <center> 290. Word Pattern </center>


## Problem Description
[Click here](https://leetcode.com/problems/word-pattern/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
It is the same as [Isomrphic Strings](https://leetcode.com/problems/isomorphic-strings/description/).

Use two hashmaps to track the 1-1 mapping between the characters in pattern and words in s.


## Approach
<!-- Describe your approach to solving the problem. -->
- split string s to get the list of words
- check base case
    - if the number of characters in pattern and the number of words in s are not the same, there is no mapping
        - return false
- set char_word = a hashmap to map characters in pattern to words in s
- set word_char = a hashmap to map words in s to characters in pattern
- traverse the characters and words in parallel <br>
*either loop in range 0 to the pattern size because both characters and words are of equal size or use zip()*
    - *for each character c and word w in pattern and words*
        - check the mapping i.e if the current char in pattern is not mapped to the current word in s or vice versa
            - return false
        - else map the char to word in char_word hashmap and vice versa
- if the loop completes without returning false, return true


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(split + traversal) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(words list + hashmap char_word + hashmap word_char) → O(n + 26 + 26) → O(26)
    - *pattern contains lowercase English letters, so char_word can have at most 26 keys*
    - *if every word is different, word_char can have up to n keys*


## Code

```python
class Solution:

    def wordPattern(self, pattern: str, s: str) -> bool:
        words = s.split()
        if len(pattern) != len(words):
            return False
        char_word, word_char = {}, {}
        for c, w in zip(pattern, words):
            if (c in char_word and char_word[c] != w) or (w in word_char and word_char[w] != c):
               return False
            char_word[c] = w
            word_char[w] = c
        return True
        
```
