# <center> 442. Find All Duplicates in an Array </center>


## Problem Description
[Click here](https://leetcode.com/problems/find-all-duplicates-in-an-array/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We have to find the duplicate numbers in the range 1 to n in constant space, so we can't use a hashset.

It is similar to the [First Missing Positive](https://leetcode.com/problems/first-missing-positive/description/).

The numbers are in the range 1 to n, while the indices are 0 to n - 1. We can use each number as an index by mapping number x to index x - 1

Traverse the array and mark the index corresponding to each number by making the value at that index negative. If the value is already negative, that index has been visited before, so the current number is a duplicate.



## Approach
<!-- Describe your approach to solving the problem. -->
- set res = a list to store duplicate numbers
- traverse the input list
    - find the index corresponding to the current number
        - *index = absolute value of current number - 1*
    - if the index is already marked i.e the number at that index is negative, the current number is a duplicate
        - add it to the result list
            - *add its absolute value because we may have made it negative for marking*
    - else mark the index by making the number at the index negative
- return the result list


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result list) → O(n) → but the problem states that the result list does not count as extra space → O(1) 


## Code

```python
class Solution:

    def findDuplicates(self, nums: List[int]) -> List[int]:
        res = []
        for n in nums:
            i = abs(n) - 1
            if nums[i] < 0:
                res.append(abs(n))
            nums[i] *= - 1
        return res
        
```
