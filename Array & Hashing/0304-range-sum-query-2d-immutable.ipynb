## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the sum of elements inside a rectangular region of a matrix for multiple queries.

The sum of the elements in a rectangular region is an overlap of four rectangles that originate at matrix[0][0]
- **rectangle sum = whole sum - top rectangle sum - left rectangle sum + top-left rectangle sum**
    - sum(r1, c1, r2, c2) = sum(r2, c2) - sum(r1 - 1, c2) - sum(r2, c1 - 1) + sum(r1 - 1, c1 - 1)
    - *the top-left rectangle is the overlap of the top and left rectangles, it is added back because its sum is subtracted twice due to overlap*

Use a 2D prefix sum to get rectangle sum
- **prefix sum of current cell = current cell + top prefix sum + left prefix sum - top-left prefix sum**
    - prefix_sum[r][c] = matrix[r][c] + prefix_sum[r-1][c] + prefix_sum[r][c-1] - prefix_sum[r-1][c-1]
    - *the top-left prefix sum represents the overlap of the top and left regions, it is subtracted because its sum is added twice due to overlap*


## Approach
<!-- Describe your approach to solving the problem. -->
**init()**
- set m = total rows, n = total columns
- initialize a 2D prefix sum array of size m x n with zeroes
- fill the prefix sum array
    - for each row r
        - for each column c
            - get top cell prefix sum if the current row is not the first row else the sum is 0
                - *top cell indices = previous row, same col = current row - 1, current col*
            - get left cell prefix sum if the current col is not the first col else the sum is 0
                - *left cell indices = same row, previous col = current row, current col - 1*
            - get top-left cell prefix sum if current row and col is not the first row and col else the sum is 0
                - *top-left cell indices = previous row, previous col = current row - 1, current col - 1*
            - get current cell prefix sum using the formula

**sumRegion()**
- get top rectangle sum using the prefix sum array if the rectangle top row is not the first row else the sum is 0
    - *top rectangle indices = previous row, last col = row1 - 1, col2*
- get left rectangle sum if the rectangle left column is not the first col else the sum is 0
    - *left rectangle indices = last row, previous col = row2, col1 - 1*
- get top-left rectangle sum if row1 and col1 are not the first row and first col else the sum is 0
    - *top-left rectangle indices = previous row, previous col = row1 - 1, col1 - 1*
- get current rectangle sum using the formula


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - init(): O(creating prefix sum array + filling prefix sum array) → O(rows × columns + row × columns) → O(m × n + m × n) → O(m × n) 
        - *but it will be done only once i.e when an object is created*
    - sumRange(): O(1)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - init(): O(prefix sum array) → O(rows × columns) → O(m × n)
    - sumRange(): O(1)


## Code

```python
class NumMatrix:

    def __init__(self, matrix: List[List[int]]):
        m, n = len(matrix), len(matrix[0])
        self.pre_sum = [[0] * n for _ in range(m)]
        for r in range(m):
            for c in range(n):
                top = self.pre_sum[r - 1][c] if r else 0
                left = self.pre_sum[r][c - 1] if c else 0
                top_left = self.pre_sum[r - 1][c - 1] if r and c else 0
                self.pre_sum[r][c] = matrix[r][c] + top + left - top_left

    def sumRegion(self, row1: int, col1: int, row2: int, col2: int) -> int:
        top = self.pre_sum[row1 - 1][col2] if row1 else 0
        left = self.pre_sum[row2][col1 - 1] if col1 else 0
        top_left = self.pre_sum[row1 - 1][col1 - 1] if row1 and col1 else 0
        return self.pre_sum[row2][col2] - top - left + top_left
        


# Your NumMatrix object will be instantiated and called as such:
# obj = NumMatrix(matrix)
# param_1 = obj.sumRegion(row1,col1,row2,col2)
```
