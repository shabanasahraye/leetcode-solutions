## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
- **Approach 1: Sorting - O(nlogn)**
    - the majority element will always be in the middle of the sorted list

- **Approach 2: Hashmap - O(n)**
    - count the occurrences of elements using a hashmap
    - return the element with count > n / 2

- **Approach 3: Moore’s Voting Algorithm - O(n) time O(1) space**
    - according to the algo, if there is a majority element in an array, it will always remain in the lead i.e its votes (count) will never become 0
    - select the first element as the candidate
    - traverse the array
        - if the current element is the same as the chosen one, increase the votes (count), else decrease the votes
        - if votes become 0, select the next element as the candidate


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 3: Moore’s Voting Algorithm
- set res = array first element as the candidate to track the majority element
- set count = 1 to track the candidate's vote count
- traverse the array from the second element
    - if count is 0
        - select the current element as the new candidate
    - if the current element is the same as res (majority element)
        - increment the count else decrement the count
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Voting Algorithm Approach: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Voting Algorithm Approach: O(1)


## Code

```python
class Solution:

    def majorityElement(self, nums: List[int]) -> int:

        # Approach 3: Moore’s Voting Algorithm
        res = nums[0]
        count = 1
        for i in range(1, len(nums)):
            if count == 0:
                res = nums[i]
            count += 1 if nums[i] == res else -1
        return res
```
