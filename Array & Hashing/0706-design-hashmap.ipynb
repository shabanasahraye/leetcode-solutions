## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Use an array of buckets with linked lists because different keys can have the same hash value, causing collisions.


## Approach
<!-- Describe your approach to solving the problem. -->
*create a separate class for the linked list nodes*

#### List Node Class

**init()**
- set the node key, value, and next pointer

#### Hashmap Class

**init()**
- set hashmap capacity = 1000
    - *1000 is the number of buckets we choose for the hashmap*
    - *we can use a smaller capacity, but it will result in more collisions and longer linked lists*
- initialize an array of size 1000 with linked list nodes to represent the buckets of the hashmap

**hash()**
- take the mod of key to get the index and return the result 

**put()**
- find the key index and set current node pointer cur = the first node at the index
- if some nodes are already there due to collision, loop until the end of the list of nodes
    - if key node is already added, update its value and return
- else create a new node for the key and add its value

**get()**
- find index and set cur = the first node at the index
- loop until you find the key
- if the key is found, return its value
- else return -1

**remove()**
- find index and set cur = the first node at the index
- loop until you find the node before the key node
- if you found the previous node
    - remove the key node by updating the next pointer of the previous node


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - init(): O(array initialization) → O(1000) → O(1)
    - hash(): O(1)
    - put(), get(), and remove(): O(n) in the worst case and O(1) on average
        - *in the worst case, all keys can have the same hash value, resulting in a linked list of size n*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - init(): O(array + linked list) → O(1000 + n) → O(n)
    - hash(): O(1)
    - put(): O(1)
    - get(): O(1)
    - remove(): O(1)
    - overall space: O(total nodes) → O(n)
        - *after n calls to put(), the linked list can contain up to n nodes*


## Code

```python
class ListNode:
    
    def __init__(self, key= -1, val= -1, next= None):
        self.key = key
        self.val = val
        self.next = next
    

class MyHashMap:
    
    def __init__(self):
        self.capacity = 1000
        self.buckets = [ListNode() for _ in range(self.capacity)]

    def hash(self, key: int) -> int:
        return key % self.capacity

    def put(self, key: int, val: int) -> None:
        cur = self.buckets[self.hash(key)]
        while cur.next:
            if cur.next.key == key:
                cur.next.val = val
                return
            cur = cur.next
        cur.next = ListNode(key, val)

    def get(self, key: int) -> int:
        cur = self.buckets[self.hash(key)]
        while cur and cur.key != key:
            cur = cur.next
        if cur:
            return cur.val
        return -1
        
    def remove(self, key: int) -> None:
        cur = self.buckets[self.hash(key)]
        while cur.next and cur.next.key != key:
            cur = cur.next
        if cur and cur.next:
            cur.next = cur.next.next
        


# Your MyHashMap object will be instantiated and called as such:
# obj = MyHashMap()
# obj.put(key,value)
# param_2 = obj.get(key)
# obj.remove(key)
```
