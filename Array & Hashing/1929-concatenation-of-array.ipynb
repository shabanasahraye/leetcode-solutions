## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
There are different approaches to concatenate the array:
- using operators:
    - concatenate using +
    - repeat using *
- using array methods:
    - extend()
    - append()


## Approach
<!-- Describe your approach to solving the problem. -->
**Using + operator**
- concatenate by using the + operator and return the result

**Using * operator**
- use array * k to create a k × n array
*k is the number of times you want to repeat the array, here k is 2* 

**Using extend()**
- use extend() and add the input array to the end of itself
- return the input array

**Using append()**
- traverse the input array
    - add the current number in the input array to the end of the array
- return the input array


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Using + operator: O(n)
    - Using * operator: O(2n) → O(n)
    - Using extend(): O(n)
    - Using append(): O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Using + operator: O(n + n) → O(n)
    - Using * operator: O(2n) → O(n)
    - Using extend(): O(n)
    - Using append(): O(n)

```python
class Solution:
    
    def getConcatenation(self, nums: List[int]) -> List[int]:
        
        # Using + operator
        # return nums + nums

        # Using * operator
        # return nums * 2

        # Using extend()
        # nums.extend(nums)
        # return nums

        # Using append()
        for i in range(len(nums)):
            nums.append(nums[i])
        return nums
        
```
