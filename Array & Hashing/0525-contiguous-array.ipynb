# <center> 525. Contiguous Array </center>


## Problem Description
[Click here](https://leetcode.com/problems/contiguous-array/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the longest subarray with an equal number of 0s and 1s. 

A subarray with an equal number of 0s and 1s has a net sum of 0. 
- for example: [0, 1, 0, 1]
- 0s = 2, 1s = 2 → net sum = 2 - 2 = 0

So we need to find the longest subarray with a target sum of 0.

General approach to find a subarray with a target sum:
- use a prefix sum and hashmap to find subarrays efficiently
    - *a prefix sum represents a sum from the start of the array up to the current index*
- in each iteration, search for a prefix sum = current prefix sum - target sum
    - *in this case, the target sum is 0, so we look for a previous prefix sum equal to the current prefix sum*
        - *previous prefix sum = current prefix sum − 0 = current prefix sum*
- if such a prefix sum exists, the subarray between the two prefix sums has the target sum
    - *because current prefix sum - previous prefix sum = target sum*


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the longest subarray length
- set net_sum = 0 to to track the current prefix sum
- create index_map = a hashmap to store the first index of each net sum
    - *key = net sum*
    - *value = first index where the net sum occurs*
    - *add net sum 0 with index -1 to represent the empty prefix before the array starts*
- traverse the input array <br>
*for each index and number at that index*
    - increment net_sum if current number is 1 else decrement it
    - if net_sum is in the hashmap, there is a subarray with net sum 0,= i.e the subarray between the first occurrence and the current index has equal numbers of 0s and 1s
        - update res with max of current subarray length and the previous max
            - *subarray with net sum 0 length = distance between the current index and the first occurrence of the current net sum*
    - else
        - add the current subarray net_sum and its index to the map
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(n)


## Code


```python
class Solution:

    def findMaxLength(self, nums: List[int]) -> int:
        res = 0
        net_sum = 0
        index_map = {0: -1}
        for i, num in enumerate(nums):
            net_sum += 1 if num == 1 else -1
            if net_sum in index_map:
                res = max(res, i - index_map[net_sum])
            else:
                index_map[net_sum] = i
        return res
          
```
