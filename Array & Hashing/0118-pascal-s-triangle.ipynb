# <center> 118. Pascal's Triangle </center>


## Problem Description
[Click here](https://leetcode.com/problems/pascals-triangle/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Each number arr[i][j] in Pascal's triangle is the sum of the two numbers directly above it: 
- the previous row's previous column arr[i-1][j-1]
- the previous row's same column arr[i-1][j]
- arr[i][j] = arr[i-1][j-1] + arr[i-1][j]


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = a 2d array with default value 1 to represent the Pascal triangle
    - *rows in range 1 to total rows* 
        - *1 because starting with 0 would result in an empty first row*
    - *cols in range 1 to the current row number* 
        - *row 1 has 1 column, row 2 has 2 columns, row 3 has 3 columns etc*
- loop to update the 2d array
    - for each row excluding the first two rows because they contain only 1s
        - for each column excluding the first and last columns because they contain 1s
            - update the current cell value using the formula
- return the 2d array (res)


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array creation + array update) → O(row × columns + row × columns) → O(n² + n²) → O(n²)
    - *n² because each row has the same number of columns as its row number. For example, row 1 has 1 column, row 2 has 2 columns etc* <br>
    *total iterations are: 1 + 2 + 3 + ... + n = n(n + 1)/2 = O(n²)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result list) → O(total rows × max column) → O(n × m) → O(n × n) → O(n²) 
    - *m = n because the last row has the maximum number of columns, and each row has the same number of columns as its row number*


## Code

```python
class Solution:

    def generate(self, num_rows: int) -> List[List[int]]:
        res = [[1 for col in range(row)] for row in range(1, num_rows + 1)]
        for r in range(2, num_rows):
            for c in range(1, r):
                res[r][c] = res[r - 1][c - 1] + res[r - 1][c]
        return res
    
```
