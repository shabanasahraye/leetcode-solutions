# <center> 560. Subarray Sum Equals K </center>


## Problem Description
[Click here](https://leetcode.com/problems/subarray-sum-equals-k/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
This problem is similar to [Contiguous Array](https://leetcode.com/problems/contiguous-array/description/).

We need to find the total number of subarrays with sum k. 

General approach to find subarray with a target sum
- use prefix sum to find subarrays and hashmap to store prefix sum frequencies
    - *a prefix sum represents a subarray from the start of the array up to the current index*
- in each iteration, search for a previous prefix sum = current prefix sum - target sum
    - *in this case, the target sum is k, so we look for a previous prefix sum = current prefix sum − k*
- if such a prefix sum exists, the subarray between the two prefix sums has the target sum
    - *because current prefix sum - previous prefix sum = target sum*


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the number of subarrays with sum k
- set pre_sum = 0 to track the current prefix sum
- set count = a hashmap to store the frequency of each prefix sum
    - *key = prefix sum*
    - *value = prefix sum frequency*
    - *add a prefix sum 0 with freq 1 because there's always an empty subarray that sums to 0*
- traverse the input array
    - add current element to pre_sum
    - if pre_sum - k is in the hashmap, there are count[pre_sum - k] subarrays with sum k
        - add this count to res
    - increment the count of the current prefix sum in the hashmap
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(n)


## Code

```python
class Solution:

    def subarraySum(self, nums: List[int], k: int) -> int:
        res = 0
        pre_sum = 0
        count = defaultdict(int, {0: 1})
        for n in nums:
            pre_sum += n
            if pre_sum - k in count:
                res += count[pre_sum - k]
            count[pre_sum] += 1
        return res
        
```
