# <center> 1930. Unique Length-3 Palindromic Subsequences </center>


## Problem Description
[Click here](https://leetcode.com/problems/unique-length-3-palindromic-subsequences/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
In a length-3 palindrome:
- the first and last characters are the same
- the middle character can be any character


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = a hashset to store all unique length-3 palindromes
- set left_chars = a hashset to track the chars seen on the left
- set right_count = a hashmap to track the frequency of chars on the right
- fill the hashmap with the frequency of each char 
    - *key = char*
    - *value = frequency*
    - *either loop to fill the hashmap or use Counter()*
- traverse the chars in the input string
    - *check the possible palindromic subseq with the current char as the middle char of the subseq*
        - remove the current char from the right_count hashmap by decrementing its count because we want to check the left and right chars for it
        - traverse the left chars
            - if the left char exists on the right with a frequency greater than 0, we found a palindrome
                - add (left char, middle char) to res to represent the palindrome
                    - *we store only (left_char, middle_char) to represent the palindrome, because the right char is always the same as the left char*
    - we will move to the next char, the current char will become leftside char, so add it to the left_chars hashset
- return the length of res as the total number of unique length-3 palindromes
 
 
## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal × hash set traversal) → O(n × 26) → O(n)
  - *26 because input contains lowercase English letters*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result hashset + left hashset + right hashmap) → O(n + 26 + 26) → O(n)
    - *result set stores at most one entry for each unique length-3 palindrome*


## Code

```python
class Solution:

    def countPalindromicSubsequence(self, s: str) -> int:
        res = set()
        left_chars = set()
        right_count = Counter(s)
        for c in s:
            right_count[c] -= 1
            for lc in left_chars:
                if lc in right_count and right_count[lc] > 0:
                    res.add((lc, c))
            left_chars.add(c)
        return len(res)
        
```
