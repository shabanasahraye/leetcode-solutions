# <center> 554. Brick Wall </center>


## Problem Description
[Click here](https://leetcode.com/problems/brick-wall/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
To find a line that crosses the minimum number of bricks, find the position where the most brick edges line up, because a line passing through a brick edge does not cross that brick.

To find the edge position, use prefix sum.
- for example, for row is [1, 2, 2, 1]:
    -  inside edges = [0+1, 1+2, 1+2+2] = [1, 3, 5]
    - *we don't include the last brick because its edge is the right edge of the wall*


## Approach
<!-- Describe your approach to solving the problem. -->
- set maxi = 0 to track the max number of aligned edges
- set edges = a hashmap to count the number of edges at each index
    - *key = edge position*
    - *value = number of edges at that index*
- for each row in the wall
    - set pre_sum = 0 to track the prefix sum and find the edge position
    - for each column (brick) in the row, skip the last brick because its edge is the wall edge
        - update prefix sum by adding the current column (brick)
        - the prefix sum represents the edge position, so increment the edge count at this position
        - update maxi 
- return the difference between the total number of rows (wall length) and the max number of aligned edges (maxi) as the min number of crossed bricks


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(rows × columns) → O(total number of bricks) → O(n)
    - *each row has a different number of columns (bricks), so the nested loop doesn't run n times for every outer iteration*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(total edge positions) → O(m)
  - *m = wall width; edge positions range from 1 to m - 1*


## Code

```python
class Solution:

    def leastBricks(self, wall: List[List[int]]) -> int:
        maxi = 0
        edges = defaultdict(int)
        for row in wall:
            pre_sum = 0
            for col in range(len(row) - 1):
                pre_sum += row[col]
                edges[pre_sum] += 1
                maxi = max(maxi, edges[pre_sum])
        return len(wall) - maxi
        
```
