## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the duplicate and missing number in the range 1 to n.

It is acombination of:
- [442. Find All Duplicates in an Array](https://leetcode.com/problems/find-all-duplicates-in-an-array/description/)
- [448. Find All Numbers Disappeared in an Array](https://leetcode.com/problems/find-all-numbers-disappeared-in-an-array/description/)

The numbers in the array are in the range 1 to n and the indices are 0 to n - 1.

Traverse the array and mark the index of the current number by making the number at that index negative. If you encounter a marked index, the current number is the duplicate.

Traverse the array again and check the indices. If you encounter an unmarked index, the current index + 1 is the missing number.


## Approach
<!-- Describe your approach to solving the problem. -->
- set duplicate = 0 to track the duplicate number
- traverse the array
    - find the index of the current number e.g index of 5 is 4
    - if the index is already marked i.e the number at that index is negative, the current number is a duplicate
        - update duplicate
            - *use the absolute value because we may have made the number negative for marking*
    - else mark the index by making the number at the index negative
- traverse the array again
    - if index is not marked (value is positive), it is the missing number
        - return the duplicate number and missing number (current index + 1) as a list


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(traversal to mark indices + traversal to check indices) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def findErrorNums(self, nums: List[int]) -> List[int]:
        duplicate = 0
        for n in nums:
            i = abs(n)
            if nums[i - 1] < 0:
                duplicate = abs(n)
            else:
                nums[i - 1] *= -1
        for i in range(len(nums)):
            if nums[i] > 0:
                return [duplicate, i + 1]
                
```
