## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Besides insertion, deletion, and random, we also need to search in O(1) time.

Insertion, deletion, and search take O(1) time in set but we can't use a set to find a random element. 

Use an array:
- insertion at the end of an array takes O(1)
- to delete in O(1)
    - swap the value to be deleted with the last value in the array and then remove the last value in O(1)
    - for swapping, we need the value index, use hashmap to map the indices and search in O(1)
- for random value, generate a random index and access the array at that index


## Approach
<!-- Describe your approach to solving the problem. -->
**init()**
- create an array to store values
- create a hashmap to map items/values to indices
    - *key = item*
    - *value = index*

**insert()**
- if the value is already present
    - return false
- get the index of the new value and store it in the hashmap
    - *the index of the new value is the current size of the list*
- add value to the array
- return true because insertion was successful

**remove()**
- if the current value (value to be deleted) is not in the hashmap
    - return false
- get the current value's index from the hashmap and the last value from the array
- replace the current value with the last value to remove the current value
- update the last value's index to the current index
- remove the last value from the array because it is now a duplicate
- remove current value from the hashmap
- return true

**getRandom()**
- generate a random index from 0 to len(vals) - 1 and return the value at that index
    - *either use randint() or choice()*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - init(): O(1) 
    - insert(): O(1)
    - remove(): O(1)
    - getRandom(): O(logn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - init(): O(1) 
    - insert(): O(1)
    - remove(): O(1)
    - getRandom(): O(1)
    - overall: O(array + hashmap) → O(n + n) → O(n)
        - *the array and hashmap can each contain up to n values after n calls to insert()*


## Code

```python
class RandomizedSet:

    def __init__(self):
        self.vals = []
        self.index = {}

    def insert(self, val: int) -> bool:
        if val in self.index:
            return False
        self.index[val] = len(self.vals)
        self.vals.append(val)
        return True

    def remove(self, val: int) -> bool:
        if val not in self.index:
            return False
        cur_index, last_value = self.index[val], self.vals[-1]
        self.vals[cur_index] = last_value
        self.index[last_value] = cur_index
        self.vals.pop()
        del self.index[val]
        return True
        
    def getRandom(self) -> int:
        # return self.vals[randint(0, len(self.vals) - 1)]
        return choice(self.vals)
        


# Your RandomizedSet object will be instantiated and called as such:
# obj = RandomizedSet()
# param_1 = obj.insert(val)
# param_2 = obj.remove(val)
# param_3 = obj.getRandom()
```
