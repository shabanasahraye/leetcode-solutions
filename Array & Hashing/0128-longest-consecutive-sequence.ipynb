# <center> 128. Longest Consecutive Sequence </center>


## Problem Description
[Click here](https://leetcode.com/problems/longest-consecutive-sequence/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the longest sequence of consecutive numbers.

- find the starting point of a sequence
    - a number x is the start of a sequence if x - 1 is not present in the input
- for each starting point, count the consecutive elements to get the sequence length
- return the maximum length

Use a hash set to search in O(1) time.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the length of the longest seq
- convert the nums array into a set
- traverse the set
    - if current number - 1 is not present in the set, the current number is the starting point of a seq
        - set size = 1 to track the length of the current seq
        - search for consecutive numbers in the subsequence and update the size
        - update res with the max of current seq size and previous maximum
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(converting array to set + set traversal × finding consecutive numbers) → O(n + n × amortized O(1)) → O(n + n) → O(n)
    - *while loop take O(n) amortized time across all starting points because each number is processed at most once*
        - *for example, in [100, 4, 200, 1, 3, 2`, it only runs for the starting points 100 and 1, for 1 and 4 iterations, respectively*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(set) → O(n)


## Code

```python
class Solution:

    def longestConsecutive(self, nums: List[int]) -> int:
        res = 0
        nums = set(nums)
        for n in nums:
            if n - 1 not in nums:
                size = 1
                while n + size in nums:
                    size += 1
                res = max(res, size)
        return res
        
```
