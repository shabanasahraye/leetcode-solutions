## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to sort the array without using built-in sorting methods.

- **Approach 1: Bubble Sort**
    - compare → swap → move largest to end → repeat
        - repeatedly compare adjacent elements and swap if in the wrong order
        - larger elements gradually move (bubble) to the end
    - time: O(n²), space: O(1)
    - *stable sorting algorithm because equal elements are not swapped, preserving their input order*

- **Approach 2: Selection Sort**
    - find minimum → swap with first element → move boundary to right → repeat
        - find the smallest element in the unsorted part and swap it with the first element of that part
    - time: O(n²), space: O(1)
    - *unstable sorting algorithm because its swap can change the order of equal elements*

- **Approach 3: Insertion Sort**
    - take element → shift larger elements right → insert → repeat
        - take one element from the unsorted part
        - shift larger elements right to make space
        - insert the element into its correct position
    - time: O(n²), space: O(1)
    - *stable sorting algorithm*

- **Approach 4: Counting Sort**
    - count values → rebuild in order
        - count how many times each value appears and reconstruct the sorted array using those counts
    - useful when the range of values is small
    - time: O(n + k), space: O(k)
    - *unstable sorting algorithm*

- **Approach 5: Merge Sort**
    - uses divide and conquer
    - divide → sort → merge → repeat
        - divide the array into two subarrays
        - recursively sort the subarrays
        - merge the sorted subarrays
    - time: O(nlogn), space: O(n)
    - *stable sorting algorithm*

- **Approach 6: Quick Sort**
    - uses divide and conquer
    - choose pivot → partition → sort left and right → repeat
        - choose an element as the pivot
        - move smaller elements left and larger elements right to place the pivot in its correct position
        - recursively repeat for the left and right parts
        - stop when each part has at most one element
    - time: O(n²), space: O(n)
        - average: O(nlogn) time, O(logn) space
    - *unstable sorting algorithm*

- **Approach 7: Heap Sort**
    - build max heap → move largest to end → restore heap → repeat
    - time: O(nlogn), space: O(1)
    - *unstable sorting algorithm*


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Bubble Sort
- set n = array size
- traverse the array
    - set swapped = false at the beginning of each pass to check whether any elements were swapped
    - traverse the unsorted part of array
        - *n - i - 1 → to compare only the unsorted part because after each pass the largest element is already at the end*
        - if the current element is greater than the next element
            - swap them
            - set swapped = true
    - *after each outer pass, the largest unsorted element moves to the end*
    - if swapped is false, no elements were swapped during the pass, so the array is already sorted
        - break
- return the sorted array

### Approach 2: Selection Sort
- set n = array size
- traverse the array
    - set min_index = current index i to assume the current element is the smallest
    - traverse the remaining unsorted portion
        - if the current element is smaller than the element at min_index
            - update min_index to the current index j
    - swap the current element with the smallest element
    - *after each outer pass, the smallest unsorted element moves to the beginning of the unsorted portion*
- return the sorted array

### Approach 3: Insertion Sort
- traverse the array starting from index 1 (second element because the first is considered sorted)
    - set key = current element to insert
    - set j = i - 1 to point to the last element in the sorted portion
    - while j is within the sorted portion and nusm[j] > key
        - shift the larger element one position to the right
        - move j one position to the left
    - place key into its correct position
- return the sorted array

### Approach 4: Counting Sort
- find the min and max values in the array
- set count = an array of size max_value - min_value + 1 initialized with 0 to store the frequency of every value between the min and max
- traverse the array
    - increment current value count in the count array
        - *use num - mini as the count index to handle negative values and start indexing from 0
            - *mini is always ≤ every num, so num - mini is always 0 or greater*
            - *e.g mini = -3, num = -1 → -1 - (-3) = 2 → count[2]*
- set index = 0 to track the position where the next sorted value should be placed
- traverse the counting array from smallest to largest
    - while the current value still has remaining occurrences
        - place the current value
            - *use i + mini to convert the count index back to the original value*
        - increment index
        - decrease its count
- return the sorted array

### Approach 5: Merge Sort
- set n = size of the array
- check base case
    if n is greater than 1, there are more than 1 elements in the array
    - *divde*
        - find the midpoint 
        - add the elements before the mid index to left subarray and elments from the mid index onward to the right subarray
    - *sort*
        - recursively sort the left and right subarrays
    - *merge*
        - initialize three pointers (i, j, k) for the left, right, and main array
        - traverse both subarrays
            - compare the current elements of the left and right subarrays and place the smaller element into the main array
            - move the corresponding pointer forward
            - move k forward
        - check the remaining elements in the left and right subarray and add them to the main array
- return the sorted array

### Approach 6: Quick Sort
- define a helper function to sort the portion of the array between left and right <br>
*quick_sort(left, right)*
    - if the portion has at most one element, it is already sorted
        - return
    - choose the last element as the pivot
    - set i = left to track where the next smaller element should be placed
        -*i marks the pivot's correct position*
    - traverse the elements before the pivot
        - if the current element is smaller than the pivot
            - swap it with the element at i
            - move i forward
    - swap the pivot with the element at i
        - *the pivot is now in its correct position*
        - *all smaller elements are on the left and all larger elements are on the right*
    - recursively sort the portion before the pivot
        - *i - 1 is the last index on the left*
    - recursively sort the portion after the pivot
        *i + 1 is the first index on the right*
- call quick_sort() on the entire array
- return the sorted array

### Approach 7: Heap Sort
- set n = size of the array
- define a helper function to maintain the max-heap property <br>
*heapify(size, root) → fix the heap of size x, starting from root x*
    - loop until the max-heap property is restored
        - set largest = root
        - calculate the left and right child indices
            - *left child index = 2 × root + 1*
            - *right child index = 2 × root + 2*
        - if the left child exists and is greater than the current largest element
            - update largest
        - if the right child exists and is greater than the current largest element
            - update largest
        - if root is already the largest
            - break
        - else swap and move root to the largest child
- *build the max heap*
    - start from the last non-leaf node and move toward the root
        - call heapify(n, i)
            - *n = full heap size, i = current root being fixed*
- *sort the array*
    - traverse the array in reverse
        - swap the largest element root = 0 with the last unsorted element (nums[i])
        - reduce the heap size and call heapify(i, 0)
            - *i becomes the new heap size because nums[i] is now sorted*
            - *we reduce the heap size because the largest element is now in its final position*
- return the sorted array


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Bubble Sort Approach: O(n passes × n comparisons per pass) → O(n²)
    - Selection Sort Approach: O(n elements × n elements searched per element) → O(n²)
    - Insertion Sort Approach: O(n elements × n shifts/comparisons per element) → O(n²)
    - Counting Sort Approach: O(finding min + finding max + counting + rebuilding array) → O(n + n + k + n) → O(n + k)
        - *n = number of elements*
        - *k = range of values*
    - Merge Sort Approach: O(recursion) → O(total recursion levels × cost at each level) → O(logn × n) → O(nlogn)
        - *recursion levels = logn because the array is divided in half at each level*
        - *cost at each level = n because merging processes all n elements*
    - Quick Sort Approach: O(recursion levels × partitioning at each level) → O(n × n) → O(n²)
        - *worst case occurs when the pivot creates highly unbalanced partitions*
    - Heap Sort Approach: O(build heap + moving largest elements to the end × heapify to restore the heap property) → O(n + n × logn) → O(nlogn)
        - *heapify starts at one node and moves down at most the height of the heap, which is O(logn)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Bubble Sort Approach: O(1)
    - Selection Sort Approach: O(1)
    - Insertion Sort Approach: O(1)
    - Counting Sort Approach: O(counting array) → O(range of values) → O(k)
    - Merge Sort Approach: O(recursion stack + left array + right array) → O(recursion depth/levels + n1 + n2) → O(logn + n) → O(n)
    - Quick Sort Approach: O(recursion stack) → O(recursion depth/levels) → O(n)
    - Heap Sort Approach: O(1)
        - *the heap is built directly inside the input array*


## Code

```python
# Approach 1: Bubble Sort
class Solution:

    def sortArray(self, nums: List[int]) -> List[int]:
        n = len(nums)
        for i in range(n):
            swapped = False
            for j in range(n - 1 - i):
                if nums[j] > nums[j + 1]:
                    nums[j], nums[j + 1] = nums[j + 1], nums[j]
                    swapped = True
            if not swapped:
                break
        return nums


# Approach 2: Selection Sort
class Solution:

    def sortArray(self, nums: List[int]) -> List[int]:
        n = len(nums)
        for i in range(n):
            min_index = i
            for j in range(i + 1, n):
                if nums[j] < nums[min_index]:
                    min_index = j
            nums[i], nums[min_index] = nums[min_index], nums[i]
        return nums
        
        
# Approach 3: Insertion Sort
class Solution:

    def sortArray(self, nums: List[int]) -> List[int]:
        for i in range(1, len(nums)):
            key = nums[i]
            j = i - 1
            while j >= 0 and nums[j] > key:
                nums[j + 1] = nums[j]
                j -= 1
            nums[j + 1] = key
        return nums


# Approach 4: Counting Sort
class Solution:

    def sortArray(self, nums: List[int]) -> List[int]:
        mini = min(nums)
        maxi = max(nums)
        count = [0] * (maxi - mini + 1)
        for num in nums:
            count[num - mini] += 1
        index = 0
        for i in range(len(count)):
            while count[i] > 0:
                nums[index] = i + mini
                index += 1
                count[i] -= 1
        return nums


# Approach 5: Merge Sort
class Solution:

    def sortArray(self, nums: List[int]) -> List[int]:
        n = len(nums)
        if n > 1:

            # divide 
            mid = n // 2
            left, right = nums[:mid], nums[mid:]

            # sort
            self.sortArray(left)
            self.sortArray(right)

            # merge
            i, j, k = 0, 0, 0
            while i < len(left) and j < len(right):
                if left[i] <= right[j]:
                    nums[k] = left[i]
                    i += 1
                else:
                    nums[k] = right[j]
                    j += 1
                k += 1
            while i < len(left):
                nums[k] = left[i]
                i += 1
                k += 1
            while j < len(right):
                nums[k] = right[j]
                j += 1
                k += 1

        return nums


# Approach 6: Quick Sort
class Solution:

    def sortArray(self, nums: List[int]) -> List[int]:

        def quick_sort(left, right):
            if left >= right:
                return
            pivot = nums[right]
            i = left
            for j in range(left, right):
                if nums[j] < pivot:
                    nums[i], nums[j] = nums[j], nums[i]
                    i += 1
            nums[i], nums[right] = nums[right], nums[i]
            quick_sort(left, i - 1)
            quick_sort(i + 1, right)

        quick_sort(0, len(nums) - 1)
        return nums


# Approach 7: Heap Sort
class Solution:

    def sortArray(self, nums: List[int]) -> List[int]:
        n = len(nums)

        def heapify(size, root):
            while True:
                largest = root
                left = 2 * root + 1
                right = 2 * root + 2
                if left < size and nums[left] > nums[largest]:
                    largest = left
                if right < size and nums[right] > nums[largest]:
                    largest = right
                if largest == root:
                    break
                nums[root], nums[largest] = nums[largest], nums[root]
                root = largest

        # build max heap
        for i in range(n // 2 - 1, -1, -1):
            heapify(n, i)

        # move largest element to the end
        for i in range(n - 1, 0, -1):
            nums[0], nums[i] = nums[i], nums[0]
            heapify(i, 0)

        return nums
        
```
