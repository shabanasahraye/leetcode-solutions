# <center> 238. Product of Array Except Self </center>


## Problem Description
[Click here](https://leetcode.com/problems/product-of-array-except-self/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Use Prefix and suffix product.

answer[i] <br>
= product of all elements except nums[i] <br>
= product of all elements before nums[i] × product of all elements after nums[i] <br>
= prefix product × suffix product


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = size of the input array
- set res = an array of size n with default value 1 
- set prefix = 1 to track the prefix product
- set suffix = 1 to track the suffix product
- traverse the array from start to end to get the prefix product for each element
    - for i in range 0 to n
        - multiply res[i] by the current prefix product to store the product of all elements before nums[i]
        - update the prefix product by multiplying it with nums[i]
- traverse the array in reverse to get the suffix product for each element
    - for i in range n - 1 to 0
        - multiply res[i] by the current suffix product to include the product of all elements after nums[i]
        - update the suffix product by multiplying it with nums[i]
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(creating result array + finding prefix product + finding suffix product) → O(n + n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result array) → O(n) → but according to the follow-up the output array doesn't count as extra space → O(1)


## Code

```python
class Solution:

    def productExceptSelf(self, nums: List[int]) -> List[int]:
        n = len(nums)
        res = [1] * len(nums)
        prefix, suffix = 1, 1
        for i in range(n):
            res[i] *= prefix
            prefix *= nums[i]
        for i in range(n - 1, -1, -1):
            res[i] *= suffix
            suffix *= nums[i]
        return res
         
```
