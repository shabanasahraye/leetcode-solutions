# <center> 724. Find Pivot Index </center>


## Problem Description
[Click here](https://leetcode.com/problems/find-pivot-index/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
If left sum == right sum, the current index is the pivot index.

For each element, we need to find the left and right sum:
- left sum = sum of all elements before the current index = prefix sum
- right sum = sum of all elements after the current index = suffix sum


## Approach
<!-- Describe your approach to solving the problem. -->
- set left_sum = 0 to track the prefix sum i.e the sum of elements before the current index
- set right_sum = total sum of the array to track the suffix sum
- traverse the array <br>
*for each index, number in the array*
    - subtract the current element from the right_sum to get the suffix sum for the current index
      - if left_sum is equal to the right_sum
            - return the current index
      - add the current element to left_sum so it becomes the prefix sum for the next index
- if the loop completes without returning anything, there's no pivot, return -1


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(finding sum + array traversal) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def pivotIndex(self, nums: List[int]) -> int:
        left_sum = 0
        right_sum = sum(nums)
        for i, n in enumerate(nums):
            right_sum -= n
            if left_sum == right_sum:
                return i
            left_sum += n
        return -1
        
```
