# <center> 1523. Count Odd Numbers in an Interval Range </center>


## Problem Description
[Click here](https://leetcode.com/problems/count-odd-numbers-in-an-interval-range/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to count the odd numbers in the inclusive range [low, high].

Let:
- total numbers = n = high − low + 1
- if n is even, half the numbers are odd
- if n is odd, there is one extra odd number only if low is odd
  - *when n is odd, low and high always have the same parity, so checking either endpoint works*


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = high - low + 1 to get the total numbers in the inclusive range
- if n is odd and low is odd
  - return n // 2 + 1
- else return n // 2


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(1)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)



## Code

```python
class Solution:

    def countOdds(self, low: int, high: int) -> int:
        n = high - low  + 1
        if (n & 1) and (low & 1):
            return n // 2 + 1
        return n // 2
        
```
