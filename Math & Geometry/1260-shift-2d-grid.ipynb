## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to shift the grid to the right k times.

- **Approach 1: O(m × n) space**
    - initialize an m × n result matrix, treat the 2D grid as a 1D array, shift each element's index to the right by k positions, then map the shifted index back to the 2D matrix
  - convert 2D index to 1D
    - (row, col) → row × total columns + col
  - shift the 1D index
    - (1D index + k) % (2D array size)
  - convert 1D index back to 2D
    - row = 1D index // total columns
    - col = 1D index % total columns
- **Approach 2: O(1) space**
    - treat the 2D grid as a 1D array and rotate it to the right by k positions using the reversal algorithm


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1
- set m = total rows, n = total columns
- set k
- if k is 0
    - return the grid as it is
- initialize m × n matrix with zeroes
- for each row
    - for each column
        - convert the current 2D index into its corresponding 1D index and shift it to the right by k positions
        - convert the shifted index to 2D index i.e find row number x and column number y
        - place the current cell value at the shifted 2D index in the result matrix
- return the result matrix

### Approach 2
- set m = total rows, n = total columns
- set k = k % (m × n)
    - *this reduces unnecessary shifts because every m × n shifts return the grid to its original state*
- if k is 0, return the grid as it is
- reverse all elements in the grid (treated as a 1D array)
- reverse the first k elements
- reverse the remaining elements
- reverse()
    - while left pointer l < right pointer r
        - convert the left and right 1D indices to 2D indices and swap the values
            - *left row = l // n, left col = l % n*
            - *right row = r // n, right col = r % n*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Approach 1: O(result matrix creation + matrix traversal) → O(m × n + m × n) → O(m × n)
    - Approach 2: O(reverse all elements + reverse first k elements + reverse remaining elements) → O(m × n + k + (m × n − k)) → O(m × n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Approach 1: O(result matrix) → O(m × n)
    - Approach 2: O(1)


## Code

```python
class Solution:

    def shiftGrid(self, grid: List[List[int]], k: int) -> List[List[int]]:
        
        # Approach 1: O(mn) space
        # m, n = len(grid), len(grid[0])
        # k %= m * n
        # if not k:
        #     return grid
        # res = [[0] * n for _ in range(m)]
        # for i in range(m):
        #     for j in range(n):
        #         index = (i * n + j + k) % (m * n)
        #         x = index // n
        #         y = index % n
        #         res[x][y] = grid[i][j]
        # return res


        # Approach 2: O(1) space
        m, n = len(grid), len(grid[0])
        k %= m * n
        if not k:
            return grid

        def reverse(l: int, r: int) -> None:
            while l < r:
                lx, ly = l // n, l % n
                rx, ry = r // n, r % n
                grid[lx][ly], grid[rx][ry] = grid[rx][ry], grid[lx][ly]
                l += 1
                r -= 1

        reverse(0, m * n - 1)
        reverse(0, k - 1)
        reverse(k, m * n - 1)
        return grid
        
```
