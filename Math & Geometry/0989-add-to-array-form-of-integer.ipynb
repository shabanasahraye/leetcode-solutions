# <center> 989. Add to Array-Form of Integer </center>


## Problem Description
[Click here](https://leetcode.com/problems/add-to-array-form-of-integer/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an integer in array form and an integer k. We need to return the array form of their sum.

Follow the addition rules:
- start from the right side and add the corresponding two digits
- if sum > 9, store the rightmost digit and pass the leftmost digit as the carry


## Approach
<!-- Describe your approach to solving the problem. -->
- treat k as the carry
- traverse the digits of the number (array) from right to left
    - add the current digit and the carry (k)
    - store the rightmost digit of the sum (sum % 10) and update the carry (sum // 10)
        - *you can also use divmod()*
    - if the carry becomes 0, stop because no further digits will change
- if a carry remains after processing all digits, add the remaining digits of k to the front of the array. This can be done by:
    - either reverse the array, add the remaining digits of k to the end, and reverse the array again
        - *time complexity = O(reverse + loop to add k + reverse) → O(n + k + n) → O(n)*
        - *space complexity = O(1)*
    - or use a deque because it takes O(1) time for insertion at both ends
        - *time complexity O(deque creation + k) → O(n + k) → O(n)*
        - *space complexity O(deque) → O(n)*
- return num


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(traversal + reverse + remaining digits + reverse) → O(n + n + k + n) → O(n)
    - *k = number of remaining digits in the integer k*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def addToArrayForm(self, num: List[int], k: int) -> List[int]:
        for i in range(len(num) - 1, -1, -1):
            sum = num[i] + k
            num[i], k = sum % 10, sum // 10
            if not k: 
                break
        if k:
            num.reverse()
            while k:
                digit, k = k % 10, k // 10
                num.append(digit)
            num.reverse()
        return num
        
```
