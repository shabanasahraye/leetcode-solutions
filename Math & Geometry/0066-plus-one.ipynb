# <center> 66. Plus One </center>


## Problem Description
[Click here](https://leetcode.com/problems/plus-one/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array of digits representing a +ve integer and need to add 1 to it.

Use the addition rules:
- start from the right side
- add the digits
- pass the carry to the next digit

A carry occurs only when the sum is a two-digit number. Since we only add 1, there will be a carry only if the digit is 9 (9 + 1 = 10).


## Approach
<!-- Describe your approach to solving the problem. -->
- set carry = 1
    - *initially the carry is 1 because we have to add 1*
- traverse the digits array from right to left
    - if the current digit is 9
        - replace it with 0
            - *9 + 1 = 10, we keep the rightmost digit i.e 0, and pass the carry i.e 1*
    - else the digit is less than 9
        - add the carry (1) to it
        - set carry as 0
        - break the loop
- if carry is still 1 after the loop
    - insert 1 at position 0
- return the digits array


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal + insert) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def plusOne(self, digits: List[int]) -> List[int]:
        carry = 1
        for i in range(len(digits) -1, -1, -1):
            if digits[i] == 9:
                digits[i] = 0
            else:
                digits[i] += 1
                carry = 0
                break
        if carry:
            digits.insert(0, 1)
        return digits
        
```
