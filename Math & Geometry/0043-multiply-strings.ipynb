# <center> 43. Multiply Strings </center>


## Problem Description
[Click here](https://leetcode.com/problems/multiply-strings/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given two numbers as strings and need to return their product as a string.

Since we cannot convert the entire numbers to integers, simulate multiplication digit by digit. Convert each character digit into an integer and apply multiplication rules.

Multiplication rules:
- start from the right side
- each time take a digit of num1 and multiply it with all digits of num2
- if the product is a two-digit number, keep the rightmost digit and pass (add) the leftmost digits as a carry to the next digit


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if num1 is 0 or num2 is 0
        - return 0 because multiplication with 0 results in 0
- set m = size of string1, n = size of string2
- set res = a list of size m + n initialized with zeroes to store the multiplication result digits
    - *m + n because multiplying an m-digit number and an n-digit number can have at most m + n digits*
- traverse num1 digits from right to left 
    - for each num1 digit, traverse num2 digits from right to left
        - multiply num1 digit at position i with num2 digit at position j
            - *get the digit by subtracting the ASCII value of '0' from the current character's ASCII value*
        - add the product with the cary at index i + j + 1 in the result list
        - store the rightmost digit of the sum
            - *rightmost digit = sum % 10*
        - pass the leftmost digit as carry to the next digit at position i + j
            - *carry = leftmost digit = sum // 10* 
- convert the result list to a string, remove leading zeroes, and return the result


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(num1 traversal × num2 traversal + result conversion) → O(m × n + m + n) → O(m × n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result list) → O(m + n)


## Code

```python
class Solution:

    def multiply(self, num1: str, num2: str) -> str:
        if num1 == '0' or num2 == '0':
            return '0'
        m, n = len(num1), len(num2)
        res = [0] * (m + n)
        for i in range(m - 1, -1, -1):
            for j in range(n - 1, -1, -1):
                prod = (ord(num1[i]) - 48) * (ord(num2[j]) - 48)   # 48 is ord('0')
                summ = prod + res[i + j + 1]
                res[i + j + 1] = summ % 10
                res[i + j] += summ // 10
        return ''.join(map(str, res)).lstrip('0')
        
```
