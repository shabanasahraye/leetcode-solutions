# <center> 1572. Matrix Diagonal Sum </center>


## Problem Description
[Click here](https://leetcode.com/problems/matrix-diagonal-sum/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
For a square matrix, we need to sum the primary and secondary diagonals.

**Primary diagonal**
- elements where row index equals column index
- indices: (0,0), (1,1), (2,2)
- matrix[i][i]

**Secondary diagonal**
- elements where row index + column index equals n - 1
- indices: (0,2), (1,1), (2,0)
- matrix[i][n - i - 1] or matrix[i][~i] in Python


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = number of rows in the matrix
- set res = 0 to track the diagonal sum
- traverse each row
  - add sum of primary diagonal element (matrix[i][i]) and the secondary diagonal element (matrix[i][~i]) to res
- if n is odd, the middle element is counted twice
  - subtract middle element matrix[n // 2][n // 2] once
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(matrix traversal) → O(n)
    - *n = number of rows*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def diagonalSum(self, mat: List[List[int]]) -> int:
        n = len(mat)
        res = 0
        for i in range(n):
            res += mat[i][i] + mat[i][~i]   # mat[i][~i] is equivalent to mat[i][n- i - 1]
        if n & 1:
            res -= mat[n // 2][n // 2]
        return res
```
