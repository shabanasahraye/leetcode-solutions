## Problem Description
[Click here](https://leetcode.com/problems/robot-bounded-in-circle/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
A robot is bounded in a circle if, after executing the instructions once:
- it returns to the starting position (0, 0), or
- it ends facing a direction other than north (0, 1)
    - *if the robot is not facing north, repeating the instructions will eventually make its path loop back*

north = +y, south = -y, east = +x, west = -x

**Instruction G changes the position**
- moves the robot one step in the current direction
    - x → x + dir_x
    - y → y + dir_y
- north (0, 1) → (x, y) → (x, y + 1)
- south (0, -1) → (x, y) → (x, y - 1)
- east (1, 0) → (x, y) → (x + 1, y)
- west (-1, 0) → (x, y) → (x - 1, y)

**Instructions L and R rotate the current direction**
- L rotates the direction 90° counterclockwise
    - dir_x → -dir_y
    - dir_y → dir_x
    - north (0, 1) → west (-1, 0)
    - west (-1, 0) → south (0, -1)
    - south (0, -1) → east (1, 0)
    - east (1, 0) → north (0, 1)
- R rotates the direction 90° clockwise
    - dir_x → dir_y
    - dir_y → -dir_x
    - north (0, 1) → east (1, 0)
    - east (1, 0) → south (0, -1)
    - south (0, -1) → west (-1, 0)
    - west (-1, 0) → north (0, 1)


## Approach
<!-- Describe your approach to solving the problem. -->
- set direction variables dir_x = 0, dir_y = 1 because the robot's initial direction is north (0, 1)
- set position variables x = 0, y = 0 because the robot's initial position is (0, 0)
- traverse the instructions string
    - if the instruction is 'G'
        - update position (x, y)
    - if the instruction is 'L'
        - update direction (dir_x, dir_y)
    - if the instruction is 'R'
        - update direction (dir_x, dir_y)
- return true if the robot returns to the starting position or ends facing a direction other than north, else return false


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(input array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def isRobotBounded(self, instructions: str) -> bool:
        dir_x, dir_y = 0, 1
        x, y = 0, 0
        for i in instructions:
            if i == 'G':
                x, y = x + dir_x, y + dir_y
            elif i == 'L':
                dir_x, dir_y = -1 * dir_y, dir_x
            elif i == 'R':
                dir_x, dir_y = dir_y, -1 * dir_x
        return (x, y) == (0, 0) or (dir_x, dir_y) != (0, 1)
        
```
