# <center> 7. Reverse Integer </center>


## Problem Description
[Click here](https://leetcode.com/problems/reverse-integer/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an integer. We need to reverse its digits.

To reverse an integer, repeatedly remove the rightmost digit from the original number and place it on the left side of the result.
- to get the rightmost digit, take the number modulo 10
- to remove the rightmost digit, divide the number by 10 (integer division)
- to place a digit on the left side, multiply by 10 to shift the existing number one place left and add the digit

For example: x = 123
- x = 123, res = 0
    - digit = 123 % 10 = 3
    - res = 0 × 10 + 3 = 3
    - x = 123 // 10 = 12
- x = 12, res = 3
    - digit = 12 % 10 = 2
    - res = 3 × 10 + 2 = 32
    - x = 12 // 10 = 1
- x = 1, res = 32
    - digit = 1 % 10 = 1
    - res = 32 × 10 + 1 = 321
    - x = 1 // 10 = 0

*Note: check for integer overflow during reversing*


## Approach
<!-- Describe your approach to solving the problem. -->
- set int_min and int_max to the 32-bit signed integer limits
- set sign to -1 if x is negative, else +1
- make x positive 
    - *this is to handle mod for negative values because Python treats negative and positive mod differently*
- set res = 0 to track the reversed integer
- loop while x is not 0 because there are still digits to process
    - extract the rightmost digit from x and append it to res
    - if the reversed integer is out of bounds
        - return 0
    - remove the rightmost digit from x
- return res * sign


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(number of digits) → O(log₁₀|x|)
    - *log₁₀|x| because each time x is divided by 10*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def reverse(self, x: int) -> int:
        int_min = -2 ** 31
        int_max = 2 ** 31 - 1
        sign = -1 if x < 0 else 1
        x *= sign
        res = 0
        while x:
            res = res * 10 + x % 10
            if res > int_max or res * sign < int_min:
                return 0
            x //= 10
        return res * sign
        
```
