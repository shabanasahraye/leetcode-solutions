## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
This problem is similar to [Spiral Matrix](https://leetcode.com/problems/spiral-matrix/description/), but instead of reading an existing matrix, we fill a new matrix in spiral order.

An n × n matrix can be divided into square cycles. <br>
The first cycle is formed by the first row, last column, last row, and first column. The second cycle is formed by the second row, second-last column, second-last row, and second column, and so on.

Create an n × n matrix. Divide it into cycles and fill each cycle in clockwise order: top → right → bottom → left.


## Approach
- check base cases
    - if n is 0, return an empty list
- initialize nxn matrix with zeroes
- initialize four variables to track the indices for the corner points
    - *top = 0*
    - *left = 0*
    - *right = rightmost index i.e total columns - 1*
    - *bottom = bottom index i.e total rows - 1*
- set val = 1 to track the value to fill in
-  run a loop until all cycles are processed i.e while left <= right and top <= bottom
    - *traverse the current cycle in clockwise order: top row → right column → bottom row → left column, and fill the cells*
    - traverse the top row from left to right
        - fill the current cell
        - increment val after filling the cell
    - increment top after the loop to point to the top of the next cycle
    - traverse the right col from top to bottom
        - fill the current cell
        - increment val after filling the cell
    - decrement right after the loop to point to the right column of the next cycle
    - if the updated pointers are out of bounds, break the loop
        - *this handles cases where only one row or one column remains*
    - traverse the bottom row in reverse from right to left
        - fill the current cell
        - increment val after filling the cell
    - decrement bottom
    - traverse the left column in reverse from bottom to top
        - fill the current cell
        - increment val after filling the cell
    - increment left
- return the matrix


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(matrix creation + matrix traversal) → O(n² + n²) → O(n²)
    - *matrix creation and filling each visit every cell once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result matrix) → O(n²)


## Code

```python
class Solution:

    def generateMatrix(self, n: int) -> List[List[int]]:
        if not n:
            return []
        matrix = [[0] * n for _ in range(n)]
        left, right = 0, n - 1
        top, bottom = 0, n - 1
        val = 1
        while left <= right and top <= bottom:
            # move left to right and fill top row
            for c in range(left, right+1):
                matrix[top][c] = val
                val += 1
            top += 1
            # move top to bottom and fill right col
            for r in range(top, bottom+1):
                matrix[r][right] = val
                val += 1
            right -= 1
            if not (left <= right and top <= bottom):
                break
            # move right to left and fill bottom row
            for c in range(right, left-1, -1):
                matrix[bottom][c] = val
                val += 1
            bottom -= 1
            # move bottom to top and fill left col
            for r in range(bottom, top-1, -1):
                matrix[r][left] = val
                val += 1
            left += 1
        return matrix
        
```
