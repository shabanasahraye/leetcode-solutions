# <center> 2028. Find Missing Observations </center>


## Problem Description
[Click here](https://leetcode.com/problems/find-missing-observations/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the values of the missing n rolls from the given mean.

To solve:
- find the sum of the missing n rolls
    - *missing sum = total sum - sum of m rolls = mean * (m + n) - sum of m rolls*
- check if the missing sum can be distributed among n rolls
    - if yes, find the average *(missing sum / n)* and distribute it
    - distribute the remainder *(missing sum % n)* also if any


## Approach
<!-- Describe your approach to solving the problem. -->
- compute the total sum i.e mean * (m + n)
- compute the missing sum i.e. total sum - sum of m rolls
- if the missing sum is less than n or more than n * 6
    - return an empty array because it can't be distributed
- find the average and remainder
- create a result array of size n and initialize it  with the average value
    - *this distributes the missing sum equally among all rolls*
- distribute the remainder
    - loop in range of remainder
        - increment the value in the result array by 1
- return the result array


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(finding sum of m rolls + creating result array + distributing remainder) → O(m + n + n) → O(m + n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result array) → O(n)


## Code

```python
class Solution:

    def missingRolls(self, rolls: List[int], mean: int, n: int) -> List[int]:
        total_sum = (len(rolls) + n) * mean
        missing_sum = total_sum - sum(rolls)
        if missing_sum < n or missing_sum > n * 6:
            return []
        avg, remainder = missing_sum // n, missing_sum % n
        res = [avg] * n
        for i in range(remainder):
            res[i] += 1
        return res
        
```
