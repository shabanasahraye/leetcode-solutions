# <center> 1071. Greatest Common Divisor of Strings </center>


## Problem Description
[Click here](https://leetcode.com/problems/greatest-common-divisor-of-strings/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the largest string that divides both input strings.

If a string divides both strings, its length must divide both string lengths. Therefore, the longest possible candidate has length equal to the GCD of the two string lengths.

Example: 
- str1 = 'ABCABC', length = 6
- str2 = 'ABC', length = 3
- gcd(6, 3) = 3
- the longest possible divisor string has length 3, so the answer is str2[0:3] = 'ABC'


## Approach
<!-- Describe your approach to solving the problem. -->
- check edge cases
    - if both strings are the same
        - return either string
    - if str1 + str2 is not equal to str2 + str1, no string can divide both strings
        - return an empty string
- find the GCD of the lengths of both strings
    - use Euclid's algorithm or the library function math.gcd()
- return the prefix of length gcd from either string


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(str1 + str2 comparison + gcd) → O((m + n) + log(min(m, n))) → O(m + n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(str1 + str2 + gcd recursion) → O(m + n + log(min(m, n))) → O(m + n)


## Code

```python
class Solution:

    def gcdOfStrings(self, str1: str, str2: str) -> str:
        if str1 == str2:
            return str1
        elif str1 + str2 != str2 + str1:
            return ''
        
        def gcd(a: int, b: int) -> int:
            if not b:
                return a
            return gcd(b, a % b)

        g = gcd(len(str1), len(str2))
        return str2[:g]
        
```
