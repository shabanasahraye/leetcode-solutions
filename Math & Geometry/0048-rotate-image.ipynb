## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to rotate an n × n matrix 90 degrees clockwise in-place.

- **Approach 1: Transpose and reverse O(n² + n²)**
    - for clockwise rotation = first take transpose and then reverse the rows
    - for anticlockwise rotation = vice versa

- **Approach 2: Rotate matrix cycles O(n²)**
    - an n × n matrix contains n/2 square cycles
    - each cycle consists of four sides:
        - top row
        - right column
        - bottom row
        - left column
    - for clockwise rotation: rotate each group of four elements in a clockwise direction
        - top → right
        - right → bottom
        - bottom → left
        - left → top
    - for anticlockwise rotation: vice versa


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: Rotate matrix cycles
- set n = size of the matrix
- traverse each cycle using two loops
    - *outer loop handles the current cycle*
    - *inner loop handles each group of four elements in the cycle*
    - for each cycle
        - for each group of four elements
            - swap the elements clockwise
                - top left → top right
                - top right → bottom right
                - bottom right → bottom left
                - bottom left → top left
            - *positions:*
                - *top left:     matrix[i][j]*
                - *top right:    matrix[j][~i]*
                - *bottom right: matrix[~i][~j]*
                - *bottom left:  matrix[~j][i]*
                - *~x = n - 1 - x, which gives the mirrored index*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Approach 2: O(matrix traversal) → O(n × n) → O(n²)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - Approach 2: O(1)


## Code

```python
class Solution:

    def rotate(self, matrix: List[List[int]]) -> None:
        """
        Do not return anything, modify matrix in-place instead.
        """

        # Approach 2: Rotate matrix cycles
        n = len(matrix)
        for i in range(n // 2):
            for j in range(i, n - i - 1):
                # ~num = n - 1 - num
                matrix[i][j], matrix[~j][i], matrix[~i][~j], matrix[j][~i] = matrix[~j][i], matrix[~i][~j], matrix[j][~i], matrix[i][j]
    
                
```
