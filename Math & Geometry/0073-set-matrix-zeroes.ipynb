# <center> 73. Set Matrix Zeroes </center>


## Problem Description
[Click here](https://leetcode.com/problems/set-matrix-zeroes/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to set the entire row and column of every 0 to 0 while using O(1) extra space.

Instead of using extra arrays to remember which rows and columns should become 0, reuse the matrix itself.
- if a cell is 0
    - set the first cell of its row to 0
    - set the first cell of its column to 0
- the first cell of each row and column acts as a marker indicating whether that row or column should later become 0

After marking all rows and columns, traverse the matrix in reverse order. If the marker for the current row or column is 0, set the current cell to 0.
- *traverse in reverse to avoid overwriting the markers before using them*


## Approach
<!-- Describe your approach to solving the problem. -->
- set col0 = 1 to track whether the first column originally contains a 0
    - *1 means the first column doesn't contain a 0, and 0 means it does*
    - *the first row (matrix[0][j]) stores column markers, and the first column (matrix[i][0]) stores row markers*
    - *matrix[0][0] belongs to both the first row and first column, so it is used only as the first-row marker*
    - *col0 is the marker for the first column because its cells are reused as row markers*
- set m = total rows, n = total columns
- traverse the matrix
    - for each row i
        - if value in the first col of this row is 0
            - set col0 to 0
        - mark the current row and col
            - traverse all columns except the first 
                - if current cell at index (i, j) is 0
                    - mark the current row and column by setting:
                        - the first cell of the current row (matrix[i][0]) to 0
                        - the first cell of the current column (matrix[0][j]) to 0
- traverse the matrix again in reverse order
    - for each row i
        - for each column j
            - if the marker for the current row or column is 0
                - set the current cell to 0
        - if first col marker is 0
            - set the value in the first col of this row to 0


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(matrix traversal + matrix traversal) → O(rows x cols + rows x cols) → O(m x n + m x n) → O(m x n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def setZeroes(self, matrix: List[List[int]]) -> None:
        """
        Do not return anything, modify matrix in-place instead.
        """
        col0 = 1
        m = len(matrix)
        n = len(matrix[0])
        for i in range(m):
            if matrix[i][0] == 0:
                col0 = 0
            for j in range(1, n):
                if matrix[i][j] == 0:
                    matrix[i][0] = 0
                    matrix[0][j] = 0
        for i in range(m - 1, -1, -1):
            for j in range(n - 1, 0, -1):
                if matrix[i][0] == 0 or matrix[0][j] == 0:
                    matrix[i][j] = 0
            if col0 == 0:
                matrix[i][0] = 0
                
```
