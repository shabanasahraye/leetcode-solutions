# <center> 202. Happy Number </center>


## Problem Description
[Click here](https://leetcode.com/problems/happy-number/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
A number is happy if repeatedly replacing it with the sum of the squares of its digits eventually reaches 1, which forms the cycle 1 → 1.

Use Floyd’s Cycle detection algorithm
- it is based on two-pointers: fast and slow
- the slow pointer moves ahead one step at a time, and the fast pointer moves ahead two steps at a time
- the pointers meet if there is a cycle

If the pointers meet at 1, the number is happy, else they meet in a different cycle, so the number is not happy.


## Approach
<!-- Describe your approach to solving the problem. -->
- define a helper function to get the sum of squares of digits <br>
*sum_square_digits(number)*
    - set res = 0 to track the sum
    - loop until the number becomes 0
        - get the rightmost digit of the number using mod operator
        - get the square of the digits and add it to res
        - remove the digit from the number
    - return res
- set slow = n and fast = sum of squares of digits of n
- loop while the pointers don't meet
    - move slow pointer 1 time
        - replace slow with the sum of squares of digits of slow
    - move fast pointer 2 times
        - replace fast with the sum of squares of digits of fast
        - replace fast with the sum of squares of digits of fast
- return whether the meeting point is 1


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(cycle detection × sum of squares of digits) → O(constant × logn) → O(logn)
    - *the cycle detection loop runs only a constant number of times because the sequence eventually enters a fixed cycle*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def isHappy(self, n: int) -> bool:
        
        def sum_square_digits(n: int) -> int:
            res = 0
            while n:
                digit = n % 10
                res += digit * digit
                n //= 10
            return res

        slow, fast = n, sum_square_digits(n)
        while slow != fast:
            slow = sum_square_digits(slow)
            fast = sum_square_digits(fast)
            fast = sum_square_digits(fast)
        return fast == 1
        
```
