# <center> 149. Max Points on a Line </center>


## Problem Description
[Click here](https://leetcode.com/problems/max-points-on-a-line/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the maximum number of points that lie on the same straight line.

Points with the same slope (relative to a fixed point) lie on the same line.

Fix one point and compute its slope with every other point. Use a hashmap to count how many points have each slope and keep track of the maximum count.

P1 = (x1, y1)
P2 = (x2, y2)
slope = (y2 − y1) / (x2 − x1)


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 1 to track the max number of points on the same line
    - *1 because there's at least one point*
- set n = length of the input points array
- create a hashmap to count the frequency of each slope 
- loop in range n
    - get the first point p1
    - clear the hashmap
    - loop in range i + 1 to n to compute slope with every other point
        - get the second point p2
        - if both points have the same x-coordinate, the line is vertical
            - set the slope to float('inf')
        - else compute the slope
            - *index 0 is x, index 1 is y*
        - increment the current slope count by 1
        - update res with the max of the previous max and the current slope count + 1 
            - *+1 includes the starting point
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(outer loop × inner loop) → O(n × n) → O(n²)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(n)



## Code

```python
class Solution:

    def maxPoints(self, points: List[List[int]]) -> int:
        res = 1
        n = len(points)
        count = defaultdict(int)
        for i in range(n):
            p1 = points[i]
            count.clear()
            for j in range(i + 1, n):
                p2 = points[j]
                if p2[0] == p1[0]:
                    slope = float('inf')
                else:
                    slope = (p2[1] - p1[1]) / (p2[0] - p1[0])
                count[slope] += 1
                res = max(res, count[slope] + 1)
        return res
        
```
