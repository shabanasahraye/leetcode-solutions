## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
A number x is a palindrome if: it is equal to its reverse, or its left half is equal to the reversed right half.

- **Approach 1**
    - reverse the entire number and compare it with the original number
- **Approach 2**
    - reverse only the right half and compare it with the left half
    - this also handles the integer overflow

For number reversal logic, see [Reverse Integer Solution](https://leetcode.com/problems/reverse-integer/solutions/3845859/o-logx-time-o-1-space-solution-explained/)
 
 
## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1
- check base cases
    - if the integer x is negative or ends with a zero i.e x is not 0 but a number that ends with 0
        - return false becuase it can't be a palindrome
- set y = x and rev = 0 to track the digits in the original and reversed part
- loop until y is not 0 i.e there are digits left to be processed
    - extract the rightmost digit of y and append it to rev
    - divide y by 10 to remove the rightmost digit 
- compare x and reverse of x and return the result

### Approach 2
- check base cases
    - if the integer x is negative or ends with a zero i.e x is not 0 but a number that ends with 0
        - return false becuase it can't be a palindrome
- set left = x and right = 0 to store the left half and reversed right half
- loop until we have processed half of the digits
    -  extract the rightmost digit from left and append it to right 
    - divide x by 10 to remove the digit from the left half
-  ccompare the left and right halves and return the result
    - *after the loop, left contains the remaining left half and right contains the reversed right half (including the middle digit for odd-length numbers)*
    - *compare:*
        - *left == right for even length*
        - *left == right // 10 for odd length (remove the middle digit)*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - Approach 1: O(loop) → O(log₁₀x)
        - *log₁₀x because each iteration divides x by 10*
    - Approach 2: O(loop) → O(log₁₀x)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Approach 1: O(1)
    - Approach 2: O(1)


## Code

```python
class Solution:

    def isPalindrome(self, x: int) -> bool:
        
        # Approach 1
        # if x < 0 or x != 0 and x % 10 == 0:
        #     return False
        # y, rev = x, 0
        # while y:
        #     rev = rev * 10 + y % 10
        #     y //= 10
        # return x == rev

        # Approach 2
        if x < 0 or x != 0 and x % 10 == 0:
            return False
        left, right = x, 0
        while left > right:
            right = right * 10 + left % 10
            left //= 10
        return left == right or left == right // 10
        
```
