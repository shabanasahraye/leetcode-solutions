# <center> 6. Zigzag Conversion </center>


## Problem Description
[Click here](https://leetcode.com/problems/zigzag-conversion/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->

We are given a string and the number of rows. We need to arrange the characters in a zigzag pattern and return the string formed by reading the rows one by one.

For example: 
- input string = "PAYPALISHIRING", rows = 3
- zigzag pattern <br>
    ```
    P   A   H   N
    A P L S I I G
    Y   I   R
    ```
- output string = row1 letters + row2 letters + row3 letters = PAHN + APLSIIG + YIR = 'PAHNAPLSIIGYIR'

Zigzag pattern observations:
- we move from the top row to the bottom row, then back up to the top
- if we are in the first row, we move down 
- if we are in the last row, we move up 

Traverse the string, place each character in its current row, and track the current row and movement direction.


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if the number of rows is 1 or equal to the length of the string
        - return the string
- set res = an array of strings of size equal to total rows to store each row letters
- set row = 0 to track the current row index
- set down = 1 to track the direction
    - *1 means to move down, and -1 to move up*
- traverse the string
    - add the current character to the current row
    - if the current row is the first row
        - change direction to down
    - if the current row is the last row
        - change direction to up
    - update the current row using the current direction
        - *if the direction is down (+1), increment the row*
        - *if the direction is up (-1), decrement the row*
- convert the result array to a string and return it


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal + array to string conversion) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result array) → O(n)


## Code

````python
class Solution:

    def convert(self, s: str, numRows: int) -> str:
        if numRows == 1 or numRows == len(s):
            return s
        res = [''] * numRows
        row = 0
        down = 1
        for c in s:
            res[row] += c
            if row == 0:
                down = 1
            elif row == numRows - 1:
                down = -1
            row += down
        return ''.join(res)
        
````
