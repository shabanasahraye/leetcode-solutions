# <center> 54. Spiral Matrix </center>


## Problem Description
[Click here](https://leetcode.com/problems/spiral-matrix/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to return the elements of an n × n matrix in spiral order.

A matrix can be divided into square cycles.
- the first cycle is formed by the first row, last column, last row, and first column
- the second cycle is formed by the second row, second-last column, second-last row, and second column, and so on

Traverse each cycle in clockwise order: top → right → bottom → left, and add the elements to the result.


## Approach
<!-- Describe your approach to solving the problem. -->
- initialize four variables to track the indices of the corner points 
    - *top = 0*
    - *left = 0*
    - *right = rightmost index i.e total columns - 1*
    - *bottom = bottom index i.e total rows - 1*
- set res = a list to store the spiral matrix
- run a loop until all cycles are processed i.e while left <= right and top <= bottom
    - *traverse the current cycle in clockwise order: top row → right column → bottom row → left column, and add the elements to res*
    - traverse the top row from left to right 
        - add the elements to res
    - increment top after the loop to point to the top of the next cycle
    - traverse the right column from top to bottom
        - add the elements to res
    - decrement right after the loop to point to the right column of the next cycle
    - if the updated pointers are out of bounds, break the loop
        - *this handles cases where only one row or one column remains*
    - traverse the bottom row from right to left 
        - add the elements to res
    - decrement bottom
    -  traverse the left column from bottom to top 
        - add the elements to res
    - increment left
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(matrix traversal) → O(m × n)
    - *each element is visited exactly once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result list) → O(m × n)


## Code

```python
class Solution:

    def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
        left, right = 0, len(matrix[0]) - 1
        top, bottom = 0, len(matrix) - 1
        res = []
        while left <= right and top <= bottom:
            # move left to right and get top row elements
            for c in range(left, right + 1):
                res.append(matrix[top][c])
            top += 1
            # move top to bottom and get right col elements
            for r in range(top, bottom + 1):
                res.append(matrix[r][right])
            right -= 1
            if not (left <= right and top <= bottom):
                break
            # move right to left and get bottom row elements 
            for c in range(right, left - 1, -1):
                res.append(matrix[bottom][c])
            bottom -= 1
            # move bottom to top and get left col elements
            for r in range(bottom, top - 1, -1):
                res.append(matrix[r][left])
            left += 1
        return res
        
```
