## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Given a query point, We need to find total number of squares with positive area. 

Use a hashmap to store points and their frequencies.

For a square, we need four points:
- 1st point P1 is the query point
- search for a diagonal point P2 in the hashmap, P1 and P2 can form a diagonal if the width and height difference is the same (P1x - P2x == P1y - P2y)
- get the other two points by swapping coordinates
    - P3 = (P1x, P2y)
    - P4 = (P2x, P1y)

For positive area, the points shouldn't be in the same spot. No need to check all points, only check the x or y coordinates of P1 and P2.

The number of squares is calculated by multiplying the frequencies of the three stored points, we don't multiply the query point frequency because it is always 1.


# Approach
<!-- Describe your approach to solving the problem. -->
**init()**
- create a hashmap to store point frequencies
  - key = point coordinates (x, y)
  - value = frequency of the point

**add()**
- increment the frequency of the point(x, y) in the hashmap
    - *convert the point list to a tuple because lists cannot be hashmap keys*

**count()**
- set res = 0 to store the number of squares
- get the first point (query) coordinates (px, py)
- traverse all points in the hashmap <br>
*for each point (x, y) and its frequency*
    - if the 2nd point is not the same point, the 1st & 2nd points are diagonal, and the other two points exist in the hashmap, they form a square
        - multiply the frequencies of the three stored points to get the number of squares and update res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - init(): O(1)
    - add(): O(1)
    - count(): O(hashmap traversal) → O(hashmap size) → O(unique points) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - init() O(1)
    - add() O(1)
    - count(): O(1)
    - overall space: O(hashmap size) → O(n)
        - *the functions don't need extra space for the computation*
        - *size of hashmap will be n after n calls to add()*


## Code

```python
class DetectSquares:

    def __init__(self):
        self.points = defaultdict(int)

    def add(self, point: List[int]) -> None:
        self.points[tuple(point)] += 1

    def count(self, point: List[int]) -> int:
        res = 0
        px, py = point
        for (x, y), freq in self.points.items():
            if x != px and abs(px - x) == abs(py - y) and (px, y) in self.points and (x, py) in self.points:
                res += self.points[(px, y)] * self.points[(x, py)] * freq
        return res
        


# Your DetectSquares object will be instantiated and called as such:
# obj = DetectSquares()
# obj.add(point)
# param_2 = obj.count(point)
```
