# <center> 50. Pow(x, n) </center>


## Problem Description
[Click here](https://leetcode.com/problems/powx-n/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to calculate xⁿ, but n can be very large.

Since n can be very large, multiplying x repeatedly would be inefficient. <br>
Use binary exponentiation to repeatedly square the base, reducing the number of multiplications.

Example: calculate x¹⁰
- instead of multiplying x 10 times:
    - *x * x * x * x * x * x * x * x * x * x*
- use squaring:
    - x² = x * x
    - x⁴ = x² * x²
    - x⁸ = x⁴ * x⁴
    - x¹⁰ = x⁸ * x²
- this reduces the number of multiplications by reusing previously calculated powers
- instead of 10 multiplications, we use 4 multiplications to calculate x¹⁰


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - return 0 if base is 0 
    - return 1 if power is 0 
- set res = 1 to track the result
- if power is negative
    - make it positive
    - take the reciprocal of x
        - $x^{-2} = \frac{1}{x^2}$
- loop until power becomes 0
    - if power is odd
        - multiply the result by the current base first because an odd power has one unpaired base
    - square the base
    - divide n by 2 to process the next power
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(log₂n)
    - *log₂n because each iteration divides n by 2*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def myPow(self, x: float, n: int) -> float:
        if x == 0:
            return 0
        if n == 0:
            return 1
        if n < 0:
            n = abs(n)
            x = 1 / x
        res = 1
        while n:
            if n & 1:      
                res *= x
            x *= x
            n //= 2       
        return res
        
```
