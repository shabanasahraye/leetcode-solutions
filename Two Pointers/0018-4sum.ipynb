## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
For 4 sum, we need 4 numbers that sum to the target.

- **Approach 1: Reduce 4Sum → 3Sum → 2Sum**   
    - *just add an outer for-loop in the [3Sum Solution](https://leetcode.com/problems/3sum/solutions/3527300/o-n-time-o-n-space-solution-explained)*

- **Approach 2: Design a k-sum solution where k >= 2**
    - recursively reduce kSum to (k-1)Sum until reaching 2Sum
    - solve the final 2Sum using two pointers 


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: k-sum
- sort the input array to handle duplicates easily
- set n = len(nums)
- set res = an array to store the unique quadruplets
- define a recursive helper function for kSum <br>
*kSum(start index, remaining numbers needed, remaining target, current chosen numbers) → find k numbers starting from start that add up to target*
    - check pruning
        - if fewer than k numbers are available, there are not enough numbers to continue
            - return
        - if the minimum possible sum is greater than target
            - return
            - *skip this check because calculating the sum adds extra work*
        - if the maximum possible sum is smaller than target
            - return
            - *skip this check because calculating the sum adds extra work*
    - if only 2 numbers are needed
        - solve using [Two Sum II solution](https://leetcode.com/problems/two-sum-ii-input-array-is-sorted/solutions/3525831/on-time-o1-space-solution-explained-by-s-xc10/)
        - return after finishing 2Sum
    - else, more than 2 numbers are still needed
        - loop through the remaining numbers in the array
            - *start = starting index because previous numbers are already used*
            - *end = n - k + 1 to ensure enough numbers remain to complete the combination*
            - if current number is duplicate
                - skip it to avoid duplicate combinations
            - else
                - choose the current number and add it to path
                - recursively find the remaining (k-1) numbers using the remaining target
                - remove the current number from path to undo the choice
- call kSum with k = 4 to solve 4Sum
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - k-sum Approach: O(sort + kSum recursion) → O(nlogn + $n^{n-1}$) → O(nlogn + $n^{4-1}$) → (nlogn + $n^3$) → O($n^3$)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - k-sum Approach: O(recursion depth + current path + result array ) → O(n + n + $n^4$) → O($n^4$)
        - *the result can contain O(n⁴) unique quadruplets in the worst case*


## Code

```python
# Approach 2: k-sum

class Solution:

    def fourSum(self, nums: List[int], target: int) -> List[List[int]]:
        nums.sort()
        n = len(nums)
        res = []

        def kSum(start, k, target, path):
            # pruning
            if n - start < k:
                return
            # if sum(nums[start:start + k]) > target:
            #     return
            # if sum(nums[n - k:]) < target:
            #     return

            # base case: 2Sum 
            if k == 2:
                left, right = start, n - 1
                while left < right:
                    total = nums[left] + nums[right]
                    if total < target:
                        left += 1
                    elif total > target:
                        right -= 1
                    else:
                        res.append(path + [nums[left], nums[right]])
                        left += 1
                        right -= 1
                        # skip duplicates
                        while left < right and nums[left] == nums[left - 1]:
                            left += 1
                        while left < right and nums[right] == nums[right + 1]:
                            right -= 1
                return

            # reduce kSum to (k-1)Sum
            for i in range(start, n - k + 1):
                # skip duplicates
                if i > start and nums[i] == nums[i - 1]:
                    continue
                # choose → explore → undo
                path.append(nums[i])
                kSum(i + 1, k - 1, target - nums[i], path)
                path.pop()

        kSum(0, 4, target, [])
        return res
        
```
