# <center> 11. Container With Most Water </center>


## Problem Description
[Click here](https://leetcode.com/problems/container-with-most-water/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the maximum area formed by any two lines.

Use two pointers moving in opposite directions:
- one starts at the left end and the other at the right end of the array
- at each step, calculate the area formed by the two pointers
    - area = width × height = distance between the pointers × lower height = (r - l) × min(height[l], height[r])
- since the area is limited by the shorter line, move the pointer with the shorter height to try to find a taller line


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the max area
- set max_h = max height
- set left pointer l = leftmost index and right pointer r = rightmost index
- traverse the array using the pointers (while l < r)
    - get current width i.e r - l
    - calculate the current area using the lower height
        - *area = width × min(height[l], height[r])*
    - update res with the max of current area and previous max area
    - if current max area > maximum possible area (current width × max height), no need to check further because the width decreases
        - break the loop
    - if the left height is lower
        - move the left pointer
    - else
        - move the right pointer
- return res as the max area


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(finding max + two-pointer traversal) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def maxArea(self, height: List[int]) -> int:
        res = 0
        max_h = max(height)
        l, r = 0, len(height) - 1
        while l < r:
            width = r - l
            area = width * min(height[l], height[r])
            res = max(res, area)
            if res > width * max_h:
                break
            if height[l] < height[r]:
                l += 1
            else:
                r -= 1
        return res
```
