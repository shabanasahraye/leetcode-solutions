# <center> 977. Squares of a Sorted Array </center>


## Problem Description
[Click here](https://leetcode.com/problems/squares-of-a-sorted-array/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
The input array is sorted. We need the output in sorted order, but squaring can change the order of negative numbers.

Use two pointers moving in opposite directions (left and right)
- compare the left and right absolute values
- the number with the greater absolute value will have the greater square, so place its square at the rightmost position in the result
    - *for example: -5 and 2 → |-5| > |2| → (-5)² = 25 > 2² = 4*


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = input array size 
- check base cases
    - if n is 1
        - square the first element and return it
- set res = an array of size n with zeros to store the sorted squares
    - *we can also create an empty array and then add the squares but we will have to reverse the array in the end because we are building the result from right side*
- set left pointer l = leftmost index and right pointer r = rightmost index
- set k = rightmost index to track the right index in the result array
    - *we can also use the difference between the left and right pointer to track the current right index in the result array but the code is clear this way*
- loop until l <= r
    - if the left absolute value is greater, the left number square is greater
        - square the left number and place it at the current right index (k) in the result array 
        - decrement k
        - move the left pointer
    - else the right number square is greater
        - square the right number and place it at current right index (k) in the result array
        - decrement k
        - move the right pointer
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(creating result array + two-pointer traversal) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result array) → O(n)


## Code

```python
class Solution:

    def sortedSquares(self, nums: List[int]) -> List[int]:
        n = len(nums)
        if n == 1:
            return [nums[0] * nums[0]]
        res = [0] * n
        l, r = 0, n - 1
        k = r   
        while l <= r:
            if abs(nums[l]) > abs(nums[r]):
                res[k] = nums[l] * nums[l]
                k -= 1
                l += 1
            else:
                res[k] = nums[r] * nums[r]
                k -= 1
                r -= 1
        return res
        
```
