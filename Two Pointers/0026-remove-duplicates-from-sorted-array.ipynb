# <center> 26. Remove Duplicates from Sorted Array </center>


## Problem Description
[Click here](https://leetcode.com/problems/remove-duplicates-from-sorted-array/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
In a sorted array, duplicates are adjacent. 

Use two pointers moving in the same direction:
- one to traverse the array
- one to track the position for placing the next unique element (count = 1)

Traverse the array and place each unique element at the next available position on the left.


## Approach
<!-- Describe your approach to solving the problem. -->
- set i = 1 to track the position for placing the next unique element
    - *1 because the first element at index 0 is always unique*
- set j = 1 to traverse the array
- traverse the array with jnstarting from the second position
    - if the current element is different from the previous element, it is unique 
        - *or compare with nums[i - 1], the previous unique element* 
        - place the element at index i 
        - move the pointer i
    - else the current element is a duplicate, skip it
- return i as total number of unique elements


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def removeDuplicates(self, nums: List[int]) -> int:
        i = 1
        for j in range(1, len(nums)):
            if nums[j] != nums[j - 1]:  # or nums[j] != nums[i - 1]
                nums[i] = nums[j]
                i += 1
        return i
        
```
