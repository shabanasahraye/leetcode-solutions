## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Use the reversal algorithm

Example: nums = [1, 12, -3, 4, 5, 6, 2], k = 3
- right rotate = [5, 6, 2, 1, 12, -3, 4]
- left rotate = [4, 5, 6, 2, 1, 12, -3]

To right-rotate an array, we need to move the last/right k elements to the start/left
- reverse the whole array → moves the last k elements to the beginning in reverse order
    - [1, 12, -3, 4, 5, 6, 2] → [2, 6, 5, 4, -3, 12, 1]
- reverse the first k elements → corrects their order
    - [**2, 6, 5**, 4, -3, 12, 1] → [**5, 6, 2**, 4, -3, 12, 1]
- reverse the remaining n-k elements → corrects their order
    - [5, 6, 2, **4, -3, 12, 1**] → [5, 6, 2, **1, 12, -3, 4**]

To left-rotate an array, we need to move the first/left k elements to the end/right
- reverse the whole array → moves the first k elements to the end in reverse order
    - [1, 12, -3, 4, 5, 6, 2] → [2, 6, 5, 4, -3, 12, 1]
- reverse the last k elements → corrects their order
    - [2, 6, 5, 4, **-3, 12, 1**] → [2, 6, 5, 4, **1, 12, -3**]
- reverse the first remaining n-k elements → corrects their order
    - [**2, 6, 5, 4**, -3, 12, 1] → [4, 5, 6, 2, 1, 12, -3]

For reversing, either use a library function reverse() or define a function to reverse the array using two pointers.


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = size of the input array
- check base cases
    - if n is 0 or k is 0 or equal to the length of the array, the rotation will result in the same array
        - return
- mod k by n to keep k in bounds
- define a helper function for reversing 
    - *reverse(array start, array end)*
        - set l = leftmost index i.e start
        - set r = rightmost index i.e array end position - 1
        - loop until left pointer < right pointer
            - swap the elements at index l and r
            - move pointers i.e increment l and decrement r
- reverse all n elements
- reverse first k elements
- reverse remaining last n-k elements


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(reverse n + reverse k + reverse n - k) → O(n + k + n - k) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n = len(nums)
        if n == 0 or k == 0 or k == n:
            return
        k %= n

        def reverse(start: int, end: int) -> None:
            l, r = start, end - 1
            while l < r:
                nums[l], nums[r] = nums[r], nums[l]
                l += 1
                r -= 1
        
        reverse(0, n)
        reverse(0, k)
        reverse(k, n)
        
```
