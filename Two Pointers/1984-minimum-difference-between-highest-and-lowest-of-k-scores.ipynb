# <center> 1984. Minimum Difference Between Highest and Lowest of K Scores </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-difference-between-highest-and-lowest-of-k-scores/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find a window (subarray) of size k with the minimum difference between its max and min element.

Sort the array because after sorting:
- any contiguous k elements are as close to each other as possible
- the min element is at the start and the max element is at the end, so we can easily find the difference

Use a fixed size sliding window to check every window of size k and find the minimum difference.


## Approach
<!-- Describe your approach to solving the problem. -->
- sort the array
- set res = int max or the last number of the sorted array (array max) to track the minimum difference
- set left pointer l = 0 and right pointer r = k - 1 to track the left and right end of the window of size k
- traverse the sorted array
    - find the difference of the current window and update res with min of current window difference and previous min difference
    - move both pointers to slide the window
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort + sliding window) → O(nlogn + n) → O(nlogn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort) → O(n)


## Code

```python
class Solution:

    def minimumDifference(self, nums: List[int], k: int) -> int:
        nums.sort()
        res = nums[-1]
        l, r = 0, k - 1
        while r < len(nums):
            res = min(res, nums[r] - nums[l])
            l += 1
            r += 1
        return res
```
