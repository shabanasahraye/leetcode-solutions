# <center> 167. Two Sum II - Input Array Is Sorted </center>


## Problem Description
[Click here](https://leetcode.com/problems/two-sum-ii-input-array-is-sorted/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
The array is sorted, so smaller numbers are on the left and greater numbers are on the right.

Use two pointers moving in opposite directions (left and right) to compare the number:
- each time, select a smaller number from the left and a greater number from the right. If their sum is:
    - less than the target → the left number is too small, so move the left pointer
    - greater than the target → the right number is too large, so move the right pointer
    - equal to the target → we found the target sum


## Approach
<!-- Describe your approach to solving the problem. -->
- set l = leftmost position to track lefttmost element
- set r = rightmost position to track rightmost element
- loop until left pointer < right pointer 
    - compute the current sum by adding elements at positions l and r
    - if the current sum is less than the target
        - move the left pointer i.e increment l
    - if the current sum is greater than the target
        - move the right pointer i.e decrement r
    - else current sum is equal to the target
        - return l + 1, r + 1 as the positions of the two numbers that sum to the target


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(two pointer traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        l, r = 0, len(numbers) - 1
        while l < r:
            cur_sum = numbers[l] + numbers[r]
            if cur_sum < target:
                l += 1
            elif cur_sum > target:
                r -= 1
            else:
                return l + 1, r + 1
                
```
