## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
In a palindrome, the left and right characters are the same.

Use two pointers moving in opposite directions (left and right) to compare the characters. If they are not equal, check if they can be made equal by skipping a left or right character. 


## Approach
<!-- Describe your approach to solving the problem. -->
- set left pointer l = 0
- set right pointer r = rightmost position i.e length of string - 1
- traverse the string and check the left and right characters
    - if chars at positions l and r are not equal
        - check if removing the left char or right char makes the string a palindrome, and return the result
            - *for this check, we define a helper function*
            - *nested loops, slicing, or recursion can also be used, slicing will increase space complexity*
    - else left and right chars are equal
        - move pointers to continue checking the string
- if the loop completes without returning false, the string is a valid palindrome, return true


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal × palindrome check) → O(n)
    - *the outer loop can take up to n/2, and the helper can take up to n/2, once a mismatch is found, the outer loop stops, so they do not both fully execute*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1) 


## Code

```python
class Solution:
    
    def validPalindrome(self, s: str) -> bool:
        
        def is_palindrome(l: int, r: int) -> bool:
            while l < r:
                if s[l] != s[r]:
                    return False
                l += 1
                r -= 1
            return True

        l, r = 0, len(s) - 1
        while l < r:
            if s[l] != s[r]:
                return is_palindrome(l, r - 1) or is_palindrome(l + 1, r)
            l += 1
            r -= 1
        return True
        
```
