# <center> 1851. Minimum Interval to Include Each Query </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-interval-to-include-each-query/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a list of intervals and queries. For each query, we need to find the size of the smallest interval that contains it.

interval size = interval end - interval start + 1

Use a min heap to efficiently track the smallest valid interval while processing the queries in sorted order.


## Approach
<!-- Describe your approach to solving the problem. -->
- sort intervals by start time 
- create an empty min heap
- create a hashmap to store the answer (min interval) for each query
    - *the queries are processed in sorted order, so their original order is lost*
    - *the hashmap maps each query to its answer, allowing us to rebuild the result array in the original query order at the end*
    - *another approach is to sort `(query, original_index)` pairs instead of using a hashmap*
- set i = 0 to track the interval index
- traverse the sorted queries
    - *don't do inplace sort because we need the queries original order to build the final result*
    - while there are intervals to process and the current interval starts before or at the current query (interval start <= query)
        - get the start time l and end time r of the current interval
        - push the interval size and end time into the min heap
            - *index 0 = interval size*
            - *index 1 = interval end*
        - increment i
    - while the heap is not empty and the interval ends before the current query (interval end < query)
       - pop interval
    - if heap is not empty after the loop, there's a min interval for the query
        - add the heap top interval size to the hashmap
    - else
        - add -1 for the query in the hashmap
- traverse the original queries and build the result array and return it
    - *for each query, look up its answer in the hashmap and append it to the result array*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort intervals + sort queries + heap push/pop + result array) → O(nlogn + qlogq + nlogn + q) → O(nlogn + qlogq)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort + heap + hashmap + result array) → O(n + n + q + q) → O(n + q)
    - *Python's sort uses O(n) auxiliary space in the worst case*


## Code

```python
class Solution:

    def minInterval(self, intervals: List[List[int]], queries: List[int]) -> List[int]:
        intervals.sort()
        min_heap = []
        res = {}
        i = 0
        for q in sorted(queries):
            while i < len(intervals) and intervals[i][0] <= q:
                l, r = intervals[i]
                heapq.heappush(min_heap, (r - l + 1, r))
                i += 1
            while min_heap and min_heap[0][1] < q:
                heapq.heappop(min_heap)
            res[q] = min_heap[0][0] if min_heap else -1
        return [res[q] for q in queries]
        
```
