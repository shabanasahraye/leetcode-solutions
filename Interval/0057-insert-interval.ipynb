# <center> 57. Insert Interval </center>


## Problem Description
[Click here](https://leetcode.com/problems/insert-interval/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a sorted list of non-overlapping intervals and a new interval. We need to insert the new interval and merge any overlapping intervals.

The intervals are already sorted by start time. Two intervals overlap if:
- current interval start ≤ new interval end
- new interval start ≤ current interval end

Traverse the intervals to find the correct position for the new interval. Merge it with any overlapping intervals.


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if there are no input intervals
        - return an array containing the new interval
    - if the new interval belongs at the beginning or the end
        - if the last interval end < new interval start
            - insert the new interval at the end 
            - return the intervals
        - if the first interval start > new interval end
            - insert the interval at the start 
            - return the intervals 
- set res = an empty array to store the merged intervals
- traverse the intervals <br>
*for each interval in the intervals*
    - if the new interval end < current interval start, e found the insert position and there is no overlap
        - add the new interval to the result array
        - return res + the remaining intervals from the input array
    - if the new interval start > current interval end, we have not found the insert position yet
        - add the current interval to the result array
    -  else there is overlap
        - merge the current interval and new interval by updating the start and end of the new interval
            - *update the start to the minimum of both start times*
            - *update the end to the maximum of both end times*
-  if the loop completes, add the merged new interval to the end of the result array
- return the result array


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(input array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result array) → O(n)


## Code

```python
class Solution:

    def insert(self, intervals: List[List[int]], newInterval: List[int]) -> List[List[int]]:
        if not intervals:
            return [newInterval]
        if intervals[-1][1] < newInterval[0]:
            intervals.append(newInterval)
            return intervals
        if intervals[0][0] > newInterval[1]:
            intervals.insert(0, newInterval)
            return intervals
        
        res = []
        for i, interval in enumerate(intervals):
            if newInterval[1] < interval[0]:
                res.append(newInterval)
                return res + intervals[i:]
            elif newInterval[0] > interval[1]:
                res.append(interval)
            else:
                newInterval[0] = min(newInterval[0], interval[0])
                newInterval[1] = max(newInterval[1], interval[1])
        res.append(newInterval)
        return res
        
```
