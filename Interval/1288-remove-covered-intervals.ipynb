# <center> 1288. Remove Covered Intervals </center>


## Problem Description
[Click here](https://leetcode.com/problems/remove-covered-intervals/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a list of intervals. We need to remove all intervals that are covered by another interval and return the number of remaining intervals.

An interval [a, b] is covered by [c, d] if c ≤ a and b ≤ d.

Sort the intervals by start (left bound). If two intervals have the same start, sort by end (right bound) in descending order.

Traverse the sorted intervals while tracking the largest end seen so far. An interval is not covered if its end > the largest end seen so far; otherwise, it is covered.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to count remaining intervals
- sort interval by start, if the start is same, sort by the highest end in reverse order
- set prev_end = 0 to track the largest end seen so far
- traverse the intervals <br>
*for each interval start and end*
    - if current interval end > prev_end, the interval is not covered
        - increment res
        - update prev_end because the current end is now the largest
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort + traversal) → O(nlogn + n) → O(nlogn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort) → O(n)
    - *Python's sort uses O(n) auxiliary space in the worst case.*


## Code

```python
class Solution:

    def removeCoveredIntervals(self, intervals: List[List[int]]) -> int:
        res = 0
        intervals.sort(key=lambda x: (x[0], -x[1]))
        prev_end = 0
        for _, end in intervals:
            if end > prev_end:
                res += 1
                prev_end = end
        return res
        
```
