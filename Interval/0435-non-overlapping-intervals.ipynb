# <center> 435. Non-overlapping Intervals </center>


## Problem Description
[Click here](https://leetcode.com/problems/non-overlapping-intervals/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a list of intervals. We need to return the minimum number of intervals to remove so that the remaining intervals are non-overlapping.

Two intervals overlap if the current interval's start time < the previous interval's end time.

Sort the intervals by end time. Traverse the sorted intervals and greedily keep the interval that ends first. If the current interval overlaps with the previously kept interval, remove the current interval because it ends later or at the same time.


## Approach
<!-- Describe your approach to solving the problem. -->
- sort the intervals by end time
- set res = 0 to count the intervals to remove
- set prev_end = end time of the first interval
    - *tracks the end time of the last interval kept*
-  traverse the remaining intervals <br>
*start from the second interval*
    - if the current interval start time >= prev_end, there is no overlap
        - update prev_end to the current interval's end time
    - else there is overlap
        - increment res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort + traversal) → O(nlogn + n) → O(nlogn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort) → O(n)
    -  *Python's sort uses O(n) auxiliary space in the worst case*


## Code

```python
class Solution:

    def eraseOverlapIntervals(self, intervals: List[List[int]]) -> int:
        intervals.sort(key= lambda x: x[1])
        res = 0
        prev_end = intervals[0][1]
        for i in range(1, len(intervals)):
            if intervals[i][0] >= prev_end:
                prev_end = intervals[i][1]
            else:
                res += 1
        return res
        
```
