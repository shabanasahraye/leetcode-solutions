# <center> 56. Merge Intervals </center>


## Problem Description
[Click here](https://leetcode.com/problems/merge-intervals/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a list of intervals. We need to merge all overlapping intervals and return the resulting non-overlapping intervals.

Two intervals overlap if the current interval's start time is less than or equal to the previous interval's end time.

Sort the intervals by start time. Traverse the sorted intervals and merge the overlapping ones.


## Approach
<!-- Describe your approach to solving the problem. -->
- sort the intervals by start time
- set res = an array to store the merged intervals
- add the first interval to res
- traverse the intervals
    - if the current interval's start time > the end time of the last interval in res, there is no overlap
        - add the current interval to res
    - else the intervals overlap
        - update the end time of the last interval in res to the maximum of the two end times (current and previous interval)
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort + traversal) → O(nlogn + n) → O(nlogn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort + result array) → O(n + n) → O(n)
    - *Python's sort uses O(n) auxiliary space in the worst case*


## Code

```python
class Solution:

    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort()
        res = [intervals[0]]
        for interval in intervals:
            if interval[0] > res[-1][1]:
                res.append(interval)
            else:
                res[-1][1] = max(res[-1][1], interval[1])
        return res
```
