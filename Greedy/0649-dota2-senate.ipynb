## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a string representing senators from two parties: Dire (D) and Radiant (R). They compete until one party remains.
- senators on the left have an earlier turn to vote
- senators on the left compete against senators on the right
- the winner eliminates the opposing senator from the left and moves to the right for the next round

Use queue because it allows deletion from the left and insertion from the right in O(1) time. <br>
Senators vote from the left (front of the queue), and surviving senators move to the right (back of the queue) for the next round.


## Approach
<!-- Describe your approach to solving the problem. -->
- create a separate queue for each party to store the indices of available senators
- for each senator (character c) in the senators list
    - if char is 'D'
        - add the position to the dire queue
    - else add the position to the radiant que
- loop until one of the queues is empty i.e senators will compete until there is one winner
    - pop both queues to get the senators' position for fight/voting
    - if the dire senator position is smaller, Dire wins because he gets to vote first
        - add the dire senator position to the end of the queue for the next round
        - add the dire senator position to the right of the queue to move him to the next round
    - else move the radiant senator to the next round
- when one queue becomes empty, the remaining party has no opponents and wins, return the party with remaining senators


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal + queue loop) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dire queue + radiant queue) → O(n + n) → O(n)


## Code

```python
class Solution:

    def predictPartyVictory(self, senate: str) -> str:
        dire, radiant = deque(), deque()
        for i, c in enumerate(senate):
            if c == 'D':
                dire.append(i)
            else:
                radiant.append(i)
        while radiant and dire:
            d_turn = dire.popleft()
            r_turn = radiant.popleft()
            if d_turn < r_turn:
                dire.append(d_turn + len(senate))
            else:
                radiant.append(r_turn + len(senate))
        return 'Dire' if dire else 'Radiant'
        
```
