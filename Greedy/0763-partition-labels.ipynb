# <center> 763. Partition Labels </center>


## Problem Description
[Click here](https://leetcode.com/problems/partition-labels/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a string. We need to split it into as many partitions as possible such that each character appears in only one partition.

For example: abcafadcbghtkhg
- partition 1 = abcafadcb ('b' has the farthest last occurrence)
- partition 2 = ghtkhg ('g' has the farthest last occurrence)

This is a greedy problem because instead of trying all possible partitions, we can extend the current partition to the farthest last occurrence seen so far.

We need the last occurrence of each character, so use a hashmap.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = an array to store each partition size
- create a hashmap to store the last occurrence of each character
    - *key = character*
    - *value = last occurring index*
- set left pointer l = 0 to track the start index of a partition
- set right pointer r = 0 to track the last index of a partition
- traverse the string <br>
*for each character c at index i*
    - update the current partition end
        - *r = max(r, last occurrence of the current character)*
    - if the current index reaches the current partition end (i == r), the partition is complete
        - add the partition size to res
            - *partition size = partition end - partition start + 1*
        - move l to the start of the next partition (r + 1)
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(hashmap creation + string traversal) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result array + hashmap) → O(number of partitions + unique characters) → O(26 + 26) → O(1)
    - *26 because the string contains only lowercase English letters*


## Code

```python
class Solution:

    def partitionLabels(self, s: str) -> List[int]:
        res = []
        last_index = {c: i for i, c in enumerate(s)}
        l, r = 0, 0
        for i, c in enumerate(s):
            r = max(r, last_index[c])
            if i == r:
                res.append(r - l + 1)
                l = r + 1
        return res
        
```
