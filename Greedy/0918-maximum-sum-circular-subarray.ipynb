# <center> 918. Maximum Sum Circular Subarray </center>


## Problem Description
[Click here](https://leetcode.com/problems/maximum-sum-circular-subarray/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a circular array. We need to return the maximum subarray sum, which can come from:
- a normal subarray that does not wrap around
- a circular subarray that wraps from the end of the array to the beginning

This is a greedy problem because instead of checking all possible subarrays, we can use Kadane's algorithm to find the required subarray sums in one traversal.

maximum subarray sum = max(normal subarray sum, circular subarray sum)
- normal subarray sum = maximum subarray sum (found using Kadane's algorithm)
- circular subarray sum = total array sum − minimum subarray sum
    - *the minimum subarray is the contiguous subarray with the smallest sum*
    - *subtracting the minimum subarray sum from the total array sum leaves the maximum possible circular subarray*

Therefore, <br>
maximum subarray sum = max(maximum subarray sum, total array sum − minimum subarray sum)


## Approach
<!-- Describe your approach to solving the problem. -->
- set total = 0 to track the total array sum
- set cur_max = 0 to track the maximum subarray sum ending at the current index
- set cur_min = 0 to track the minimum subarray sum ending at the current index
- set global_max = a very small number to track the maximum subarray sum found so far
- set global_min = a very large number to track the minimum subarray sum found so far
- traverse the array  
  *for each number num in the array*
    - add the number to the total array sum
    - update the current maximum and minimum subarray sums using Kadane's algorithm
      - *cur_max = max(cur_max + num, num)*
      - *cur_min = min(cur_min + num, num)*
    - update the maximum and minimum subarray sums found so far
      - *global_max = max(global_max, cur_max)*
      - *global_min = min(global_min, cur_min)*
- if global_max < 0, all numbers are negative, so return global_max
- else return the maximum of:
    - the normal subarray sum (global_max)
    - the circular subarray sum (total - global_min)


## Complexity
- Time complexity: O(array traversal) → O(n)
<!-- Add your time complexity here, e.g. $$O(n)$$ -->

- Space complexity: O(1)
<!-- Add your space complexity here, e.g. $$O(n)$$ -->


## Code

```python
class Solution:

    def maxSubarraySumCircular(self, nums: List[int]) -> int:
        total = 0
        cur_max = 0
        cur_min= 0
        global_max = -math.inf
        global_min = math.inf
        for num in nums:
            total += num
            cur_max = max(cur_max + num, num)
            cur_min= min(cur_min + num, num)
            global_max = max(global_max, cur_max)
            global_min = min(global_min, cur_min)
        return global_max if global_max < 0 else max(global_max, total - global_min)
        
```
