# <center> 12. Integer to Roman </center>

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an integer and need to convert it into a Roman numeral.

This is a greedy problem because at each step we choose the largest possible Roman symbol value that can be subtracted from the remaining number.

Use a hashmap to map Roman symbols to their decimal values. Traverse the symbols from largest to smallest, add each symbol as many times as possible, and reduce the remaining number until it becomes zero.


## Approach
<!-- Describe your approach to solving the problem. -->
- create a hashmap mapping Roman symbols to their values
- set res = empty string to track the result roman number
- traverse the symbols in reverse order to start from the largest value
    - find how many times the current symbol can be used by dividing num by its value
    - if the count is greater than 0
        - add the symbol count times to the res
        -  update num to keep the remaining value
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(hashmap traversal) → O(13) → O(1)
    - *13 because the hashmap contains 13 Roman symbols*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap) → O(13) → O(1)


## Code

```python
class Solution:

    def intToRoman(self, num: int) -> str:
        roman = {'I': 1, 'IV': 4, 'V': 5, 'IX': 9, 'X': 10, 'XL': 40, 'L': 50, 'XC': 90, 'C': 100, 'CD': 400, 'D': 500, 'CM': 900, 'M': 1000}
        res = ''
        for symbol, val in reversed(roman.items()):
            count = num // val
            if count:
                res += symbol * count
                num = num % val
        return res
        
```
