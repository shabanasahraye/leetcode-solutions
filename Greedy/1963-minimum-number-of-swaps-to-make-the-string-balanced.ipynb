# <center> 1963. Minimum Number of Swaps to Make the String Balanced </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-number-of-swaps-to-make-the-string-balanced/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to make the brackets balanced by swapping [ and ].

This is a greedy problem because whenever we find an extra closing bracket, we immediately fix the imbalance instead of waiting until the end.

- **Approach 1: Count Swaps**
    - track the number of opening and closing brackets
    - we need a swap if the closing brackets are more than the opening to fix the imbalance

- **Approach 2: Count Imbalance**
    - track the maximum number of extra closing brackets at any point
    - each swap fixes two extra closing brackets, so the answer is max_close / 2


## Approach 
<!-- Describe your approach to solving the problem. -->
### Approach 1: Count Swaps
- set start = 0 to track the number of opening brackets
- set end = 0 to track the number of closing brackets
- set res = 0 to track the min swaps required to make the string balanced
- traverse the string
    - if the current char is an opening bracket
        - increment start
    - else the char is a closing bracket
        - increment end
        - if there are more closing brackets than opening, we need a swap
            - increment res
            - increment start because the swap brings an opening bracket to the current position
            - decrement end because the swap removes the extra closing bracket from the current position

- return res

### Approach 2: Count Imbalance
- set close = 0 to track the current excess of closing brackets
- set max_close = 0 to track the maximum number of excess closing brackets
- traverse the string
    - decrement close if char is an opening bracket
    - increment close if char is a closing bracket
    - update max_close
        - *we need max_close because close may return to 0 by the end of the traversal*
- return (max_close + 1) / 2
    - *divide by 2 is because each swap fixes two extra closing brackets*
    - *+1 is used to round up when max_close is odd*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Count Swaps Approach: O(string traversal) → O(n)
    - Count Imbalance Approach: O(string traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Count Swaps Approach: O(1)
    - Count Imbalance Approach : O(1)
    

## Code

```python
class Solution:

    def minSwaps(self, s: str) -> int:
        
        # Approach 1: Count Swaps
        start, end, res = 0, 0, 0
        for c in s:
            if c == '[':
                start += 1
            else:
                end += 1
                if end > start:
                    res += 1
                    start += 1
                    end -= 1
        return res

        # Approach 2: Count Imbalance
        close, max_close = 0, 0
        for c in s:
            if c == '[':
                close -= 1
            else:
                close += 1
            max_close = max(close, max_close)
        return (max_close + 1) // 2
        
```
