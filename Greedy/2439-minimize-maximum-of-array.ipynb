## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array of integers where each operation moves 1 from nums[i] to nums[i - 1]. We need to find the minimum possible maximum value of the array after any number of operations. 

- **Approach 1: Binary Search O(n × log(max − min))**
    - the minimum possible max value lies in the range [min(array), max(array)]
    - treat this range as a sorted array and apply binary search
        - if the midpoint is a valid max value, search the left half for a smaller answer, else search the right half
    - time complexity: O(binary search × validation check) → O(log(max − min) × n)
- **Approach 2: Math & Greedy Approach O(n)**
    - the minimized max value of the array is the max ceiling average among all prefixes
        - we use the ceiling average because minimizing the max value means distributing the numbers evenly  by decreasing larger values and increasing smaller ones
    - this is a greedy problem because instead of trying every possible redistribution, we can greedily compute the minimum possible max for each prefix
    - compute the ceiling average of each prefix and return the max
    - Example: [4, 7] <br>
        average = (4 + 7) / 2 = 5.5 <br>
        ceil = 6 <br>
        minimum possible max = 6
    - Formula: <br>
        = ceil (prefix average) <br>
        = ceil(prefix sum / total elements) <br>
        = ceil(prefix sum till index i / (i + 1)) <br>
        = (prefix sum + (i + 1) - 1) // (i + 1) *because ceil(a / b) = (a + b - 1) // b* <br>
        = (prefix sum + i) // (i + 1) 


## Approach
<!-- Describe your approach to solving the problem. -->
### Math & Greedy Approach
- set res = first element to track the minimized max value
    - *we use the first element to skip a step because the first prefix contains only one element, so its ceiling average is the first element itself*
- set pre_sum = first element to track the prefix sum
- traverse the array starting from the second element because the first prefix has already been processed
    - add the current number to pre_sum to extend the prefix
    - compute the current prefix ceiling average and update res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Math & Greedy Approach: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Math & Greedy Approach: O(1)


## Code

```python
class Solution:
    
    def minimizeArrayValue(self, nums: List[int]) -> int:
        res, pre_sum = nums[0], nums[0]
        for i in range(1, len(nums)):
            pre_sum += nums[i]
            res = max(res, (pre_sum + i) // (i + 1))
        return res
        
```
