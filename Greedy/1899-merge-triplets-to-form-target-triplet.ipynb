# <center> 1899. Merge Triplets to Form Target Triplet </center>


## Problem Description
[Click here](https://leetcode.com/problems/merge-triplets-to-form-target-triplet/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array of triplets and a target triplet. We need to determine whether we can merge some triplets to form the target.

This is a greedy problem because any triplet with a value greater than the corresponding target value can never be used, so we discard it immediately.

Among the remaining valid triplets, keep taking the maximum value for each position. If the resulting triplet equals the target triplet, return true else false.

result = (max(first values), max(second values), max(third values))


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = [0, 0, 0] to track the max value collected for each position
- traverse the triplets array <br>
*for each triplet t*
    - if all three values in the current triplet are less than or equal to the corresponding values in the target triplet
      - update each position in res by taking the max value
- return whether res is equal to the target triplet


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal + comparing res and target triplets) → O(n + 3) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result triplet) → O(3) → O(1)


## Code

```python
class Solution:

    def mergeTriplets(self, triplets: List[List[int]], target: List[int]) -> bool:
        res = [0, 0, 0]
        for t in triplets:
            if t[0] <= target[0] and t[1] <= target[1] and t[2] <= target[2]:
                res[0], res[1], res[2] = max(res[0], t[0]), max(res[1], t[1]), max(res[2], t[2])
        return res == target
        
```
