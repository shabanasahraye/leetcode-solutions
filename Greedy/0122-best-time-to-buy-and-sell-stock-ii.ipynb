## Intuition
<!-- Describe your first thoughts on how to solve this problem. --> 
It is the same as [Best Time to Buy and Sell I](https://leetcode.com/problems/best-time-to-buy-and-sell-stock/description/), but here we can make multiple transactions and can buy and sell on the same day. We need to maximize the total profit.

*In Buy and Sell Stock I, we can make only one transaction, so we need to choose the best day to buy and the best day to sell to maximize the profit.*

This is a greedy problem because we take every positive price difference to maximize the total profit without reconsidering previous decisions.

If the current price is greater than the previous day's price, we can make a profit by buying on the previous day and selling on the current day. Traverse the prices and add the difference (profit) whenever the current price is greater than the previous day's price.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the total profit
- traverse the prices for each day (start from the second day)
    - if the current price is greater than the previous day's price
        - add the proce difference (profit) to res
- return res as the total profit


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def maxProfit(self, prices: List[int]) -> int:
        res = 0
        for i in range(1, len(prices)):
            if prices[i] > prices[i-1]:
                res += prices[i] - prices[i-1]
        return res
        
```
