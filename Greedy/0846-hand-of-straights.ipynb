# <center> 846. Hand of Straights </center>


## Problem Description
[Click here](https://leetcode.com/problems/hand-of-straights/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to divide the array into groups of k consecutive numbers.

This is a greedy problem because instead of trying all possible groups, we can start with the smallest available card.

For example, if the smallest card is 2 and the group size is 4, the group must be 2, 3, 4, 5.

Find the smallest available card (number) and check whether the next k − 1 consecutive cards are available. If any required card is missing, return false.

Use:
- a hashmap to track the frequency of each card in O(1)
- a min heap to get the smallest available card in O(1)


## Approach
<!-- Describe your approach to solving the problem. -->
- check if the total number of cards is divisible by the group size
    - if not, return false
- count the frequency of each card using a hashmap
    - *either use a library function or create an empty hashmap and fill it*
    - *key = card*
    - *value = frequency of the card*
- create a min heap from the hashmap keys
    - *get the hashmap keys and convert them to a list*
    - *heapify the list*
- loop until the heap is empty
    - use the smallest available card as the start of the current group
        - *heap top (index 0) contains the smallest card*
    - loop to find the next numbers in the group
        - if the next card is not available
            - return false because a required card is missing
        - else use the card by decrementing its count
        - if its count becomes 0
            - if it is not the smallest card in the heap
                - return false because a required card is missing from the middle of a group
                - *removing a middle card creates a gap. The smallest card will still remain to start a future group, but the missing card makes the consecutive sequence impossible.*
            - else remove it from the heap
- if the loop completes without returning false, return true


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(build hashmap + create heap + process cards) → O(build hashmap + (create list of hashmap keys + heapify) + (while loop × for loop + heap pop)) → O(n + (m + m) + (n + mlogm)) → O(n + m + m + n + mlogm) → O(n + mlogm)
    - *n = total number of cards*
    - *m = total unique cards*
    - *the while loop and for loop together process each card exactly once, so their total work is O(n)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(hashmap + heap) → O(m + m) → O(m)


## Code

```python
class Solution:

    def isNStraightHand(self, hand: List[int], groupSize: int) -> bool:
        if len(hand) % groupSize:
            return False
        card_freq = Counter(hand)
        min_heap = list(card_freq.keys())
        heapq.heapify(min_heap)
        while min_heap:
            start = min_heap[0]
            for i in range(start, start + groupSize):
                if i not in card_freq:
                    return False
                card_freq[i] -= 1
                if card_freq[i] == 0:
                    if i != min_heap[0]:
                        return False
                    heapq.heappop(min_heap)
        return True
        
```
