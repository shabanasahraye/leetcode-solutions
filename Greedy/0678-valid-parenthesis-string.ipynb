## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a string containing three types of characters: '(' , ')' , and '\*'. The '\*' can be treated as an opening bracket, a closing bracket, or an empty string. We need to determine if the string can be made valid by replacing each '\*'.

A string is invalid if it contains unpaired opening or closing brackets.

This is a greedy problem because instead of trying every '*' replacement, we can track the range of possible unmatched opening brackets.

Since '\*' has multiple possible meanings, instead of tracking the exact number of unmatched opening brackets, track a range of possible values:
- low = minimum possible number of unmatched '('
- high = maximum possible number of unmatched '('

For each character:
- update the range based on the possible interpretation
- if high < 0, there are too many closing brackets, so the string is invalid
- after processing all characters, if low == 0, all required opening brackets can be matched


## Approach
<!-- Describe your approach to solving the problem. -->
- set low and high to 0 to track the lower and upper boundaries of the possible unmatched opening brackets
- traverse the string, *for each character c in the string s*
    - if c is an opening bracket
        - increment high and low because we got a new unpaired opening bracket 
    - if c is a closing bracket
        - decrement high because a closing bracket reduces the maximum possible unmatched '(' 
        - decrement low if there are possible unmatched opening brackets left
    - else c is an asterisk
        - for high, treat it as an opening bracket because it increases the maximum possible unmatched '('
        - for low, treat it as a closing bracket if possible because it decreases the minimum possible unmatched '('
    - if high becomes negative, the string is invalid because there are too many closing brackets, return false
- after processing all characters, check if low is 0 because all possible unmatched opening brackets must be paired


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def checkValidString(self, s: str) -> bool:
        low, high = 0, 0
        for c in s:
            if c == "(":
                high += 1
                low += 1
            elif c == ")":
                high -= 1
                if low > 0:
                    low -= 1
            else:
                high += 1
                if low > 0:
                    low -= 1
            if high < 0:
                return False
        return low == 0
        
```
