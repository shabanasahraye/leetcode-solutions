# <center> 1921. Eliminate Maximum Number of Monsters </center>


## Problem Description
[Click here](https://leetcode.com/problems/eliminate-maximum-number-of-monsters/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given the distance and speed of each monster. We need to determine the maximum number of monsters we can eliminate before any monster reaches the city.

This is a greedy problem because the monster that arrives the earliest must always be eliminated first.

Sort the monsters by their arrival time. Traverse the sorted array. In every iteration (minute), eliminate one monster and recharge the weapon. If the current minute (recharge time) is greater than a monster's arrival time, we lose.

arrival time = (distance - 1) // speed <br>
*(actual arrival time = distance / speed, we subtract 1 because monsters must be eliminated before they reach the city)*

Note: we eliminate one monster per minute, and no recharge is required at the start.


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = total number of monsters
- calculate the arrival time of each monster and sort the arrival times
- traverse the sorted arrival times <br>
*for each index (current minute), arrival time in the sorted array*
    - if current minute > monster's arrival time
        - return the current minute as the number of monsters eliminated so far
            - *the current index represents both the current minute (recharge time) and the number of monsters eliminated so far, since one monster is eliminated each minute*
- if the loop completes, we eliminate all monsters, return n


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort + traversal) → O(nlogn + n) → O(nlogn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort + arrival time array) → O(n + n) → O(n)


## Code

```python
class Solution:

    def eliminateMaximum(self, dist: List[int], speed: List[int]) -> int:
        n = len(dist)
        monster_arrival = sorted([(dist[i] - 1) // speed[i] for i in range(n)])
        for minute, arrival in enumerate(monster_arrival):
            if minute > arrival:
                return minute
        return n
        
```
