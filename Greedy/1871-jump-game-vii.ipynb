## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a binary string and a jump range (minJump to maxJump). We need to determine whether we can reach the last index from index 0 using jumps within this range.

An index i is reachable if:
- it contains '0'
- there is a previous reachable index j such that
  - i - maxJump <= j <= i - minJump

This is a dynamic programming problem because the reachability of each index depends on previously reachable indices.

Use a sliding window to track the reachable indices within the valid jump range for the current index.


## Approach
<!-- Describe your approach to solving the problem. -->
- check base case
    - if the last index value is not '0'
        - return false
- initialize a dp array of size n with false
    - *dp[i] represents whether index i is reachable*
- set dp[0] = true because we start at index 0
- set reached = 0 to track the number of reachable indices in the current sliding window
- traverse the string starting from minJump index
    - add the index entering the sliding window
        - reached += dp[i - minJump]
    - remove the index leaving the sliding window
        - if i - maxJump - 1 >= 0
            - subtract dp[i - maxJump - 1]
    - update dp[i]
        - current index is reachable if:
            - there is at least one reachable index in the current sliding window (reached > 0)
            - current character is '0'
- return dp[-1]
    - if true, the last index is reachable, else there is no valid path


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(dp array creation + loop) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(n)


## Code

```python
class Solution:

    def canReach(self, s: str, minJump: int, maxJump: int) -> bool:
        if s[-1] != '0':
            return False
        reached = 0
        dp = [False] * len(s)
        dp[0] = True
        for i in range(minJump, len(s)):
            reached += dp[i - minJump]
            if i - maxJump - 1 >= 0:
                reached -= dp[i - maxJump - 1]
            dp[i] = reached and s[i] == '0'
        return dp[-1] 
        
```
