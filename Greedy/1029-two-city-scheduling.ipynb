# <center> 1029. Two City Scheduling </center>


## Problem Description
[Click here](https://leetcode.com/problems/two-city-scheduling/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to send half people to city A and half to B with minimum cost.

This is a greedy problem because instead of checking all possible assignments, we choose the assignments with the biggest cost advantage first.

Sort the people based on the cost difference between city A and city B. The lower the cost difference, the higher the priority to be sent to city A. Send the first half of the sorted array to city A and the second half to city B.


## Approach
<!-- Describe your approach to solving the problem. -->
- sort the costs based on the difference between city A and city B costs
- sort the costs based on cost difference
    - *this places people with lower cost for city A at the start (left half) and people with lower cost for city B at the end (right half)*
- set res = 0 to track the minimum total cost
- set n = mid of array to get the number of people per city
- traverse the left half array of the sorted array
    - add city A cost of left element (i) and city b cost of right element (i + n) to res
        - *index 0 = city A cost, index 1 = city B cost*
        - *this sends the current person to city A and the corresponding person from the second half to city B*
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(sort + loop) → O(nlogn + n/2) → O(nlogn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(sort) → O(n)
    - *Python's sort uses O(n) auxiliary space in the worst case*


## Code

```python
class Solution:

    def twoCitySchedCost(self, costs: List[List[int]]) -> int:
        costs.sort(key=lambda x: x[0] - x[1])
        res = 0
        n = len(costs) // 2
        for i in range(n):
            res += costs[i][0] + costs[i + n][1]
        return res 
        
```
