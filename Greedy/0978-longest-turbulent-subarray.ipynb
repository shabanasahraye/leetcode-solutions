# <center> 978. Longest Turbulent Subarray </center>


## Problem Description
[Click here](https://leetcode.com/problems/longest-turbulent-subarray/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array of integers. We need to return the length of the longest turbulent subarray.

A subarray is turbulent if the comparison sign alternates between adjacent elements:
- nums[0] < nums[1] > nums[2] < nums[3] ... or
- nums[0] > nums[1] < nums[2] > nums[3] ...

This is a greedy problem because instead of checking all possible subarrays, we extend the current turbulent subarray as long as the comparison sign keeps alternating.

An element nums[i] extends the current turbulent subarray if:
- nums[i-2] < nums[i-1] > nums[i] or
- nums[i-2] > nums[i-1] < nums[i]

Traverse the array and track the current turbulent subarray length. If the comparison sign alternates, extend the current subarray, else start a new one.


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if there is one element
        - return 1
    - if array max and min are same, all elements are the same
        - return 1
- set res to track the longest turbulent subarray length
- set cur to track the current turbulent subarray length
    - initialize both to 2 if the first two elements are different, else 1
- traverse the array starting from the third element
    - if the comparison sign alternates
        - increment cur
    - else if the current and previous elements are different
        - start a new turbulent subarray of length 2
            - cur = 2
    - else
        - reset cur to 1
    - update res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(finding min + finding max + array traversal) → O(n + n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def maxTurbulenceSize(self, arr: List[int]) -> int:
        if len(arr) == 1:
            return 1
        if min(arr) == max(arr):
            return 1
        res, cur = 2, 2 if arr[0] != arr[1] else 1
        for i in range(2, len(arr)):
            if arr[i - 2] < arr[i - 1] > arr[i] or arr[i - 2] > arr[i - 1] < arr[i]:
                cur += 1
            elif arr[i - 1] != arr[i]:
                cur = 2
            else:
                cur = 1
            res = max(res, cur)
        return res
```
