## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array where each element represents the maximum jump length from that index. We need to reach the last index using the minimum number of jumps.

Use a greedy approach.

For each jump, explore all indices in the current reachable range and keep track of the farthest index that can be reached. Once we move beyond the current range, make a jump and update the range to the farthest reachable index.


## Approach
<!-- Describe your approach to solving the problem. -->
- set cur = -1 to track the end of the current jump range
- set next = 0 to track the farthest index reachable from the current jump range
- set res = 0 to count the number of jumps
- set i = 0 to track the current index
- traverse the array while the farthest reachable index has not reached the last index
    - if the current index > cur
        - we have exhausted the current jump range, so make another jump
        - increment res
        - update cur = next
    - update the farthest reachable index
        - next = max(next, i + nums[i])
    - increment i
- return res




## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def jump(self, nums: List[int]) -> int:
        cur, next = -1, 0
        res = i = 0
        while next < len(nums) - 1:
            if i > cur:
                res += 1
                cur = next
            next = max(next, nums[i] + i)
            i += 1
        return res
        
```
