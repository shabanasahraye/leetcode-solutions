## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array where each element represents the maximum jump length from that index. We need to reach the last index using the minimum number of jumps.

This is similar to [Jump Game](https://leetcode.com/problems/jump-game/description/) problem, but here we need to count the minimum number of jumps instead of only checking if the last index is reachable.

Use a greedy approach.

For each jump, explore all indices within the current jump range and keep track of the farthest index that can be reached. Once we move beyond the current range, make another jump and update the range to the farthest reachable index.


## Approach
<!-- Describe your approach to solving the problem. -->
- set cur = -1 to track the end of the current jump range
- set next = 0 to track the farthest index reachable from the current range
- set res = 0 to count the number of jumps
- set i = 0 to track the current index
- traverse the array until the farthest reachable index reaches the last index
    - if the current index is beyond the current jump range (i > cur)
        - make another jump
        - increment res
        - update the current jump range
            - *cur = next*
    - update the farthest reachable index
        - *next = max(next, i + nums[i])*
    - increment i
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def jump(self, nums: List[int]) -> int:
        cur, next = -1, 0
        res = i = 0
        while next < len(nums) - 1:
            if i > cur:
                res += 1
                cur = next
            next = max(next, nums[i] + i)
            i += 1
        return res
        
```
