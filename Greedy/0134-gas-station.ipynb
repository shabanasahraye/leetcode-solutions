## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given gas available at each station and the cost required to travel to the next station. We need to find the starting station from which we can travel through all stations and have enough gas.

Net gas after traveling from station i to i + 1 = gas available at station i - travel cost = gas[i] - cost[i]

This is a greedy problem because instead of trying every starting station, we can discard failed starting points.
- traverse all stations while tracking the current gas balance
- if the balance becomes negative, the current start cannot complete the circuit, so reset the balance and try the next station as the new starting point


## Approach
<!-- Describe your approach to solving the problem. -->
- if total gas is less than total cost, completing the circuit is impossible
    - return -1
    - *we skip this check because calculating total gas and cost requires extra traversal*
- set start = 0 to track the start station
- set net = 0 to track total net gas (total gas gained − total gas cost)
- set cur = 0 to track the current gas balance
- traverse all gas stations
    - calculate the gas difference at the current station
    - add the gas difference to the current gas balance
    - add the gas difference to net to track the total gas balance
    - if current gas balance becomes negative
        - reset cur to 0 
        - update the start station to the next station (i + 1)
- if net gas < 0, return -1 because we can't travel around the circuit, else return the start station


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: (array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: (1)


## Code

```python
class Solution:

    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        # if sum(gas) < sum(cost):
        #     return -1
        start, net, cur = 0, 0, 0                  
        for i in range(len(gas)): 
            diff =  gas[i] - cost[i]
            cur += diff
            net += diff          
            if cur < 0: 
                cur = 0 
                start = i + 1
        return -1 if net < 0 else start  
        
```
