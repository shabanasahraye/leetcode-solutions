# <center> 55. Jump Game </center>


## Problem Description
[Click here](https://leetcode.com/problems/jump-game/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array where each element represents the maximum jump length from that index. We need to check if we can reach the last index.

This is a greedy problem because instead of checking all possible paths, we can track the farthest reachable index.

Traverse the array and track the farthest index we can reach. If we finish traversing all reachable indices and reach the last index, return true, else false.


## Approach
<!-- Describe your approach to solving this problem. -->
- set i = 0 to track the current index
- set reach = 0 to track the farthest index reachable
- traverse the array <br>
*loop while the current index is within the array and reachable (i < len(nums) and i <= reach)*
    - update the farthest reachable index
        - *reach = max(reach, i + nums[i])*
    - increment i
- return whether we reached the end of the array (i == len(nums))


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def canJump(self, nums: List[int]) -> bool:
        i = 0
        reach = 0
        while i < len(nums) and i <= reach:
            reach = max(reach, i + nums[i])
            i += 1
        return i == len(nums)
        
```
