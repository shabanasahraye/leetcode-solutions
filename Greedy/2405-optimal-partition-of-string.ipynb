## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to partition the string so that each partition contains unique characters.

This is a greedy problem because at each step we keep adding characters until a duplicate forces us to start a new partition. 

Traverse the string and make a new partition (substring) whenever you find a duplicate character.

To find duplicates:
1. use a hash set to store chars in the current partition
2. use bit manipulation to track chars, set a bit for each char, if the bit is already set, the char is a duplicate


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Hashset
- set res = 1 to track the minimum number of partitions
    - *1 because the string size is at least 1 i.e at least there is one partition*
- set seen = a hash set to track duplicates i.e to check elements in the current partition
- traverse the string
    - if the current char is in the set (current partition), it is a duplicate char, we need a new partition
        - increment res
        - clear the hash set
    - else it is not present in the current partition
        - add it to the set
- return res

### Approach 2: Bit Manipulation
- set res = 1 to track the minimum number of partitions 
- set seen = 0 to track chars using bits 
    - *the string contains lowercase English letters, so we only need 26 bits*
    - *an integer has at least 32 bits, so we can use one bit for each letters*
    - *for example, bit 0 represents a, bit 1 represents b, ..., bit 25 represents z*
    - *0 because all bits in 0 are unset*
- traverse the string 
    - find the bit index for the current char and set it using left shift operator
        - *bit index = ord(char) - ord('a')*
        - *left shift operator shifts the bits to the left by the given number of positions*
        - *for example, for c: bit index = ord('c') - ord('a') = 99 - 97 = 2, so 1 << 2 = 0001 << 2 = 0100*
    - check if the corresponding bit in seen is already set using AND operator, if yes the char is a duplicate, so we need a new partition
        - *`&` returns 1 only when both corresponding bits are set*
        - *for example: 1100 & 0100 = 0100 because the second bit is set in both*
        - increment res
        - reset seen to 0
    - else set the corresponding bit in seen using OR operator
        - *`|` returns 1 if either corresponding bit is set*
        - *we use `|` to keep the previously set bits and set the new bit*
        - *for example: 1000 | 0100 = 1100*
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Hashset Approach: O(string traversal) → O(n)
    - Bit Manipulation Approach: O(string traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - Hashset Approach: O(hashset) → O(lowercase english letters) → O(26) → O(1)
    - Bit Manipulation Approach: O(1)


## Code

```python
class Solution:

    def partitionString(self, s: str) -> int:

        # Approach 1: Hashset
        # res = 1
        # seen = set()
        # for c in s:
        #     if c in seen:
        #         res += 1
        #         seen.clear()
        #     seen.add(c)
        # return res

        # Approach 2: Bit Manipulation
        res = 1
        seen = 0
        for c in s:
            bit = 1 << (ord(c) - 97)
            if seen & bit: 
                res += 1
                seen = 0
            seen |= bit
        return res
        
```
