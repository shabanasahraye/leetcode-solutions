# <center> 91. Decode Ways </center>


## Problem Description
[Click here](https://leetcode.com/problems/decode-ways/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a string of digits where '1' to '26' represent the letters 'A' to 'Z'. We need to find the total number of valid ways to decode the string.

This is a Dynamic Programming problem because the number of ways to decode the current position depends on previous positions.

At each position, we can:
- decode the current digit alone if it is not 0
- decode the current and previous digits together if they form a valid number from 10 to 26

ways(i) = ways(i - 1) + ways(i - 2) *when the corresponding decodings are valid*


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if there is a leading zero, we can't decode the string, return 0
- set n = string size
- set dp = an array of size n + 1 where dp[i] represents the number of ways to decode the first i characters
    - *index = number of characters considered*
    - *value = ways to decode*
- set dp[0] = dp[1] = 1 
    - *dp[0] = 1 because there is one way to decode an empty string (choose nothing)*
    - *dp[1] = 1 because a single non-zero digit has exactly one valid decoding (the leading zero case has already been handled by returning 0)*
- traverse the DP array from index 2 to n
    - check for the above two conditions and fill the dp array
        - if the last digit is not 0
            - update dp[i]
        - if the last two digits form a valid number from 10 to 26 i.e the second last digit is 1 or 2 and the last digit is less than 7
            - update dp[i]
- return dp[n], the number of ways to decode the entire string


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(n)


## Code

```python
class Solution:

    def numDecodings(self, s: str) -> int:
        if s[0] == '0': 
            return 0
        n = len(s)
        dp = [0] * (n + 1)
        dp[0] = dp[1] = 1          
        for i in range(2, n + 1):
            if s[i - 1] != '0':
                dp[i] += dp[i - 1]
            if s[i - 2] == '1' or s[i - 2] == '2' and s[i - 1] < '7':
                dp[i] += dp[i - 2]
        return dp[n]
        
```
