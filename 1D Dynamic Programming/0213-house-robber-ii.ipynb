## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given houses arranged in a circle, where each house contains some money. We need to find the maximum amount of money we can rob without robbing two adjacent houses.

This is the same as [House Robber I](https://leetcode.com/problems/house-robber/) but in House Robber II, the houses are arranged in a circle, which means the first and last houses are adjacent and cannot be robbed together. 

This is a Dynamic Programming problem because the maximum money up to the current house depends on the previous houses.

Consider two cases:
- rob from the first house to the second last house (exclude the last house)
- rob from the second house to the last house (exclude the first house)

Use the House Robber I solution for both cases and return the maximum of the two results.


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = total number of houses
- check base cases
    - if there is only one house
        - return its value
    - if there are two houses
        - return the maximum value
- define a helper function that finds the maximum money that can be robbed within a given range using the [House Robber I Approach](https://leetcode.com/problems/house-robber/solutions/3578630/o-n-time-o-1-space-solution-explained/)
    - initialize rob1 and rob2
    - traverse the given range
        - for each house
            - find the maximum money by either:
                - robbing the current house (current house money + rob1) or
                - skipping the current house (rob2)
            - update rob1 and rob2 for the next house
    - return rob2
- return the maximum of the two cases:
    - robbing houses from index 0 to n - 2
    - robbing houses from index 1 to n - 1


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(House Robber I on case i + House Robber I on case ii) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        if n == 1 or n == 2:
            return max(nums)
        
        def helper(l: int, r: int) -> int:
            rob1, rob2 = 0, 0
            for i in range(l, r):
                res = max(nums[i] + rob1, rob2)
                rob1 = rob2
                rob2 = res
            return res
            
        return max(helper(0, n - 1), helper(1, n))
        
```
