# <center> 740. Delete and Earn </center>


## Problem Description
[Click here](https://leetcode.com/problems/delete-and-earn/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array of numbers. If we choose a number nums[i]:
- we earn points from all its duplicates
- we cannot choose its adjacent values nums[i] - 1 and nums[i] + 1

This problem is similar to the [House Robber](https://leetcode.com/problems/house-robber/description) problem because choosing one value prevents us from choosing neighboring values.

This is a Dynamic Programming problem because the maximum points for the current number depend on the maximum points from previous numbers.

For each number, we have two choices:
- take the current number and add its points
- skip the current number and keep the previous maximum points

First, combine all occurrences of the same number into total points, then use dynamic programming to find the maximum points.


## Approach
<!-- Describe your approach to solving the problem. -->
- set dp = an array of size max(nums) + 1 with zeroes
    - *dp[i] represents the total points earned by choosing number i*
    - *for example: nums = [2,2,3,3,3,4], dp = [0,0,4,9,4]*
- for each number n in nums
    - add n to dp[n] because each occurrence contributes n points
- set prev = 0 to track the maximum points before the previous number
- set cur = 0 to track the maximum points up to the current number
- traverse the dp array
    - for each value
        - store the current cur value before updating
        - update cur with the maximum of the two choices:
            - take the current number and add its points to the previous maximum (prev + n)
            - skip the current number and keep the current maximum (cur)
        - update prev with the old cur value
            - *prev should store the old cur value because it represents the maximum points before the previous number*
- return cur as the maximum points we can earn


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(input traversal + dp array traversal) → O(n + max(nums)) → O(n + M)
    - *M = max(nums)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(max(nums)) → O(M)


## Code

```python
class Solution:

    def deleteAndEarn(self, nums: List[int]) -> int:
        dp = [0] * (max(nums) + 1)  
        for n in nums:
            dp[n] += n
        prev = cur = 0
        for n in dp:
            cur, prev = max(cur, prev + n), cur
        return cur
        
```
