# <center> 691. Stickers to Spell Word </center>


## Problem Description
[Click here](https://leetcode.com/problems/stickers-to-spell-word/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a list of stickers (words) and a target string. We have to find the minimum number of words required to create the target string using some or all characters from the given words.

This problem is similar to [Word Break](https://leetcode.com/problems/word-break/description/), but here we can use individual characters from a word instead of using the complete word.

This is a Dynamic Programming problem because the minimum words needed for the current state depends on the minimum words needed for previous states.

To find the minimum words required to create the target string, we compute the minimum words required to form every subset of target characters. The order of characters in a word does not matter because we can use individual characters from a word and rearrange them.

For example words = ['with', 'example', 'science'], target string = 'thehat'
- we need to form all characters of the target string: t, h, e, h, a, t
- use the word 'with' to get the chars t and h
    - remaining chars = e, h, a, t
- use the word 'example' to get the chars e and a 
    - remaining chars = h, t
- we can't use the word 'science' because it doesn't contain any required chars
- using the word 'with' to get the chars h and t
    - remaining chars = none
- total words used = 'with' and 'example' = 2


## Approach
<!-- Describe your approach to solving the problem. -->
- set subsets = total subsets (bitmasks) of target characters = $2^{target\:string\:size}$
- set dp = an array of size equal to total subsets with default value of a very large number i.e float infinity     
    - *dp[mask] represents the minimum stickers required to form the characters represented by mask*
- set dp[0] to 0 because no sticker is required to spell an empty string
- create a hashmap to store the target string characters and indices 
    - *key = character*
    - *value = list of indices of the character occurrences*
    - *we are using a hashmap to avoid the loop to check if the sticker character is present in the target string*
- traverse all the subsets <br>
*for each subset i*
    - skip if we can't reach that subset from the starting subset
    - traverse the stickers to find stickers that can be used to spell the current subset <br>
    *for each sticker in the stickers list*
        - set cur = current subset i
        - for each character in the current sticker
            - if the character is present in the target string
                - loop over its indices in the target string
                    - if the current state doesn't contain that character occurrence (i.e. the corresponding bit is not set)
                        - mark that character position as matched by setting its corresponding bit to 1
                        - break the loop
        - update dp[cur] with the minimum number of required stickers 
- if dp[-1] is not float infinity, we can create the target string, return dp[-1], else return -1
    - *the last state (all bits set) represents the complete target string* <br>


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(target map creation + subset traversal × stickers × sticker length × target length) → O(m + $2^m$ × n × k × m) → O($2^m$ × n × k × m) → O($2^m$ × n × 10 × 15) → O($2^m$ × n)
    - *m = target length (≤ 15)*
    - *n = number of stickers*
    - *k = maximum sticker length (≤ 10)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array + target hashmap) → O($2^m$ + m) → O($2^m$)


## Code

```python
class Solution:

    def minStickers(self, stickers: List[str], target: str) -> int:
        subsets = 1 << len(target)
        dp = [float('inf')] * subsets
        dp[0] = 0
        target_map = defaultdict(list)
        for i, c in enumerate(target):
            target_map[c].append(i)
        for i in range(subsets):
            if dp[i] == float('inf'):
                continue
            for sticker in stickers:
                cur = i
                for c in sticker:
                    if c in target_map:
                        for j in target_map[c]:
                            if not (cur >> j & 1):
                                cur |= 1 << j 
                                break
                dp[cur] = min(dp[cur], dp[i] + 1)
        return dp[-1] if dp[-1] != float('inf') else -1
        
```
