## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given n stairs and can climb either 1 or 2 steps at a time. We need to find the total number of different ways to reach the top.

This is a Dynamic Programming problem because the answer for the current step depends on the answers to previous steps.

To reach step n, we can come from:
- step n - 1 by taking 1 step or
- step n - 2 by taking 2 steps

ways(n) = ways(n - 1) + ways(n - 2)

This follows the Fibonacci pattern because we need the previous two values to compute the next.


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if n is less than or equal to 3 
        - return n because:
            - *n = 1, fib(n+1) = fib(2) = 1 way to reach the top*
            - *n = 2, fib(2+1) = fib(3) = 2 ways to reach the top*
            - *n = 3, fib(3+1) = fib(4) = 3 ways to reach the top*
- initialize two variables n1 and n2 to track the previous two numbers
    - *n1 = 2 and n2 = 3 (because we start from fib4)*
- start from Fibonacci number 4 and loop until n
    - compute the next Fibonacci number by adding n1 and n2 
    - update n1, n2
        - *(res is for clarity, we can skip res and directly update n1 and n2 i.e n1, n2 = n2, n1 + n2)*
- return res *(or n2 if we directly update the numbers)*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def climbStairs(self, n: int) -> int:
        if n <= 3:
            return n
        n1 = 2
        n2 = 3
        for _ in range(4, n + 1):
            res = n1 + n2
            n1 = n2
            n2 = res
        return res
        
```
