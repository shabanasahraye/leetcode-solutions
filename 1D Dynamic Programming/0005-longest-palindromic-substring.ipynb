## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a string and need to find its longest palindromic substring.

In a palindrome, the left and right characters are the same.

- **Approach 1: Dynamic Programming (O(n²) space)**
    - store whether every substring is a palindrome in a DP table
    - A substring is a palindrome if its first and last characters are the same and the inner substring is also a palindrome
    - this avoids recomputing palindrome checks but requires O(n²) space

- **Approach 2: Expand Around Center (O(1) space)**
    - every palindrome is centered around either:
        - one character (odd-length palindrome), e.g. "aba"
        - two characters (even-length palindrome), e.g. "abba"
    - For each character, expand outward from both possible centers using two pointers and keep track of the longest palindrome found


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Dynammic Programming
- set n = length of the string
- if the string size is 1
    - return the input string
- set dp = a 2D array of size n × n
    - *dp[i][j] = True if the substring from index i to j is a palindrome*
- set start = 0 and max_len = 1 to track the longest palindrome
- mark every single character as a palindrome
    - *set dp[i][i] = True*
- check all substring lengths from 2 to n
    - set j = ending index of the current substring
    - if the first and last characters are the same
        - if the substring length is 2 or the inner substring is a palindrome
            - mark the current substring as a palindrome
            - if the current palindrome is longer than the longest palindrome found
                - update start and max_len 
- return the longest palindrome using start and max_len

### Approach 2: Expand Around Center
- set n = length of the string
- check base cases
    - if the string size is 1, it is a palindrome
        - return the input string as the longest palindrome
- set two variables start = 0, end = 0 to track the start and end indices of the longest palindromic substring
- traverse the string
    - break the loop if the current substring length (n - i) is less than the max length substring we have found i.e the remaining characters cannot form a longer palindrome
    - check odd-length palindromes
        - set l = i and r = i to track the left and right of an odd-length palindrome and use the current character as its center
        - loop until l and r are in bounds and the left and right characters are the same 
            - if the current palindrome (r - l) is longer than the current max length (end - start)
                - update the start and end
            - update l and r to expand to the left and right by one character
    - check even-length palindromes
        - set l = i and r = i + 1 to track the left and right of an even-length palindrome and use the current and next characters as its center
        - loop until l and r are in bounds and the left and right characters are the same 
            - if the current palindrome (r - l) is longer than the current max length (end - start)
                - update the start and end
            - update l and r to expand to the left and right by one character
- return the max length palindrome using the start and end indices


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Approach 1:  O(DP array initialization + DP traversal + slicing) → O(n² + n + n) → O(n²)
        - *the DP array has n² entries and each is computed once*
    - Approach 2: O(string traversal × (odd palindrome expansion + even palindrome expansion) + slicing) → O(n × (n + n) + n) → O(n² + n) → O(n²)
        - *there are n possible centers*
        - *each center can expand up to n characters in the worst case*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Approach 1: O(DP array) → O(n²)
    - Approach 2: O(1)

```python
class Solution:

    def longestPalindrome(self, s: str) -> str:

        ### Approach 1: Dynammic Programming
        # n = len(s)
        # if n == 1:
        #     return s
        # dp = [[False] * n for _ in range(n)]
        # start = 0
        # max_len = 1
        # for i in range(n):
        #     dp[i][i] = True
        # for length in range(2, n + 1):
        #     for i in range(n - length + 1):
        #         j = i + length - 1
        #         if s[i] == s[j] and (length == 2 or dp[i + 1][j - 1]):
        #             dp[i][j] = True
        #             if length > max_len:
        #                 start = i
        #                 max_len = length
        # return s[start:start + max_len]

        
        ### Approach 2: Expand Around Center
        n = len(s)
        if n == 1:
            return s
        start, end = 0, 0
        for i in range(n):
            # possibility of max length substring
            if n - i < (end - start) / 2:
                break
            # odd length
            l, r = i, i
            while l >= 0 and r < n and s[l] == s[r]:
                if r - l > end - start:
                    start = l
                    end = r
                l -= 1
                r += 1
            # even length
            l, r = i, i + 1
            while l >= 0 and r < n and s[l] == s[r]:
                if r - l > end - start:
                    start = l
                    end = r
                l -= 1
                r += 1
        return s[start:end+1]
        
```
