## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an integer n and need to find the minimum number of perfect square numbers whose sum equals n.

- **Approach 1: Dynamic Programming**
    - this is a Dynamic Programming problem because the minimum number of perfect squares for a number depends on smaller numbers
    - for every number i, try every perfect square j² ≤ i and use the previously computed answer for i - j²

- **Approach 2: Lagrange's Four-Square Theorem**
    - this theorem states that every positive integer can be represented as the sum of at most four perfect squares
    - the theorem steps are:
        - divide the number by 4 while it is divisible by 4
        - if dividing the reduced number by 8 leaves a remainder of 7, the answer is 4 i.e the number is a sum of four square numbers
        - else if the number can be written as the sum of one or two perfect squares, the answer is 1 or 2
        - else the answer is 3


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Dynamic Programming
- check base cases
    - if n is a perfect square for example 4, 9, 16 etc
        - return 1
- set dp = an array of size n + 1 with a default max value (use n or max integer) 
    - *dp[i] represents the minimum number of perfect squares whose sum is i*
- set dp[0] = 0 and dp[1] = 1 because 0 requires 0 perfect squares and 1 requires 1 perfect square
- for each sum i in range 2 to n
    - for each perfect square j in range 1 to sqrt(i)
        - update dp[i] by taking the minimum of its current value and dp[i - j²] + 1
            - *if we can form i − j², adding the perfect square j² forms i* 
- return number of perfect squares that sum to n i.e dp[n]
    
### Approach 2: Lagrange's Four-Square Theorem
- check base cases
- if n is a perfect square
    - return 1
- apply Lagrange's Four-Square Theorem
    - reduce n by dividing it by 4 while it is divisible by 4
    - if n % 8 == 7
        - return 4
    - check if n can be represented as the sum of two perfect squares, the answer is 1 or 2
        - *we have already handled the 1 in the base case, check for 2 using a loop*
        - loop through possible values of i
            - if n - i² is also a perfect square
                - return 2
    - if none of the above cases apply
        - return 3


## Complexity

- Time complexity:
    - Approach 1: O(perfect square check + outer loop × inner loop) → O(1 + n × √n) → O(n√n)
        - *the perfect square check uses the built-in sqrt() function, which is treated as O(1)*

        - *for each outer iteration i, the inner loop runs up to √i times*
    - Approach 2: O(perfect square check + divide by 4 loop + two square check loop) → O(1 + log₄n + √n) → O(√n)
        - *the loop checks all possible values of i from 0 to √n*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - Approach 1: O(dp array) → O(n)
    - Approach 2: O(1)
