## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an integer n and need to break it into the sum of at least two positive integers such that their product is maximum.

- **Approach 1: Dynamic Programming O(n²)**
    - this is a Dynamic Programming problem because the maximum product for the current integer depends on the maximum products of smaller integers
    - if we split an integer i into two parts j and i - j: <br>
      max product for i = max(max(j, dp[j]) × max(i - j, dp[i - j]))
    - compute the maximum product for every integer from 2 to n
- **Approach 2: Mathematical Observation O(log(n/3))**
    - examples:
        - n = 10, max product = 3 × 3 × 4
        - n = 11, max product = 3 × 3 × 3 × 2
    - if we observe the pattern, the maximum product is obtained by repeatedly breaking the integer into parts of 3 while the remaining value is greater than 4
    - according to the maxima-minima concept, the maximum product is achieved when the integer is split into equal parts, where the optimal part size is e ≈ 2.718 *(natural logarithm base)*
    - since 2 < e < 3, the best integer part is 3
    - therefore, keep breaking the number into 3s until the remaining value is 2 or 4


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Dynammic Programming
- set dp = an array of size n + 1 with zeroes
    - *dp[i] represents the maximum product obtainable by breaking integer i*
- set dp[1] = 1
    - *the maximum product obtainable from 1 is 1*
- for each integer from 2 to n
    - try every possible first split
        - split into j and i - j
        - update dp[i]
            - *dp[i] = max(dp[i], max(j, dp[j]) × max(i - j, dp[i - j]))*
- return dp[n] as the maximum product

### Approach 2:  Mathematical Observation
- check base cases
    - if n <= 3, the max prdouct is n - 1 
        - return n - 1
        - *because the integer must be broken into at least two positive integers*
        - *n = 2 break into 1 + 1, max product = 1 × 1 = 1*
        - *n = 3 break into 2 + 1, max product = 2 × 1 = 2*
- find the last remaining part, multiply it by the power of 3 and return the result
    - if n is divisible by 3, there's no last part
        - return $3^{n/3}$
        - *e.g. n = 6 → 3 + 3*
    - if n mod 3 is 1, the last part is 4
        - return $3^{(n/3)-1}$ × 4
        - *e.g. n = 10 → 3 + 3 + 4*
    - else n mod 3 is 2, the last part is 2
        - return $3^{n/3}$ × 2
        - *e.g. n = 11 → 3 + 3 + 3 + 2*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Approach 1: O(outer loop × inner loop) → O(n × n) → O(n²)
    - Approach 2: O(power) → O(log(exponent)) → O(log(n/3))
        - *the built-in `**` operator computes powers in O(log k) time, where k = n/3*
        - *with repeated multiplication, it would be O(n/3) instead*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Approach 1: O(dp array) → O(n)
    - Approach 2: O(1)
