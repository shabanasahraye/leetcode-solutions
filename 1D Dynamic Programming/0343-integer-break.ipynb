# <center> 343. Integer Break </center>


## Problem Description
[Click here](https://leetcode.com/problems/integer-break/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an integer n and need to break it into the sum of at least two positive integers such that their product is maximum.

- **Approach 1: O(n/3)**
    - examples:
        - n = 10, max product = 3 x 3 x 4 <br>
        - n = 11, max product = 3 x 3 x 3 x 2 <br>
    - If we observe the pattern, the max product is obtained by repeatedly breaking the integer into parts of 3 while the remaining value is greater than 4
    - keep the last part as 2, 3, or 4

- **Approach 2: O(log(n/3))**
    - according to the maxima-minima concept, the maximum product is achieved when the integer is split into equal parts, the optimal part size is e ≈ 2.718 *(natural logarithm base)*
    - since 2 < e < 3, the best integer part is 3
    - therefore, keep breaking the number into 3s until the remaining value is 2 or 4


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if n <= 3, the max prdouct is n - 1 
        - return n - 1
        - *because the integer must be broken into at least two positive integers*
        - *n = 2 break into 1 + 1, max product = 1 x 1 = 1*
        - *n = 3 break into 2 + 1, max product = 2 x 1 = 2*
- find the last remaining part, multiply it by the power of 3 and return the result
    - if n is divisible by 3, there's no last part
        - return $3^{n/3}$
        - *e.g. n = 6 → 3 + 3*
    - if n mod 3 is 1, the last part is 4
        - return $3^{(n/3)-1}$ × 4
        - *e.g. n = 10 → 3 + 3 + 4*
    - else n mod 3 is 2, the last part is 2
        - return $3^{n/3}$ × 2
        - *e.g. n = 11 → 3 + 3 + 3 + 2*


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(power) → O(log(exponent)) → O(log(n/3))
    - *the built-in `**` operator computes powers in O(log k) time, where k = n/3*
    - *with repeated multiplication, it would be O(n/3) instead*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def integerBreak(self, n: int) -> int:
        if n <= 3:
            return n - 1
        if n % 3 == 0:
            return 3 ** (n // 3)
        if n % 3 == 1:
            return 3 ** ((n // 3) - 1) * 4
        return 3 ** (n // 3) * 2
        
```
