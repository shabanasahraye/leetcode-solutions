## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array and need to find the length of the longest increasing subsequence (LIS).

This is a Dynamic Programming problem because the longest increasing subsequence ending at the current index depends on the longest increasing subsequences ending at previous indices.

LIS at index i = 1 (if index i element is greater) + max(LIS at previous indices with smaller values) <br>
LIS(i) = 1 + max(LIS(j)) where j < i and nums[j] < nums[i]

- **Approach 1: Dynamic Programming**
    - compute the LIS ending at each index and return the maximum length

- **Approach 2: Dynamic Programming with Binary Search**
    - instead of storing the LIS for every index, store the smallest possible ending value for increasing subsequences of each length
    - use binary search to update this list efficiently


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Dynamic Programming
- set dp = an array of size n with default value of ones 
    - *dp[i] represents the length of the longest increasing subsequence ending at index i*
    - *initialize the array with ones because the length of a LIS ending at each index is at least 1*
- for i in range 1 to n i.e 2nd element to last <br>
*start from index 1 because the LIS ending at index 0 is already 1*
    - for j in range 0 to i i.e first element to ith element
        - if the ith element is greater and length of current LIS of i <= LIS of j
            - update dp[i] using the formula above
- return the max of all LIS lengths i.e max(dp)

### Approach 2: Dynamic Programming With Binary Search
- set dp = an empty array
    - *dp[i] stores the smallest possible ending value of an increasing subsequence of length i + 1*
- traverse the input array
    - if the current number is greater than the last value in dp
        - append it to dp
    - else
        - use binary search to find the first value in dp greater than or equal to the current number
        - replace it with the current number
- return the size of dp


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - Approach 1: O(outer loop × inner loop) → O(n × n) → O(n²)
        - *for each index, the inner loop checks all previous elements (worst case: n comparisons)*
    - Approach 2: (array traversal × binary search) → O(n × logn) → O(nlogn)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Approach 1: O(dp array) → O(n)
    - Approach 2: O(dp array) → O(n)
