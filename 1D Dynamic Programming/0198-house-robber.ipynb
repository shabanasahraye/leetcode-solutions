# <center> 198. House Robber </center>


## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array where each element represents the amount of money in a house. We need to find the maximum amount of money we can rob without robbing two adjacent houses.

This is a Dynamic Programming problem because the maximum money up to the current house depends on the previous houses.

For each house, we have two choices:
- rob the current house i → add its money to the maximum money up to house i - 2
- don't rob the current house i → keep the maximum money up to house i - 1

Traverse the houses, and choose the option with the max value.
- max(option 1, option 2)
- max(rob(i-2) + rob(i), rob(i-1))


## Approach
<!-- Describe your approach to solving the problem. -->
- initialize two variables rob1 and rob2 to represent the two options
    - rob1 = maximum money up to house i - 2
    - rob2 = maximum money up to house i - 1
- traverse the houses
    - for each house
        - find the maximum money by:
            - robbing the current house (current house money + rob1) or
            - skipping the current house (rob2)
        - update rob1 and rob2
- return res or rob2


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def rob(self, nums: List[int]) -> int:
        if len(nums) == 1:
            return nums[0]
        rob1, rob2 = 0, 0
        for n in nums:
            res = max(n + rob1, rob2)
            rob1 = rob2
            rob2 = res
        return res
        
```
