## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a triangle and need to find the minimum path sum from top to bottom by moving to the lower-left or lower-right element.

This is a Dynamic Programming problem because the minimum path sum from a cell depends on the minimum path sums of the two cells directly below it.

Process the triangle from bottom to top. Starting from the last row, compute the minimum path sum for each cell by adding its value to the smaller of its two possible downward paths.


## Approach
<!-- Describe your approach to solving the problem. -->
- set dp = last row where dp[j] represents the minimum path sum from the current cell to the bottom
    - *index = column number*
    - *value = minimum path sum from that position to the bottom*
- traverse the rows in reverse <br>
*for each row starting from the second last row*
    - traverse the columns <br>
    *for each col in the row*
        - find the minimum path sum through the lower-left child
        - find the minimum path sum through the lower-right child
        - update current cell path sum with the min of the two child path sums
- return dp[0], which stores the minimum path sum from the top to the bottom


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(outer loop x inner loop) → O(total cells in triangle) → O(1 + 2 + 3 + ... + n) → O(n(n+1)/2) → O(n²)
    - *the input array is a triangle array i.e each row contains a different number of columns (elements)* 
    - *row i contains i + 1 elements*
    - *the total number of elements is 1 + 2 + ... + n = n(n+1)/2*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(last row size) → O(n)


## Code

```python
class Solution:

    def minimumTotal(self, triangle: List[List[int]]) -> int:
        dp = triangle[-1]
        for i in range(len(triangle) - 2, -1, -1):
            for j in range(i + 1):
                # dp[j] = triangle[i][j] + min(dp[j], dp[j + 1])
                lower_left = triangle[i][j] + dp[j]
                lower_right = triangle[i][j] + dp[j + 1]
                dp[j] = min(lower_left, lower_right)
        return dp[0]
        
```
