## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a list of coins and can use each coin any number of times. We need to find the minimum number of coins required to make the target amount.

This is a Dynamic Programming problem because the minimum coins needed for the current amount depend on the minimum coins needed for smaller amounts.

min coins required for current amount = 1 + min coins required for the remaining amount <br>
*(1 because we have chosen a coin)*

For example: coins = [1, 3, 4], target amount = 7
- if we choose coin 1
    - remaining amount = 7 - 1 = 6
    - min coins for amount 7 = 1 + min coins for amount 6
- if we choose coin 3
    - remaining amount = 7 - 3 = 4
    - min coins for amount 7 = 1 + min coins for amount 4
- if we choose coin 4
    - remaining amount = 7 - 4 = 3
    - min coins for amount 7 = 1 + min coins for amount 3

Compute the minimum coins for every amount from 1 to the target amount by trying every coin.


## Approach
<!-- Describe your approach to solving the problem. -->
- set m = a value greater than the target amount i.e amount + 1
- set dp = an array of size m where dp[i] represents the minimum number of coins required to make amount i
    - *index = amount*
    - *value = min required coins*
 - *set value at index 0 to 0 because 0 coins are required for amount 0*
 - *set all other values to m (a value greater than the target amount) to represent an impossible state initially*
- for each amount in the range 1 to m
    - for each coin
        - if the remaining amount is valid (a - coin ≥ 0)
            - update the minimum coins for the current amount
                - *min coins for current amount = min(current value, 1 + min coins for remaining amount)*
- if dp[target amount] is still m, the target amount cannot be formed, return -1, else return dp[target amount]


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(amount x coins) → O(m x n)
    - *for each amount from 1 to amount, we try every coin once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(amount) → O(m)


## Code

```python
class Solution:

    def coinChange(self, coins: List[int], amount: int) -> int:
        m = amount + 1
        dp = [m] * (m)
        dp[0] = 0
        for a in range(1, m):
            for c in coins:
                if a - c >= 0:
                    dp[a] = min(dp[a], 1 + dp[a - c])
        return -1 if dp[amount] == m else dp[amount]
        
```
