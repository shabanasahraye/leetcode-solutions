## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given the cost of each step and need to find the minimum cost to reach the top of the stairs.

At each step, we have two choices:
- take 1 step
- take 2 steps

This is a Dynamic Programming problem because the minimum cost to reach the current step depends on the minimum cost to reach the previous two steps.

For each step:
- choose the minimum cost between:
    - taking one step from the previous step
    - taking two steps from the step before the previous step

Traverse the cost array and compare the cost of the two choices at each step to find the minimum cost. <br>
Use two variables to store the minimum cost of the previous two steps instead of storing all values in a DP array.


## Approach
<!-- Describe your approach to solving the problem. -->
- set n1 = 0 and n2 = 0 to represent the minimum cost of the previous two steps
    - *n1 = min cost to reach the step before the previous step*
    - *n2 = min cost to reach the previous step*
- start from index 2 because the first two steps can be reached without paying extra cost
    - for each step:
        - calculate the minimum cost to reach the current step by choosing:
            - *one step: n2 + cost[i - 1]*
            - *two steps: n1 + cost[i - 2]*
        - update n1 and n2
            - *n1 = n2 (previous step becomes the step before previous)*
            - *n2 = current minimum cost*
- return n2 as the minimum cost to reach the top


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

     def minCostClimbingStairs(self, cost: List[int]) -> int:
        n1 = n2 = 0
        for i in range(2, len(cost) + 1):
            res = min(n1 + cost[i - 2], n2 + cost[i - 1])
            n1 = n2
            n2 = res
        return n2
        
```
