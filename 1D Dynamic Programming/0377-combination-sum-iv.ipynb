## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array of distinct numbers and a target value. We need to find the total number of ordered combinations that sum to the target.

This is a Dynamic Programming problem because the number of combinations for the current sum depends on the number of combinations for smaller sums.

combinations(current sum) = sum of combinations(current sum - number) <br>
C(x) = Σ C(x - i), where i ∈ nums and i ≤ x

For example: target = 4, nums = [1, 2, 3] <br>
C(4) = C(3) + C(2) + C(1)

Compute the number of combinations for every sum from 0 to the target.


## Approach
<!-- Describe your approach to solving the problem. -->
- set dp = an array of size target + 1 with zeroes
    - *dp[i] represents the number of combinations that sum to i*
- set dp[0] = 1
    - *there is one way to make sum 0: choose no numbers*
- for each sum from 1 to target
    - for each number in nums
        - if the current number is less than or equal to the current sum, we can select the current number
            - add the combinations for the remaining sum
                - *dp[current sum] += dp[current sum - current number]*
- return dp[target] as the number of ordered combinations that sum to the target


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(outer loop × inner loop) → O(target × nums array) → O(t × n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(target) → O(t)


## Code

```python
class Solution:

    def combinationSum4(self, nums: List[int], target: int) -> int:
        dp = [0] * (target + 1)
        dp[0] = 1
        for i in range(1, target + 1):
            for n in nums:
                if n <= i:
                    dp[i] += dp[i - n]
        return dp[-1]  
        
```
