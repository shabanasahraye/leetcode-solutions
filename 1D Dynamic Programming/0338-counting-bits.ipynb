## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an integer n and we need to find the numbers in the range 0 to n and return the number of set bits (1's) in their binary form. 

- **Approach 1: Brute Force**
    - generate the binary representation of each number and count its set bits
    
- **Approach 2: Dynamic Programming**
    - this is a Dynamic Programming problem because the number of set bits in the current number depends on the result of a previously computed number
    - set bits in x = set bits in (x / 2) + (1 if x is odd else 0)
        - *we add 1 if x is odd because its rightmost bit (LSB) is 1, dividing x by 2 (or right shifting by 1) removes the rightmost bit*


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: Dynamic Programming
- set res = an array of size n + 1 with zeroes
    - *res[i] stores the number of set bits in i*
- for each number in the range 1 to n <br>
*start the loop from 1 because number of set bits in 0 is 0*
    - compute its number of set bits using the formula and store it in res
        - *set bits in num = set bits in (num / 2) + rightmost bit (LSB)*
            - *num & 1 gives the rightmost bit (1 if odd, 0 if even)*
            - *you can also use num % 2 to get the rightmost bit*
            - *num / 2 can also be obtained using right shift (num >> 1)*
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Approach 2: O(creating result array + array traversal) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Approach 2: O(result array) → O(n)


## Code

```python
class Solution:

    def countBits(self, n: int) -> List[int]:

        # Approach 2: Dynamic Programming
        res = [0] * (n + 1)
        for i in range(1, n + 1):
            res[i] = res[i // 2] + (i & 1)
        return res
        
```
