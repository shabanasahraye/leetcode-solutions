# <center> 53. Maximum Subarray </center>


## Problem Description
[Click here](https://leetcode.com/problems/maximum-subarray/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array that may contain positive and negative numbers. We need to find the contiguous subarray with the maximum sum.

This is a Dynamic Programming problem because the maximum subarray sum ending at the current position depends on the maximum subarray sum ending at the previous position. It is also considered greedy because instead of checking all subarrays, we can discard subarrays with negative sums since they can only decrease future subarray sums.

Use Kadane's algorithm.
- traverse the array while tracking the current subarray sum
- if the subarray sum becomes negative, discard it by resetting it to 0 because it will only reduce the sum of any future subarray
- else continue extending the current subarray and keep track of the maximum sum seen so far


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = first element in the array to track the maximum subarray sum
- set cur_sum = 0 to track the current subarray sum
- traverse the array
    - add the current number to cur_sum to include it in the current subarray
    - update res to the max of the current subarray sum and the previous maximum subarray sum
    - if current subarray sum < 0
        - discard the current subarray by resetting cur_sum to 0
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def maxSubArray(self, nums: List[int]) -> int:
        res = nums[0]
        cur_sum = 0
        for n in nums:
            cur_sum += n
            res = max(res, cur_sum)
            if cur_sum < 0:
                cur_sum = 0
        return res
        
```
