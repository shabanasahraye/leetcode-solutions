# <center> 673. Number of Longest Increasing Subsequence </center>


## Problem Description
[Click here](https://leetcode.com/problems/number-of-longest-increasing-subsequence/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array and need to find the total number of longest increasing subsequences (LIS).

This is a Dynamic Programming problem because the LIS ending at the current index depends on the LIS ending at previous indices.

For each index, compute:
- the length of the LIS ending at that index
- the number of LIS with that length ending at that index


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = input array size
- set size = an array of size n with ones 
    - *size[i] represents the length of the LIS ending at index i*
- set freq = an array of size n with ones 
    - *freq[i] represents the number of LIS ending at index i*
- set maxi = 1 to track the maximum LIS length
- set count = 1 to track the total number of LIS
- for each index i from 1 to n - 1
    - for each previous index j from 0 to i - 1
        - if current subseq is an increasing subseq i.e index i element is greater than index j element
            - if size of the LIS ending at index j is greater than the current LIS ending at i, we found a LIS ending at i
                - update size[i] and freq[i]
            - if size of the LIS ending at j = current LIS ending at i, we found another LIS with the same length as the current LIS ending at i
                - only update freq[i]
    - if length of the current LIS ending at i is greater than maxi
        - update maxi and count
    - if length of the current LIS ending at i is equal to maxi
        - only update count
- return count as the total number of longest increasing subsequences


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(creating size array + creating freq array + outer loop × inner loop) → O(n + n + n × n) → O(n²)
    - *for each index i, the inner loop compares it with all previous indices*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(size array + freq array) → O(n + n) → O(n)


## Code

```python
class Solution:

    def findNumberOfLIS(self, nums: List[int]) -> int:
        n = len(nums)
        size, freq = [1] * n, [1] * n
        count, maxi = 1, 1
        for i in range(1, n):
            for j in range(i):
                if nums[i] > nums[j]:
                    if size[j] + 1 > size[i]:
                        size[i] = size[j] + 1
                        freq[i] = freq[j]
                    elif size[j] + 1 == size[i]:
                        freq[i] += freq[j]
            if size[i] > maxi:
                maxi = size[i]
                count = freq[i]
            elif size[i] == maxi:
                count += freq[i]
        return count
        
```
