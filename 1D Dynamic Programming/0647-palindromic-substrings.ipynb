# <center> 647. Palindromic Substrings </center>


## Problem Description
[Click here](https://leetcode.com/problems/palindromic-substrings/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
This is similar to the [Longest Palindromic Substring](https://leetcode.com/problems/longest-palindromic-substring/description/) problem, but instead of finding the longest palindrome, we count every palindrome.

In a palindrome, the left and right characters are the same.
- in an odd-length palindrome, there is a central character for example 'aba'
- in an even-length palindrome there is no central character for example 'abba'.

For each character in the string, expand to the left and right to find all odd-length and even-length palindromes using the two-pointer approach.


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = input string length
- check base cases
- if string length is 1
    - return 1 because a single character is a palindrome
- set res = 0 to count the palindromes
- traverse the string <br>
*for every char in the string*
    - check odd-length palindromes
        - set l = i and r = i to track the left and right pointers of an odd-length palindrome
        - loop until l and r are in bounds and the left and right characters are the same 
            - add 1 to res because we found a palindrome
            - update l and r to expand to the left and right by one character
    - check even-length palindromes
        - set l = i and r = i + 1 to track the left and right pointers of an even-length palindrome
        - loop until l and r are in bounds and the left and right characters are the same 
            - add 1 to res because we found a palindrome
            - update l and r to expand to the left and right by one character
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(string traversal × (odd-length palindrome expansion + even-length palindrome expansion)) → O(n × (n + n)) → O(n² + n²) → O(n²)
    - *each center expands at most n characters in the worst case*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def countSubstrings(self, s: str) -> int:
        n = len(s)
        if n == 1:
            return 1
        res = 0
        for i in range(n):
            # odd length palindrome check
            l, r = i, i
            while l >= 0 and r < n and s[l] == s[r]:
                res += 1
                l -= 1
                r += 1
            # even length palindrome check
            l, r = i, i + 1
            while l >= 0 and r < n and s[l] == s[r]:
                res += 1
                l -= 1
                r += 1
        return res   
```
