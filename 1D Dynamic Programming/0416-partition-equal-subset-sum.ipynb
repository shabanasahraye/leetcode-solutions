# <center> 416. Partition Equal Subset Sum </center>


## Problem Description
[Click here](https://leetcode.com/problems/partition-equal-subset-sum/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array and need to determine whether it can be partitioned into two subsets of equal sum.

target subset sum = total array sum / 2

This is a Dynamic Programming problem because whether we can form the current target sum depends on whether we could form smaller sums using the previous elements.

It is similar to the 0/1 Knapsack problem because for each element we have two choices:
- include it in the current subset
- exclude it (leave it for the other subset)

Observations:
- each element must be placed entirely in one subset, not split into parts
- each subset must have a sum equal to half of the total array sum, therefore if the total sum is odd, partitioning is impossible

For example: 
- arr = [2, 4, 3, 3]
    - array sum = 12 → even
    - subset sum = 6
    - partition is possible because the array sum is even and 2 + 4 = 6
- arr = [4, 3, 3]
    - array sum = 10 → even
    - subset sum = 5
    - partition is not possible because the array sum is even, but no subset sums to 5
- arr = [4, 3, 3, 5] <br>
    - array sum = 15 → odd <br>
    - partition is not possible because the array sum is odd


## Approach
<!-- Describe your approach to solving the problem. -->
- find the sum of the input array
- check base cases
    - if the array sum is odd, partitioning is impossible
        - return false
- find the target subset sum = total sum / 2
- set dp = a boolean list of size equal to the target sum + 1 with a default value of false
    - *dp[i] represents whether it is possible to get sum i using the input array elements we’ve seen so far*
- set dp[0] to true because we can always get sum 0 by selecting no elements 
- traverse the input array <br>
*for each number in the input array*
    - traverse the dp list in reverse order <br>
    *for each sum i in range target sum  to current number*
       - check whether we can form the current sum by including or excluding the current number, then update dp[i]
            - if dp[i] is already true, we can already form sum i (exclude the current number)
            - if dp[i - current number] is true, including the current number forms sum i
    - if dp[target] becomes true, the target subset sum can be formed
        - return true
- return dp[target], which represents whether the target subset sum can be formed


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(outer loop × inner loop) → O(input array × target subset sum) → O(n × k)
    - *k = target subset sum = total array sum / 2*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(target subset sum) → O(k)


## Code

```python
class Solution:

    def canPartition(self, nums: List[int]) -> bool:
        summ = sum(nums)
        if summ & 1:
            return False
        target = summ // 2
        dp = [False] * (target + 1)
        dp[0] = True
        for num in nums:
            for i in range(target, num - 1, -1):
                dp[i] = dp[i] or dp[i - num]
            if dp[-1]:
                return True
        return dp[-1]
        
```
