# <center> 139. Word Break </center>


## Problem Description
[Click here](https://leetcode.com/problems/word-break/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a string and a dictionary of words. We need to determine whether the entire string can be segmented into one or more dictionary words.

*Note: we have to use the complete word, we can't use a part of it.*

This is a Dynamic Programming problem because whether a prefix of the string can be segmented depends on whether a previous prefix can be segmented.

For each position, try every word in the dictionary. If the prefix before the word can be segmented and the current prefix ends with that word, then the current prefix can also be segmented.


## Approach
<!-- Describe your approach to solving the problem. -->
- set dp = a boolean array of size len(s) + 1, where dp[i] represents whether the first i characters of the string can be segmented
    - *index = prefix length*
    - *value = whether the prefix can be segmented*
- set value at index 0 to True, because an empty string can always be segmented
- traverse all prefix lengths
    - for each prefix length i
        - for each word w in the dictionary
            - if the current prefix is at least as long as the current word, the prefix before the word can be segmented, and the current prefix ends with the current word
                - the current prefix can be segmented, set dp[i] = True
                - no need to check the remaining words, break the inner loop
- return the last index value from the dp array i.e dp[-1] that represents whether the entire string s can be segmented or not


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(dp array initialization + outer loop × inner loop × endswith() check) → O(input string size + input string size × words array size × max word length) → O(m + m × n × L) → O(m × n × 20) → O(m × n)
    - *L = maximum word length (L ≤ 20 in this problem)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(dp array) → O(string s size) → O(m)


## Code

```python
class Solution:

    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        dp = [False] * (len(s) + 1)
        dp[0] = True
        for i in range(1, len(s) + 1):
            for w in wordDict:
                if i >= len(w) and dp[i - len(w)] and s[:i].endswith(w):
                    dp[i] = True
                    break
        return dp[-1]
        
```
