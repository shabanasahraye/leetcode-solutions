## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
The input is in the form [x1, x2, ..., xn, y1, y2, ..., yn]
We need to rearrange it into alternating order: [x1, y1, x2, y2, ..., xn, yn]

- **Approach 1: Two-Pointers**
    - use two pointers, one starting at the first x element (index 0) and the other at the first y element (index n)
    - O(n) space
- **Approach 2: Bit Manipulation**
    - each value is at most 1000, so it fits within 10 bits (2¹⁰ = 1024)
    - we can store both the corresponding x and y values in one integer using bit manipulation
    - O(1) space


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Two-Pointers
- set res = an array to store the shuffled array
- set left pointer l = index 0 and right pointer r = index n
- loop n times
    - add x element at index l to the result
    - add the corresponding y element at index r to the result
    - increment l and r
- return the result array

### Approach 2: Bit Manipulation
- traverse the first half of the array
    - shift the x value left by 10 bits to make space for its corresponding y value
    - combine the shifted x value and y using the OR operator
- *after this loop, each element in the first half stores both the corresponding x and y values. Traverse these pairs in reverse order to extract x and y without overwriting values that have not been processed yet.*
- set j = rightmost position
- traverse the first half in reverse order
    - extract y using the & operator and store it at position j
        - *the rightmost 10 bits store y, so `num & 1023` extracts y because 1023 = 2¹⁰ − 1 = 1111111111₂*
    - extract x by shifting 10 bits to the right and store it at position j - 1
        - *the leftmost bits store x, so shifting right by 10 removes the rightmost 10 bits and extracts x*
    - decrement j by 2
- return nums


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - Two-Pointers Approach: O(loop) → O(n)
    - Bit Manipulation Approach: O(loop 1 + loop 2) → O(n + n) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - Two-Pointers Approach: O(result array) → O(2n) → O(n)
        - *input array size is 2n* 
    - Bit Manipulation Approach: O(1)


## Code

```python
class Solution:

    def shuffle(self, nums: List[int], n: int) -> List[int]:
        
        # Two-Pointers Approach
        # res = []
        # l, r = 0, n
        # for i in range(n):
        #     res.append(nums[l])
        #     res.append(nums[r])
        #     l += 1
        #     r += 1
        # return res

        # Bit Manipulation Approach
        for i in range(n):
            nums[i] = nums[i] << 10
            nums[i] = nums[i] | nums[i+n]
        j = len(nums) - 1
        for i in range(n-1, -1, -1):
            nums[j] = nums[i] & 1023 
            nums[j-1] = nums[i] >> 10  
            j -= 2
        return nums
        
```
