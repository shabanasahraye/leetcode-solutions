# <center> 268. Missing Number </center>


## Problem Description
[Click here](https://leetcode.com/problems/missing-number/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array containing n distinct numbers in the range [0, n], with exactly one number missing.

missing number = sum of all numbers from 0 to n − sum of the input array

the sum of all numbers from 0 to n is n(n + 1) / 2


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = input array size
- calculate the expected sum of numbers from 0 to n using the formula
- calculate the sum of the input array
- return the difference between the expected sum and the input array sum


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(finding input array sum) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def missingNumber(self, nums: List[int]) -> int:
        n = len(nums)
        return n * (n + 1) // 2 - sum(nums)
        
```
