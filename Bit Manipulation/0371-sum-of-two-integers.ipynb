## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given two integers a and b and need to return their sum without using the `+` or `-` operators.

To add two numbers, we start from the right side and add the digits. If the result is a two-digit number, we keep the rightmost digit as the result and pass the leftmost digit as the carry to the next position.

To perform addition without using the `+` operator, use bit manipulation:
- use XOR to compute the sum bits without the carry
    - *XOR returns 1 if the bits are different and 0 if they are the same*
- use AND to compute the carry
    - *AND returns 1 only when both bits are 1*
- use LEFT SHIFT (<<) to move the carry to the next bit
- repeat until no carry remains

Formula:
- a + b = (a XOR b) + ((a AND b) << 1)

Examples:
- 0 + 0 = 0 (result bit = 0, carry = 0)
- 0 + 1 = 1 (result bit = 1, carry = 0)
- 1 + 0 = 1 (result bit = 1, carry = 0)
- 1 + 1 = 10 (result bit = 0, carry = 1)


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if a is 0
        - return b because 0 + x = x
    - if b is 0
        - return a because x + 0 = x
- initialize a mask variable
    - *0xffffffff is a 32-bit hexadecimal mask equal to 4294967295 in decimal, you can also use its binary representation*
    - *Python integers are not limited to 32 bits. We use a 32-bit mask (0xffffffff) to limit the result to 32 bits and prevent an infinite loop caused by carry bits*
- while carry exists within 32 bits
    - get carry using AND
    - compute the sum without carry using XOR and store it in a
    - shift the carry to the next bit using LEFT SHIFT and store it in b
- convert the result back to a signed 32-bit integer if needed and return it


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: (loop) → O(32) → O(1)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def getSum(self, a: int, b: int) -> int:
        if a == 0:
            return b
        if b == 0:
            return a
        mask = 0xffffffff 
        while b & mask:
            carry = a & b 
            a = a ^ b
            b = carry << 1
        return (a & mask) if b > 0 else a
        
```
