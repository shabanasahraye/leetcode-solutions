# <center> 191. Number of 1 Bits </center>


## Problem Description
[Click here](https://leetcode.com/problems/number-of-1-bits/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a 32-bit unsigned integer and need to return the number of set bits (1s) in its binary representation.

Use Brian Kernighan's Algorithm
- repeatedly unset the rightmost set bit (change the rightmost 1 to 0)
- the loop ends when there are no set bits left, i.e. n becomes 0
- the number of loop iterations equals the number of set bits (1s)

For example: n = 101100
- iteration 1, n = 101**1**00 → 101**0**00
- iteration 2, n = 10**1**000 → 10**0**000
- iteration 3, n = **1**00000 → **0**00000
- total set bits = total loop iterations = 3


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to count the set bits
- loop until the number is zero
    - unset the rightmost set bit by computing `n &= (n - 1)`
    - increment res because one set bit was removed
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(number of set bits) → O(32) → O(1)
    - *a 32-bit integer has at most 32 set bits*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def hammingWeight(self, n: int) -> int:
        res = 0
        while n:
            n &= n-1
            res += 1
        return res
        
```
