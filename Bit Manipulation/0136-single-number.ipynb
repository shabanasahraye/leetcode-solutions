# <center> 136. Single Number </center>


## Problem Description
[Click here](https://leetcode.com/problems/single-number/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an array where every element appears twice except for one element that appears only once. We need to return that single element.

Use the XOR (^) operator because duplicate numbers cancel each other out. <br>
For example: <br>
a ^ b ^ a = a ^ a ^ b = 0 ^ b = b

XOR properties:
- XOR of two equal numbers is 0 (a ^ a = 0)
- XOR of 0 and a number is the number itself (0 ^ a = a)
- XOR is commutative, so the order does not matter (a ^ b = b ^ a)


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the single number
- traverse the array
    - take XOR between the current number and res, then update
        - *duplicate numbers cancel out, leaving only the single number*
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(array traversal) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def singleNumber(self, nums: List[int]) -> int:
        res = 0
        for n in nums:
            res ^= n
        return res
        
```
