## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a 32-bit unsigned integer and need to return its bits in reverse order.

It is similar to [Reverse Integer](https://leetcode.com/problems/reverse-integer/description/), but here we reverse bits instead of digits.

To reverse the bits of n, repeatedly extract the rightmost bit, add it to the result, and remove the bit from n.

Use the AND (&) operator to get the rightmost bit, the LEFT SHIFT (<<) operator to add a bit to the result, and the RIGHT SHIFT (>>) operator to remove the rightmost bit.

Example: Reverse 1101 into 1011
- n = 1101, res = 0000
    - get the rightmost bit: 1101 & 1 = 1
        - *you can also use the modulo operator (% 2) because the rightmost bit is 1 for odd numbers and 0 for even numbers*
    - left shift res to make space: 0000 << 1 = 0000
    - add the extracted bit: 0000 + 1 = 0001
    - right shift n to remove the bit: 1101 >> 1 = 0110
    - n = 0110, res = 0001
- n = 0110, res = 0001
    - get the rightmost bit: 0110 & 1 = 0
    - left shift res: 0001 << 1 = 0010
    - add the extracted bit: 0010 + 0 = 0010
    - right shift n: 0110 >> 1 = 0011
    - n = 0011, res = 0010
- n = 0011, res = 0010
    - get the rightmost bit: 0011 & 1 = 1
    - left shift res: 0010 << 1 = 0100
    - add the extracted bit: 0100 + 1 = 0101
    - right shift n: 0011 >> 1 = 0001
    - n = 0001, res = 0101
- n = 0001, res = 0101
    - get the rightmost bit: 0001 & 1 = 1
    - left shift res: 0101 << 1 = 1010
    - add the extracted bit: 1010 + 1 = 1011
    - right shift n: 0001 >> 1 = 0000
    - n = 0000, res = 1011


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to store the reversed bits
- loop 32 times because n is a 32-bit integer
    - left shift res by 1
        - *this shifts the bits left to make space for the new bit*
    - if the rightmost bit of n is 1
        - add 1 to res
            - *you can also use mod operator for even-odd check*
    - else the rightmost bit of n is 0
        - do nothing because the new bit in res is already 0
    - ight shift n to remove the rightmost bit
        - *right shifting by one bit is equivalent to integer division by 2*
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop) → O(32) → O(1)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(1)


## Code

```python
class Solution:

    def reverseBits(self, n: int) -> int:
        res = 0
        for _ in range(32):
            res <<= 1
            if n & 1:
                res += 1
            n >>= 1
        return res
        
```
