# <center> 67. Add Binary </center>


## Problem Description
[Click here](https://leetcode.com/problems/add-binary/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given two binary strings a and b. We need to return their sum as a binary string.

Follow binary addition rules:
- start from the right 
- add two bits at a time along with the carry
- store the rightmost bit as the result digit 
- pass the leftmost bit as the new carry

For example:
- 0 + 0 = 0 (carry = 0)
- 0 + 1 = 1 (carry = 0)
- 1 + 1 = 10 (result bit = 0, carry = 1)
- 1 + 1 + 1 = 11 (result bit = 1, carry = 1)


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = an array to store the result digits
- set i and j = rightmost indices of strings a and b to traverse from right to left
- set carry = 0
-  loop until both strings are processed and no carry remains
    - get the current bits of a and b (if the index is valid, otherwise use 0)
    - add the bits and carry to get the current sum
    - get the rightmost bit of the current sum using mod and add it to res 
       - *% 2 because it's binary*
    - get leftmost bit using integer division and add it to carry 
    - decrement i and j
- reverse the result array, convert it to a string, and return it


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(loop + reverse + join) → O(k + k + k) → O(k) → O(max(m, n))
    - *m = string a size, n = string b size*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(result array) → O(max(m, n))


## Code

```python
class Solution:

    def addBinary(self, a: str, b: str) -> str:
        res = []
        i, j = len(a) - 1, len(b) - 1
        carry = 0
        while i >= 0 or j >= 0 or carry:
            bit_a = int(a[i]) if i >= 0 else 0
            bit_b = int(b[j]) if j >= 0 else 0
            bit_sum = bit_a + bit_b + carry
            res.append(str(bit_sum % 2))
            carry = bit_sum // 2
            i -= 1
            j -= 1
        return ''.join(res[::-1])
        
```
