# <center> 2101. Detonate the Maximum Bombs </center>


## Problem Description
[Click here](https://leetcode.com/problems/detonate-the-maximum-bombs/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given bombs (nodes) with a blast radius. We need to find the maximum number of bombs that can be detonated starting from any single bomb.

A bomb x can detonate bomb y if the distance between them is within bomb x's blast radius, creating a directed edge from x to y.

Treat the bombs as a directed graph where bombs are nodes and an edge from x to y means bomb x can detonate bomb y.

The number of bombs a given bomb can detonate is the total number of reachable nodes from that bomb. Run DFS from every node to find the maximum number of reachable nodes.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the max bombs
- create a hashmap to represent the adjacency list (graph)
    - *key = nodes*
    - *values = list of neighbor nodes*
- find and add directed edges to the graph. For each pair of bombs, check if one bomb can directly detonate the other. Each bomb is represented as (x, y, range), where (x, y) is its location and range is its blast radius
    - for each index i, x1, y1 and range1 in the bomb list
        - for each index j, x2, y2, and range2
            - if i and j are the same, this is the same bomb
                - skip to avoid adding a self-edge
            - find the distance between bombs i and j 
            - if distance <= bomb i's range, bomb i can detonate bomb j
                - add an edge from bomb i to bomb j
            - if distance <= bomb j's range
                - add an edge from bomb j to bomb i
- define a helper function for DFS <br>
*dfs(source node, visited nodes set)*
    - if the node is already visited
        - return 0
    - mark the node as visited
    - visit neighbors of the node and do DFS to find other connected nodes
    - the visited set stores all connected/reachable nodes from the source node, return its length
- for each node
    - do DFS to find the total connected nodes and update res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph creation + graph traversal) → O(adjacency list creation + DFS from every node) → O(V² + (V + E) x V) → O(V x (V + E)) → O(n x (n + n²)) → O(n² + n³) → O(n³)
    - *V = total vertices/nodes = total bombs = n*
    - *E = total directed edges = at most n²*
    - *graph creation checks every pair of bombs, so it takes O(V²)*
    - *DFS is run from every node and visits all nodes and edges in the worst case*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(graph + DFS recursion stack + seen hashset) → O(E + V + V) → O(V + E) → O(n + n²) → O(n²)
    - *the graph stores at most n² directed edges*
    - *the DFS recursion stack can store up to V bombs*
    - *the visited set stores up to V bombs*


## Code

```python
class Solution:

    def maximumDetonation(self, bombs: List[List[int]]) -> int:
        res = 0
        graph = defaultdict(list)
        for i, (xi, yi, ri) in enumerate(bombs):
            for j, (xj, yj, rj) in enumerate(bombs):
                if i == j: continue
                dist = sqrt((xi - xj)**2 + (yi - yj)**2)
                if dist <= ri:
                    graph[i].append(j)
                if dist <= rj:
                    graph[j].append(i)

        def dfs(node: int, seen: Set[int]) -> int:
            if node in seen:
                return 0
            seen.add(node)
            for neighbor in graph[node]:
                dfs(neighbor, seen)
            return len(seen)

        for i in range(len(bombs)):
            res = max(res, dfs(i, set()))
        return res
        
```
