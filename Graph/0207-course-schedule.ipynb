# <center> 207. Course Schedule </center>


## Problem Description
[Click here](https://leetcode.com/problems/course-schedule/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to determine whether we can complete all courses given their prerequisites. <br>
This problem is about detecting a cycle in a directed graph.

Create a directed graph with courses as vertices/nodes and prerequisite relationships as directed edges. <br>
Use DFS to detect cycles. If a cycle exists, some courses depend on each other, so we cannot complete all courses. Otherwise, we can.


## Approach
<!-- Describe your approach to solving the problem. -->
- create a hashmap for the graph (adjacency list)
    - *key = node (course)*
    - *value = list of neighbors (prerequisite courses)*
- create a hashmap to track the status of a node
    - *key = node (course)*
    - *value = status (0 = unvisited, 1 = visited, -1 = visiting i.e on the current path)*
- fill the hashmap to create the graph
    - for each edge (course, prerequisite) in the prerequisites list
        - add the prerequisite to the neighbors list of the course
            - *this creates a directed edge from the course to its prerequisite*
- define a helper function for DFS <br>
*dfs(current node/course)*
    - if status is -1, we encountered the current node again in the path, there is a cycle
        - return false
    - if status is 1, the current node is visited and cycle-free (all prerequisite paths have been checked)
        - return true
    - else it is an unvisited node
        - add it to the current path by updating the status to -1
        - visit all the neighbors (prerequisite courses) to check for cycles
            - if there is a cycle and we can't finish the course i.e dfs returns false
                - return false
        - if the neighbors loop finishes without returning false, the current node is cycle-free and all of its prerequisites can be completed
            - remove it from the current path by updating the status to 1
            - return true
- for each course
    - do DFS to check if we can finish this course, if we can't i.e dfs returns false
        - return false
- if the loop finishes without returning false, we can finish all courses
    - return true


## Complexity
- Time complexity: O(graph creation + graph traversal) → O(adjacency list creation + adjacency list DFS) → O(E + (V + E)) → O(V + E)
    - *V = total vertices/nodes (courses), represented by the rows/keys in the adjacency list*
    - *E = total directed edges (prerequisite relationships), represented by the neighbor lists in the adjacency list*
    - *each node is visited once and each edge is processed once during DFS*

- Space complexity: O(graph + state hashmap + DFS recursion stack) → O(adjacency list + state hashmap + DFS recursion depth) → O((V + E) + V + V) → O(V + E)
    - *state hashmap stores the status of each course*
    - *recursion depth is V because DFS may visit every course before returning (worst case: courses form a single chain of prerequisites)*


## Code

```python
class Solution:

    def canFinish(self, numCourses: int, prerequisites: List[List[int]]) -> bool:
        graph = defaultdict(list)
        state = defaultdict(int)
        for a, b in prerequisites:
            graph[a].append(b)

        def dfs(node: int) -> bool:
            if state[node] == -1:
                return False
            if state[node] == 1:
                return True
            state[node] = -1
            for neighbor in graph[node]:
                if not dfs(neighbor):
                    return False
            state[node] = 1
            return True

        for course in range(numCourses):
            if not dfs(course):
                return False
        return True
```
