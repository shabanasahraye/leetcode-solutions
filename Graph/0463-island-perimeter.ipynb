## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an m x n binary matrix. 1 represents a land cell and 0 represents a water cell. Each cell is a square, and the land cells form an island. We need to find the island's perimeter.

Perimeter of a cell = total sides = 4 <br>
but here we will only count the sides that are adjacent to water <br>
Perimeter of the island = total sides adjacent to water

Traverse each cell, if you find a land cell:
- **Approach 1: Check All Four Neighbors**
    - check all 4 neighbors (top, bottom, left, right)
    - if a neighbor is water or out of bounds, increment the perimeter count
- **Approach 2: Count Shared Edges (Optimized)**
    - increment the perimeter count by 4 and check only two neighbors (bottom & right or top & left)
    - if a checked neighbor is also a land cell, decrement the perimeter count by 2
  

## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Check All Four Neighbors
- set m = total rows
- set n = total columns
- set res = 0 to track the perimeter
- for each cell
    - if it is a land cell
        - check its 4 neighbors (top, bottom, left, right)
        - if a neighbor is out of bounds or is a water cell
            - increment res by 1
- return res

### Approach 2: Count Shared Edges (Optimized) 
- set m = total rows
- set n = total columns
- set res = 0 to track the perimeter
- for each cell
    - if it is a land cell
        - increment res by 4
        - if the bottom cell is in bounds and is a land cell
            - decrement res by 2
        - if the right cell is in bounds and is a land cell
            - decrement res by 2
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Approach 1: O(grid traversal) → O(rows x columns) → O(m x n)
    - Approach 2: O(grid traversal) → O(rows x columns) → O(m x n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity:
    - Approach 1: O(1)
    - Approach 2: O(1)

```python
class Solution:

    def islandPerimeter(self, grid: List[List[int]]) -> int:
        
        # Approach 1: Check All Four Neighbors
        # m, n = len(grid), len(grid[0])
        # res = 0
        # for r in range(m):
        #     for c in range(n):
        #         if grid[r][c] == 1:
        #             if r == 0 or grid[r - 1][c] == 0:
        #                 res += 1
        #             if r == m - 1 or grid[r + 1][c] == 0:
        #                 res += 1
        #             if c == 0 or grid[r][c - 1] == 0:
        #                 res += 1
        #             if c == n - 1 or grid[r][c + 1] == 0:
        #                 res += 1
        # return res


        # Approach 2: Count Shared Edges (Optimized) 
        m, n = len(grid), len(grid[0])
        res = 0
        for r in range(m):
            for c in range(n):
                if grid[r][c] == 1:
                    res += 4
                    if r + 1 < m and grid[r + 1][c] == 1:
                        res -= 2
                    if c + 1 < n and grid[r][c + 1] == 1:
                        res -= 2
        return res
        
```
