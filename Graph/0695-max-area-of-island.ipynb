## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an m x n binary grid. 1 represents a land cell and 0 represents a water cell. An island (land block) is formed by connecting land cells horizontally or vertically. The area of an island is the number of land cells. We need to find the maximum island area.

The grid can be viewed as a graph where each cell is connected to its 4 neighbors. We need to find the connected component (island) with the maximum number of vertices (cells).

Traverse the graph using DFS to find connected components (islands). When you find an unvisited land cell, it represents the start of an island. Explore all connected land cells, count them to get the island's area, and mark them as visited. For marking a vertex/cell as visited, use a hash set or replace its value with 2 (you can use any value except 0 and 1).


## Approach
<!-- Describe your approach to solving the problem. -->
- set m = total rows and n = total columns
- set res = 0 to track the max area
- define a helper function for DFS <br>
*dfs(row, column)*
    - if row or column index is out of bound or current cell value is not 1 i.e either it is a visited land cell (value 2) or a water cell (value 0), return 0
    - else current cell is an unvisited land cell
        - mark it as visited
        - return 1 (current cell) + the areas returned by the 4 DFS calls after visiting its 4 neighbors (left, right, top, bottom)
- for each cell
    - if it is an unvisited land cell (value = 1), we found an island (connected component)
        - find its area and update res with the max of the current island area and the previous maximum area
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph traversal) → O(DFS) → O(V) → O(m x n)
    - *V = total vertices/nodes (cells) in the grid = m × n, because each cell is treated as a node in the graph*
    - *DFS visits each cell at most once because each cell's neighbors can be found directly (up, down, left, right)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(DFS recursion stack) → O(recursion depth) → O(V) → O(m x n)
    - *recursion depth is V because DFS may visit every land cell before returning (worst case: the entire grid is one island)*


## Code

```python
class Solution:

    def maxAreaOfIsland(self, grid: List[List[int]]) -> int:
        m, n = len(grid), len(grid[0])
        res = 0

        def dfs(r: int, c: int) -> int:
            if r < 0 or r == m or c < 0 or c == n or grid[r][c] != 1:
                return 0
            grid[r][c] = 2
            return 1 + dfs(r + 1, c) + dfs(r - 1, c) + dfs(r, c + 1) + dfs(r, c - 1)

        for row in range(m):
            for col in range(n):
                res = max(res, dfs(row, col))
        return res
```
