# <center> 752. Open the Lock </center>


## Problem Description
[Click here](https://leetcode.com/problems/open-the-lock/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Total lock combinations = $n^k$ = $slots^{wheels}$ = $10^4$ = 10000

There are 10⁴ possible lock combinations (0000 to 9999). We need to find the minimum number of turns required to reach the target combination. This is the same as finding the shortest path in a graph.

Treat each combination as a node in a graph. From each node, we can generate up to 8 neighboring combinations by turning one of the 4 wheels one step forward or backward.

use BFS to find the shortest path from the start node (0000) to the target while avoiding deadends.


## Approach
<!-- Describe your approach to solving the problem. -->
- check base cases
    - if the target is the start node (combination), we unlocked the lock without turning the wheel
        - return 0
    - convert the deadends to a set to search in O(1) time
    - if the start node is in the deadends, we can't unlock it
        - return -1
- set `seen` = a hashset to store the visited nodes and add the start node
- create a queue for BFS and add the start node
- set `res` = 0 to track the total turns
- traverse the graph level by level i.e loop until the queue is empty
    - for current level
        - pop queue to get current node
        - if node is the target node, return res
        - create the child nodes <br>
        *for each of the 4 wheels*
            - get current wheel digit i.e char at index i in the current node
            - turn it up and get the child node i.e increment the digit at index i to get a combination
            - if the child node is not in visited and not in deadends, add it to the queue and set
            - turn it down and get the child node i.e decrement the digit at index i to get a combination
            - if the child node is not in visited and not in deadends, add it to the queue and set
    - increment `res` after checking a level
- if the BFS completes without returning anything, there's no shortest path i.e we can't unlock it, return -1


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
Time complexity: O(graph BFS) → O(V) → O($10^4$)
    - *V = total vertices/nodes = total combinations = $10^4$*
    - *each combination is visited at most once*
    - *for each combination, we generate at most 8 neighbors (4 wheels × 2 directions), which is a constant amount of work*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
Space complexity: O(BFS queue + seen hashset) → O(V + V) → O(V) → O($10^4$)
    - *the queue stores combinations to process*
    - *the seen hashset stores visited combinations*


## Code

```python
class Solution:

    def openLock(self, deadends: List[str], target: str) -> int:
        if target == '0000':
            return 0
        deadends = set(deadends)
        if '0000' in deadends:
            return -1
        seen = seen = {'0000'}
        q = deque(['0000'])
        res = 0
        while q:
            for _ in range(len(q)):
                cur = q.popleft()
                if cur == target:
                    return res
                for i in range(4):
                    digit = int(cur[i])
                    up = cur[:i] + str((digit + 1) % 10) + cur[i+1:]
                    if up not in seen and up not in deadends:
                        q.append(up)
                        seen.add(up)
                    down = cur[:i] + str((digit - 1) % 10) + cur[i+1:]
                    if down not in seen and down not in deadends:
                        q.append(down)
                        seen.add(down)
            res += 1
        return -1
```
