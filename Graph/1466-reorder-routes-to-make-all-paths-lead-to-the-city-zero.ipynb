# <center> 1466. Reorder Routes to Make All Paths Lead to the City Zero </center>


## Problem Description
[Click here](https://leetcode.com/problems/reorder-routes-to-make-all-paths-lead-to-the-city-zero/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given n cities (nodes) and n - 1 one-way roads (edges). We need to reorient roads that don't lead to the capital (city 0).

Visit all cities starting from city 0 and count the roads that need to be reoriented.

To do this, create an undirected graph and do DFS. Start DFS from the root node (city 0). If we traverse a forward edge (an edge directed away from city 0), it must be reversed because every path should lead to city 0. Count the total forward edges.


## Approach
<!-- Describe your approach to solving the problem. -->
- use a hashmap to create an undirected graph (adjacency list)
    - *key = node (city)*
    - *value = list of neighbors*
    - *we can also use a nested list*
- create a stack for DFS and add the root node (city 0)
- create a set to store visited nodes
- set res = 0 to count the forward edges (roads that need to be reoriented)
- for each edge (start city, end city) in the connection list
    - add a bidirectional edge to the graph i.e add the end city as a neighbor to the neighbors list of the start city and vice versa
    - also store an indicator with each neighbor to show whether the original edge is forward or backward
    - in the input list the edges are forward (from start city to end city), so use 1 for forward and 0 for backward
- do DFS i.e loop until the stack is empty
    - pop stack to get current node
    - mark the current node as visited
    - for each neighbor and edge indicator in the neighbor's list of the current node
        - if the neighbor has not been visited
            - if the edge is a forward edge i.e value is 1
                - increment res
            - add the neighbor to the stack
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph creation + graph traversal) → O(adjacency list creation + adjacency list DFS) → O(E + V + E) → O(V + E) → O(n + n) → O(n)
    - *V = total vertices/nodes = total cities = n*
    - *E = total edges = total roads = n - 1*
    - *DFS visits each city once and processes each road once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(graph + DFS stack + visited set) → O(adjacency list + DFS stack + visited nodes) → O((V + E) + V + V) → O(V + E) → O(n)
    - *the graph stores all cities and roads*
    - *the DFS stack stores at most V cities*
    - *the visited set stores all visited cities*


## Code

```python
class Solution:

    def minReorder(self, n: int, connections: List[List[int]]) -> int:
        graph = defaultdict(list)
        stack = [0]
        seen = set()
        res = 0
        for a, b in connections:
            graph[a].append((b, 1))
            graph[b].append((a, 0))
        while stack:
            node = stack.pop()
            seen.add(node)
            for neighbor, edge_type in graph[node]:
                if neighbor not in seen:
                    stack.append(neighbor)
                    if edge_type == 1: 
                        res += 1
        return res
        
```
