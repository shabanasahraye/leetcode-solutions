## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
An extra edge has been added to a connected graph and we need to find it (the redundant edge).


Adding an edge to a connected graph creates a cycle. So, the edge that creates the cycle is the redundant edge.
We can detect the cycle in a connected graph using:
- *Depth-First Search (DFS)* <br>
    - for each edge, temporarily remove it and use DFS to check whether its two nodes are still connected
    - if they are, that edge is redundant because there is another path between the two nodes
    - time complexity = O(V + E) per search = O(E × (V + E)) ≈ O(n²)
- *Union-Find (Disjoint Set)* <br>
    - a data structure used to efficiently track groups of connected nodes in a graph
    - has two operations:
        - *find*: identifies which group a node belongs to
        - *union*: merges two different groups into one
    - if two nodes are already in the same group, connecting them again creates a cycle
    - use union by rank (size) and path compression to optimize the operations
    - time complexity = O(n)


## Approach
<!-- Describe your approach to solving the problem. -->
### Union-Find Approach
- create a list to store parent of each node/vertex
- create a list to store the rank (size) of each tree. Initially, each vertex represents a tree and acts as a root node of the tree, so the size is 1 for each tree
- define a helper function for the find operation, it will find the root parent (root node) using path compression <br>
*find(node)*
    - set p = immediate parent of the node
    - loop until we find the root parent
        - do path compression by updating the parent of the current node to its grandparent
        - update p
    - return p
- define a helper function for the union operation, it will take two nodes and merge the lower rank node (smaller tree) with the higher rank node (larger tree) <br>
*union(node1, node2)*
    - find the root parent of the nodes
    - if the parent of both nodes are the same, the nodes are already connected, the union can't be performed because it will result in a cycle
        - return false
    - if the rank of node1 parent is greater than the rank of node2 parent, node1 belongs to a taller tree
        - add node2 to the node1 tree by updating the root parent of node2
        - update the rank (size) of node1 parent
    - else node2 parent rank is higher
        - add node1 to the node2 tree by updating the root parent of node1
        - update the rank (size) of node2 parent
- for each edge (node1, node2)
    - check if we can add the edge i.e connect (union) the nodes
        - if we can't, the edge forms a cycle, return the edge 
        - else, connect the nodes
  
  
## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Union-Find Approach: O(parent list creation + rank list creation + cycle detection using Union-Find) → O(V + V + E × α(V)) → O(n + n + n × α(n)) → O(n + n + n × constt) → O(n)
        - *V = total vertices (nodes)*
        - *E = total edges*
        -*α(n) is the inverse Ackermann function, which grows very slowly and is nearly constant*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Union-Find Approach: O(parent list + rank list) → O(V + V) → O(n + n) → O(n)
        - *V = total vertices or nodes*


## Code

```python
class Solution:

    def findRedundantConnection(self, edges: List[List[int]]) -> List[int]:
        
        # Union-Find Approach
        
        parent = [i for i in range(len(edges) + 1)]
        rank = [1] * (len(edges) + 1)

        def find(n: int) -> int:
            p = parent[n]
            while p != parent[p]:
                parent[p] = parent[parent[p]]
                p = parent[p]
            return p

        def union(n1: int, n2: int) -> bool:
            p1, p2 = find(n1), find(n2)
            if p1 == p2:
                return False
            if rank[p1] > rank[p2]:
                parent[p2] = p1
                rank[p1] += rank[p2]
            else:
                parent[p1] = p2
                rank[p2] += rank[p1]
            return True

        for n1, n2 in edges:
            if not union(n1, n2):
                return [n1, n2]
```
