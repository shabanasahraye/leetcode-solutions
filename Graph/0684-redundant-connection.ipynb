## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
An extra edge has been added to a connected graph and we need to find it (the redundant edge).

Adding an edge to a connected graph creates a cycle. So, the edge that creates the cycle is the redundant edge.
We can detect the cycle in a connected graph using:
- **Depth-First Search (DFS)** <br>
    - process each edge one by one
    - for each edge, temporarily remove it and use DFS to check whether its two nodes are still connected
    - if they are connected, adding this edge creates a cycle, so it is the redundant edge
    - time complexity = O(V + E) per search = O(E × (V + E)) ≈ O(n²)
- **Union-Find (Disjoint Set)** <br>
    - a data structure used to efficiently track groups of connected nodes in a graph
    - has two operations:
        - *find*:  finds the root (representative) of the group to determine which group a node belongs to
        - *union*: merges two different groups into one
    - if two nodes are already in the same group, connecting them again creates a cycle
    - to make these operations efficient, Union-Find is usually optimized using:
        - **union by size:** attach the tree with smaller size (fewer nodes) to the tree with larger size
        - **union by rank:** attach the tree with smaller rank (height) to the tree with larger rank
        - **path compression:** during the find operation, make each visited node point closer to the root, flattening the tree
        - both *union by size + path compression* and *union by rank + path compression* achieve nearly constant time per operation
    - we will use union by size and path compression
    - time complexity = O(n × α(n)) ≈ O(n)


## Approach
<!-- Describe your approach to solving the problem. -->
### Union-Find Approach
- create a list to store the parent of each node/vertex
- create a list to store the size of each tree
    - initially, each vertex represents a tree and acts as a root node of the tree, so the size is 1 for each tree
- define a helper function for the find operation, it will find the root parent (root node) using path compression <br>
*find(node)*
    - loop until the current node is its own parent (root node)
        - do path compression by updating the parent of the current node to its grandparent
        - update the current node to its new parent
    - return the root parent
- define a helper function for the union operation, it takes two nodes and merges the smaller tree with the larger tree <br>
*union(node1, node2)*
    - find the root parent of the nodes
    - if both nodes have the same root parent, they are already connected, so adding this edge creates a cycle
        - return false
    - if the size of node1's tree is greater than or equal to the size of node2's tree
        - add node2's tree to node1's tree by updating the root parent of node2
        - update the size of node1's tree
    - else
        - add node1's tree to node2's tree by updating the root parent of node1
        - update the size of node2's tree
- for each edge (node1, node2)
    - check if we can add the edge i.e connect (union) the nodes
        - if we can't, the edge forms a cycle, return the edge 
        - else, connect the nodes
  
  
## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Union-Find Approach: O(parent list creation + size list creation + cycle detection using Union-Find) → O(V + V + E × α(V)) → O(n + n + n × α(n)) → O(n + n + n × constt) → O(n)
        - *V = total vertices (nodes)*
        - *E = total edges*
        - *α(n) is the inverse Ackermann function, which grows very slowly and is nearly constant*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Union-Find Approach: O(parent list + size list) → O(V + V) → O(n + n) → O(n)
        - *V = total vertices or nodes*


## Code

```python
class Solution:

    def findRedundantConnection(self, edges: List[List[int]]) -> List[int]:
        
        # Union-Find Approach
        
        parent = [i for i in range(len(edges) + 1)]
        size = [1] * (len(edges) + 1)

        def find(n: int) -> int:
            while n != parent[n]:
                parent[n] = parent[parent[n]]
                n = parent[n]
            return n

        def union(n1: int, n2: int) -> bool:
            p1, p2 = find(n1), find(n2)
            if p1 == p2:
                return False
            if size[p1] >= size[p2]:
                parent[p2] = p1
                size[p1] += size[p2]
            else:
                parent[p1] = p2
                size[p2] += size[p1]
            return True

        for n1, n2 in edges:
            if not union(n1, n2):
                return [n1, n2]
                
```
