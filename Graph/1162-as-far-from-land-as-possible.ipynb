# <center> 1162. As Far from Land as Possible </center>


## Problem Description
[Click here](https://leetcode.com/problems/as-far-from-land-as-possible/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the water cell with the maximum distance to its nearest land cell.

**Approach 1** <br>
- do BFS from each water cell to find its nearest land cell, then return the maximum distance among all water cells. 
- time complexity = O(total water cells × grid BFS) → O(W × V) → O(W × n²)

**Approach 2** <br>
- the maximum distance from a water cell to its nearest land cell is the same as the maximum distance from a land cell to its nearest water cell
- do multisource BFS i.e. start BFS from all land cells at the same time, expand toward the nearest water cells, and track the maximum distance 
- traverse level by level, all land cells start at level 0 and the last level that reaches a water cell is the answer
- time complexity = O(grid BFS) → O(V) → O(n²)


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2
- set res = -1 to track the max distance
- set n = total rows
- create a list and add all 4 directions (right, left, bottom, top)
- create a queue for BFS
- traverse the grid
    - if you get a land cell (cell value 1) 
        - add its coordinates (row, column) to the queue
- if q is empty (all cells are water) or q size is equal to the total grid cells (all cells are land)
    - return -1
- loop until q is empty
    - for each level
        - pop queue to get coordinates of current level cell
        - explore all 4 neighbors to find unvisited water cells <br>
        *for each direction*
            - get the neighbor cell coordinates
            - if the indices are inbound and the neighbor is an unvisited water cell
                - add it to the queue
                - mark the neighbor as visited by updating the value to -1
                    - *(we can use any value or a hashset for marking visited cells)*
    - after each level, increment res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Approach 2: O(grid traversal + graph traversal) → O(rows × columns + adjacency matrix BFS) → O(V + V) → O(V) → O(n × n) → O(n²)
        - *V = total vertices/nodes = total grid cells = n × n*
        - *the first traversal visits every cell once to add all land cells to the BFS queue*
        - *BFS visits each cell at most once while expanding from the land cells*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Approach 2: O(BFS queue) → O(total cells in queue) → O(V) → O(n × n) → O(n²)
        - *the BFS queue stores at most V cells*


## Code

```python
class Solution:

    def maxDistance(self, grid: List[List[int]]) -> int:
        res = -1
        n = len(grid)
        directions = [(0, 1), (0, -1), (1, 0), (-1, 0)]
        q = deque()
        for r in range(n):
            for c in range(n):
                if grid[r][c] == 1:
                    q.append((r, c))
        if not q or len(q) == n * n:
            return res
        while q:
            for _ in range(len(q)):
                row, col = q.popleft()
                for x, y in directions:
                    r, c = x + row, y + col
                    if 0 <= r < n and 0 <= c < n and grid[r][c] == 0:
                        q.append((r, c))
                        grid[r][c] = -1
            res += 1
        return res
        
```
