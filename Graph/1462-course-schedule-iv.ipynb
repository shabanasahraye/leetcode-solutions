## Problem Description
[Click here](https://leetcode.com/problems/course-schedule-iv/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
For a query (a, b), we need to find if course a is a prerequisite of b.
 
Find all direct and indirect prerequisites for every course and store them in a hashmap. Use the hashmap to answer the queries.

Create a directed graph with courses as vertices/nodes and prerequisite relationships as edges. Use DFS with memoization to find all prerequisites for each course.


## Approach
<!-- Describe your approach to solving the problem. -->
- create a list to store query result
- create a hashmap to store the prerequisites for each course
    - *key = node (course)*
    - *value = set of all neighbor nodes (prerequisites)*
- create a hashmap for the graph (adjacency list)
    - *key = node (course)*
    - *value = list of direct neighbor nodes (prerequisites)*
- create graph
    - for each edge (course, prerequisite) in the prerequisites list
        - add the prerequisite as a neighbor to the course node
- define a helper function for DFS <br>
*dfs(current node/course)*
    - if the node is not in the prerequisites hashmap
        - find all of its direct and indirect neighbors (prerequisites) <br>
        *for each direct neighbor of the node*
            - do DFS on the direct neighbors to find the indirect neighbors
            - update the current course's neighbors set by taking the union of the direct and indirect neighbors sets
    - add the current node to its prerequisite set
    - return the current node prerequisite set
- for each course
    - do DFS to find all direct and indirect prerequisites
- for each query (a, b)
    - if course a is in the prerequisite set of course b, add true to the result list, else add false
- return the result list


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph creation + graph traversal + queries traversal) → O(adjacency list creation + DFS with memoization for each node + total queries) → O(E + adjacency list DFS * total nodes + Q) → O(E + (V + E) * V + Q) → O(V * (V + E) + Q)
    - *V = total vertices/nodes = total courses*
    - *E = total directed edges = total prerequisite relationships*
    - *each course is processed once, and each edge is traversed while computing prerequisite sets*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(graph + DFS recursion stack + prerequisite hashmap + result list) → O(adjacency list + recursion depth + prerequisite hashmap + total queries) → O((V + E) + V + V + Q) → O(V + E + Q)
    - *the graph stores all courses and prerequisite relationships*
    - *the prerequisite hashmap stores the prerequisite set for each course*
    - *recursion depth is V because DFS may visit every course before returning (worst case: courses form a single chain)*
    - *the result list stores one answer for each query*


## Code

```python
class Solution:

    def checkIfPrerequisite(self, numCourses: int, prerequisites: List[List[int]], queries: List[List[int]]) -> List[bool]:
        res = []
        pre_req = defaultdict(set) 
        graph = defaultdict(list)
        for a, b in prerequisites:
            graph[b].append(a)
        
        def dfs(node: int):
            if node not in pre_req:
                for neighbor in graph[node]:
                    pre_req[node] |= dfs(neighbor)
            pre_req[node].add(node)
            return pre_req[node]

        for course in range(numCourses):
            dfs(course)
        for u, v in queries:
            res.append(u in pre_req[v])
        return res
        
```
