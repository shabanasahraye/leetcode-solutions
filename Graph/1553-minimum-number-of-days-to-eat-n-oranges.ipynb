# <center> 1553. Minimum Number of Days to Eat N Oranges </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-number-of-days-to-eat-n-oranges/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. --> 
We need to find the min days to eat all oranges. Each day, we can either eat one orange, half of the oranges, or two-thirds of the oranges.

We start with n oranges and reduce the number until none are left. At each step, we have two choices:
- eat n/2 oranges (if n is divisible by 2) or
- eat 2n/3 oranges (if n is divisible by 3)

Do DFS with memoization to explore both choices and return the minimum number of days.


## Approach
<!-- Describe your approach to solving the problem. -->
- create a hashmap for memoization to store the minimum days required to eat x oranges
    - *key = oranges left*
    - *value = min days required to eat x oranges*
- add the base cases to the hashmap (days required to eat 0 and 1 oranges)
- define a helper function for DFS <br>
*dfs(oranges left)*
    - if x is in the hashmap, we have already found the min days required to eat x oranges, return it
    - recursively compute the minimum days for both choices and update the hashmap with the min of the two
    - return the min days from the hashmap
- call DFS to find the min days required to eat n oranges 


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(DFS with memoization) → O(total unique states) → O(log²n)
    - *there are log²n unique states because each state is determined by repeatedly dividing n by 2 and/or 3*
    - *memoization ensures each reachable state is computed only once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(DFS recursion stack + hashmap) → O(recursion depth + memoization) → O(logn + log²n) → O(log²n)
    - *recursion depth is logn because x is repeatedly divided by 2 or 3*
    - *the hashmap stores log²n unique states*


## Code

```python
class Solution:

    def minDays(self, n: int) -> int:
        dp = {0: 0, 1: 1}

        def dfs(x: int) -> int:
            if x in dp:
                return dp[x]
            dp[x] = 1 + min(x % 2 + dfs(x // 2), x % 3 + dfs(x // 3))
            return dp[x]

        return dfs(n)
        
```
