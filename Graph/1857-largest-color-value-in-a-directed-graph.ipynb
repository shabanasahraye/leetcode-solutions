# <center> 1857. Largest Color Value in a Directed Graph </center>


## Problem Description
[Click here](https://leetcode.com/problems/largest-color-value-in-a-directed-graph/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the maximum frequency of any color along a path in a directed graph. If the graph contains a cycle, there is no valid path.

Use topological sorting to process nodes in order, detect cycles, and track the color frequencies along each path.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the max color value 
- set seen = 0 to track the total visited nodes
- set n = total nodes
- create a hashmap to represent the graph (adjacency list)
    - *key = node*
    - *value = list of neighbor nodes*
- create a queue for BFS topological sorting
- create a list of size n to store the in-degree (total number of incoming edges) of each node
    - *index = node*
    - *value = in-degree*
- create a list to store color frequencies for each node
    - *index = node*
    - *value = hashmap storing the frequency of each color in the longest path ending at that node*
        - *hashmap key = color*
        - *value = color frequency*
- add edges to the graph
    - for each source node, destination node in the edges
        - add an edge from the destination node to the source node to process nodes in reverse topological order
        - increment the in-degree of the source node
- add nodes with 0-indegree to the queue
- do topological sort
    - loop until q is empty
        - pop q to get a node to process
        - get the color of the current node and increment its frequency in the longest path ending at this node
        - update res
        - increment seen because we have visited a new node
        - visit the neighbors of the current node
            - for each neighbor node:
                - update the neighbor's color frequencies using the maximum frequencies from the current node's path
                - reduce the neighbor's in-degree by 1 because we processed one incoming edge
                - if the neighbor's in-degree becomes 0
                    - add it to the queue for topological processing
- if we have not visited all n nodes, there is a cycle, return -1
- else return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph (adjacency list) creation + adding 0-indegree nodes + topological sort) → O(E + V + (V + E)) → O(V + E) → O(n + m)
    - *V = total vertices or nodes*
    - *E = total edges*
    - *each node is processed once, and each edge is processed once during topological sorting*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(graph + in-degree list + color count list + BFS queue) → O((V + E) + V + (V × C) + V) → O(V + E + V × C) → O(V + E + V × const) → O(V + E) → O(n + m)
    - *V = total vertices or nodes*
    - *E = total edges*
    - *C = total possible colors = 26 (constant)*
    - *the graph stores all nodes and edges*
    - *the in-degree list stores one value for each node*
    - *the color count list stores a frequency map for each node, so it can store up to V × C → V x 26 entries*
    - *the BFS queue stores at most V nodes*


## Code

```python
class Solution:

    def largestPathValue(self, colors: str, edges: List[List[int]]) -> int:
        res, seen, n = 0, 0, len(colors)
        graph, q = defaultdict(list), deque()
        in_degree = [0] * n
        count = [defaultdict(int) for _ in range(n)]
        for u, v in edges:
            graph[v].append(u)
            in_degree[u] += 1
        for node in range(n):
            if not in_degree[node]:
                q.append(node)
        while q:
            cur = q.popleft()
            count[cur][colors[cur]] += 1
            res = max(res, count[cur][colors[cur]])
            seen += 1
            for v in graph[cur]:
                for c in count[cur]:
                    count[v][c] = max(count[v][c], count[cur][c])
                in_degree[v] -= 1
                if not in_degree[v]:
                    q.append(v)
        if seen < n:
            return -1
        return res
        
```
