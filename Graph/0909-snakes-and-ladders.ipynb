# <center> 909. Snakes and Ladders </center>


## Problem Description
[Click here](https://leetcode.com/problems/snakes-and-ladders/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an n × n board and need to find the minimum number of moves to reach the last position (n²). This is the same as finding the shortest path in a graph.

Treat each board position as a node in a graph. From each position, we can move to the next 1 to 6 positions (dice roll), making this a shortest-path problem. Use Breadth-First Search (BFS).


## Approach
<!-- Describe your approach to solving the problem. -->
- set n = total rows
- set target = last position value
- the board has n² positions numbered in a bottom-to-top zigzag order, reverse the board first so that position 1 is in the first row, making it easy to convert a position number into row and column coordinates
- create a hash set to store the visited positions
- create a queue for BFS. Add (root node / start position, moves required to reach root node) to queue
- do BFS i.e loop until the queue is empty
    - pop queue to get current position and moves required to reach it
    - roll the dice, (a dice has 6 values, loop 6 times to get all possible next positions) <br>
    *for each dice value*
        - get the next position
        - get the coordinates (row, column) of the next position
            - *row = (next position - 1) / total rows*
            - *column = (next position - 1) % total rows*
        - if the row is odd, column index = total columns - 1 - column
            - *(in columns, the positions alternate i.e even rows have positions in ascending order and odd in descending order)*
        - if the value at the next position is not -1, the position contains a snake or ladder
            - update the next position
        - if the next position is the target position
            - increment moves and return it
        - if the next position is not visited
            - add the position to the set
            - add the position and moves to the queue
- if the BFS completes without returning anything, we can't reach the last position, return -1


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(board reversal + graph traversal) → O(board reversal + BFS) → O(n² + V) → O(n² + n²) → O(n²)
    - *V = total vertices/nodes = total board positions = n²*
    - *each position is visited at most once*
    - *each position generates at most 6 next positions (dice rolls), which is constant work*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(seen hashset + BFS queue) → O(total positions + total positions) → O(V + V) → O(V) → O(n²)
    - *the seen hashset stores the visited positions*
    - *the BFS queue stores at most V positions*


## Code

```python
class Solution:

    def snakesAndLadders(self, board: List[List[int]]) -> int:
        n = len(board)
        target = n * n
        board.reverse()
        seen = set()
        q = deque([(1, 0)])
        while q:
            square, moves = q.popleft()
            for i in range(1, 7):
                next_sq = square + i
                r, c = (next_sq - 1) // n, (next_sq - 1) % n
                if r % 2:
                    c = n - 1 - c
                if board[r][c] != -1:
                    next_sq = board[r][c]
                if next_sq == target:
                    return moves + 1
                if next_sq not in seen:
                    seen.add(next_sq)
                    q.append((next_sq, moves + 1))
        return -1
        
```
