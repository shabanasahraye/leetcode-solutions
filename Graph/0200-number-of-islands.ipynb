# <center> 200. Number of Islands </center>


## Problem Description
[Click here](https://leetcode.com/problems/number-of-islands/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given an m × n binary grid. '1' represents a land cell and '0' represents a water cell. An island is formed by connecting land cells horizontally or vertically. We need to return the total number of islands.

The grid can be viewed as a graph where each cell is connected to its 4 neighbors. We need to find the connected components (islands).

Traverse the graph using DFS. Whenever we find an unvisited land cell, explore its neighbors to find all connected land cells and mark them as visited. Count the number of times we start DFS from an unvisited land cell, because each such DFS discovers exactly one island.

To mark a cell as visited, use a hash set or replace its value with '2' (you can use any value except '0' and '1').


## Approach
<!-- Describe your approach to solving the problem. -->
- set m = total rows
- set n = total columns
- set res = 0 to track the total number of subgraphs (islands)
- define a helper function for DFS <br>
*dfs(row, column)*
    - if row or column index is out of bound or current cell value is not 1 i.e either it is a visited land cell (value 2) or a water cell (value 0)
        - return
    - else cell value is 1 i.e it is an unvisited land cell
        - mark it as a visited land cell by replacing the value with 2
        - visit its 4 neighbors (left, right, top, bottom) to find other connected land cells
- for each cell in the grid
    - if it is an unvisited land cell i.e value is 1, we found a subgraph (island)
        - increment res by 1 
        - do DFS to find and mark the connected land cells
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph traversal) → O(DFS) → O(V) → O(m * n)
    - *V = total vertices/nodes (cells) in the grid = m × n, because each cell is treated as a node in the graph*
    - *DFS visits each cell at most once because each cell's neighbors can be found directly (up, down, left, right)*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(DFS recursion stack) → O(recursion depth) → O(V) → O(m * n)
    - *recursion depth is V because DFS may visit every land cell before returning (worst case: the entire grid is one island)*


## Code

```python
class Solution:

    def numIslands(self, grid: List[List[str]]) -> int:
        m, n = len(grid), len(grid[0])
        res = 0

        def dfs(r: int, c: int) -> None:
            if r < 0 or r == m or c < 0 or c == n or grid[r][c] != '1':
                return
            grid[r][c] = '2' 
            dfs(r, c - 1)
            dfs(r, c + 1)
            dfs(r + 1, c)
            dfs(r - 1, c)

        for row in range(m):
            for col in range(n):
                if grid[row][col] == '1':
                    res += 1
                    dfs(row, col)
        return res
```
