## Problem Description

## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to return the minimum time required to rot all oranges or -1 if it is impossible to rot all oranges.

The grid is an adjacency matrix (graph), and rotten oranges spread to their neighboring oranges (top, bottom, left, right) level by level.

Do BFS to traverse the graph and rot neighboring fresh oranges level by level. For each level, increment the minute counter. In the end, if all fresh oranges are rotten, return the total minutes, else return -1.


## Approach
<!-- Describe your approach to solving the problem. -->
- set m = total rows, n = total columns
- create a queue to store rotten oranges for BFS (level-order traversal)
- set fresh = 0 to track fresh oranges
- set res = 0 to track the time (minutes) required to rot all oranges
- for each cell in the grid
    - if it is a fresh orange (value 1)
        - increment fresh
    - if it is a rotten orange (value 2)
        - add its coordinates to the queue
- traverse the graph level by level while the queue is not empty and fresh oranges remain
    - at each level
        - pop queue to get the rotten orange in the current level
        - visit all 4 neighbors (bottom, top, left, right)
            - if the indices are in bounds and neighbor is a fresh orange (value 1)
                - mark it as rotten by changing its value to 2
                - add its coordinates to the queue
                - decrement fresh
    - after each level traversal, a minute is passed, increment res
- return res if no fresh oranges are left else return -1


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(grid traversal + graph traversal) → O(rows × columns + BFS) → O(V + V) → O(V) → O(m × n)
    - *V = total vertices/nodes (grid cells) = m × n*
    - *the first traversal counts fresh oranges and adds all initially rotten oranges to the queue*
    - *BFS visits each cell at most once while spreading the rot*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(BFS queue) → O(total rotten oranges) → O(V) → O(m × n)
    - *the BFS queue stores at most V cells*


## Code

```python
class Solution:

    def orangesRotting(self, grid: List[List[int]]) -> int:
        m, n = len(grid), len(grid[0])
        q = collections.deque()
        fresh, res = 0, 0
        for r in range(m):
            for c in range(n):
                if grid[r][c] == 1:
                    fresh += 1
                elif grid[r][c] == 2:
                    q.append((r, c))
        while fresh > 0 and q:
            for i in range(len(q)):
                r, c = q.popleft()
                for i, j in [(r + 1, c), (r - 1, c), (r, c - 1), (r, c + 1)]:
                    if 0 <= i < m and 0 <= j < n and grid[i][j] == 1:
                        grid[i][j] = 2
                        q.append((i, j))
                        fresh -= 1
            res += 1
        return res if fresh == 0 else -1
        
```
