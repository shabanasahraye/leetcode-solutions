## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given n cities (nodes) connected by roads (edges) with distances (edge weights). We need to find the minimum score of a path from city 1 to city n, where the score of a path is the minimum edge weight on that path.

Since all cities are connected to city 1, use DFS to traverse all reachable nodes and track the minimum edge weight among all visited roads.


# Approach
<!-- Describe your approach to solving the problem. -->
- set res = infinity to track the minimum edge weight
- create a hashset to track visited nodes (cities)
- create a hashmap for the adjacency list representation of the graph
  - *key = node (city)*
  - *value = list of (neighbor node, edge weight) pairs*
- create the graph (adjacency list)
    - for each edge (source node, destination node, distance or edge weight)
        - add a bidirectional weighted edge i.e add (destination node, edge weight) to the neighbor nodes list of the source node and vice versa
- define a helper function for DFS <br>
*dfs(current node)*
  - mark res as nonlocal 
    - *this is to access outer function variables in nested functions*
    - *res is a float variable, not an object like the seen hashset, and to update (reassign) a variable defined outside the function, we need to declare it as nonlocal*
  - if the node is in the hashset, it is already visited
    - return
  - add the node to the set to mark it visited
  - traverse the neighbors of the current node <br>
  *for each neighbor node, edge weight of the current node*
    - update res by taking min of the current edge weight and previous min 
    - do dfs to explore the neighbor node
- start DFS from city 1 to visit all connected cities
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph creation + graph traversal) → O(adjacency list creation + adjacency list DFS) → O(E + V) → O(V + E) → O(n + n) → O(n)
    - *V = total vertices/nodes = cities = n*
    - *E = total edges = roads*
    - *each node is visited once, and each edge is processed once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(graph + DFS recursion stack + hashset) → O((V + E) + V + V) → O(V + E) → O(n)
    - *the graph stores all cities and roads*
    - *the DFS recursion stack can store up to V cities*
    - *the visited set stores up to V cities*


## Code

```python
class Solution:

    def minScore(self, n: int, roads: List[List[int]]) -> int:
        res = float('inf')
        seen = set()
        graph = defaultdict(list)
        for u, v, distance in roads:
            graph[u].append((v, distance))
            graph[v].append((u, distance))
        
        def dfs(node: int) -> None:
            nonlocal res
            if node in seen:
                return
            seen.add(node)
            for neighbor, distance in graph[node]:
                res = min(res, distance)
                dfs(neighbor)

        dfs(1)
        return res
        
```
