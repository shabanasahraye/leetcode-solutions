# <center> 1905. Count Sub Islands </center>


## Problem Description
[Click here](https://leetcode.com/problems/count-sub-islands/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given two m x n binary grids and need to count islands in grid2 that are completely contained within grid1.

The grids can be viewed as graphs where each land cell is connected to its 4 neighbors. Use DFS to find each island in grid2 and check if all corresponding cells in grid1 are also land. If every cell in the island exists in grid1, it is a sub-island.

For marking a cell as visited, use a hash set or update its value to 2 (we can use any value except 1).


## Approach
<!-- Describe your approach to solving the problem. -->
- set m = total rows and n = total columns
- define a helper function for DFS <br>
*dfs(row, column)*
    - if the indices are out of bounds or the current cell in grid2 is not land (value != 1), 
        - return 1 because this path does not invalidate the island
    - else the current cell is an unvisited land cell (start of an island in grid2) 
        - mark it as visited by updating the value to 2
        - visit its 4 neighbors (left, right, top, bottom) using DFS and check the corresponding cells in grid1   
            - return 1 only if the current cell and all neighboring cells belong to a sub-island else return  0
- set res = 0 to count the sub-islands
- for each cell in grid2
    - if it is an unvisited land cell (value = 1), we found the start of an island
        - do dfs to find other cells on the island and check if it is a sub-island and add the result to res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph traversal) → O(adjacency matrix DFS) → O(V) → O(m × n)
    - *V = total vertices/nodes = total grid cells = m × n*
    - *DFS visits each cell in grid2 at most once*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(DFS recursion stack) → O(recursion depth) → O(V) → O(m × n)
    - *recursion depth can reach V in the worst case when the entire grid is one island*


## Code

```python
class Solution:

    def countSubIslands(self, grid1: List[List[int]], grid2: List[List[int]]) -> int:
        m, n = len(grid2), len(grid2[0])

        def dfs(r: int, c: int) -> int:
            if r < 0 or c < 0 or r == m or c == n or grid2[r][c] != 1:
                return 1
            grid2[r][c] = 2
            return dfs(r + 1, c) & dfs(r - 1, c) & dfs(r, c - 1) & dfs(r, c + 1) & grid1[r][c]

        res = 0
        for r in range(m):
            for c in range(n):
                if grid2[r][c] == 1:
                    res += dfs(r, c)
        return res
```
