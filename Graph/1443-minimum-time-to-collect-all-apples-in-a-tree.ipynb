# <center> 1443. Minimum Time to Collect All Apples in a Tree </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-time-to-collect-all-apples-in-a-tree/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
This is a graph problem because we are given n nodes and an edges list. We need to find the minimum time to collect all apples and return to the root node.

The time is equal to the number of edges traversed because crossing one edge takes one second.

Do DFS to visit every node. If a subtree contains an apple, count the edges needed to visit that subtree and return to the current node. <br>
*1 second to traverse an edge + 1 second to return*


## Approach
<!-- Describe your approach to solving the problem. -->
- create a hashmap to represent the graph (adjacency list)
    - *key = node*
    - *value = list of neighbor nodes*
    - *we can also use a 2D array for the adjacency list*
- traverse the edges list and fill the graph
    - for each edge between nodes a and b
        - add forward edge i.e add node b to the neighbor list of node a
        - add backward edge i.e node a to the neighbor list of node b
- define a helper function for DFS <br>
*dfs(current node, parent node)*
    - set time = 0 to track the time required to collect all apples in the current subtree and return to the current node
    - traverse the neighbor nodes of the current node
        - if the neighbor is not the parent node
            - recursively collect apples in the child subtree and update time
      - if the current subtree contains an apple (time > 0 or current node has an apple) and the current node is not the root node
          - add 2 to time 
              - *2 = 1 second to traverse the edge to this node + 1 second to return*
      - return time
- start dfs from the root node and return the result


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph creation + graph DFS) → O(adjacency list creation + adjacency list DFS) → O(E + V + E) → O(V + E) 
  - *V = total vertices/nodes = n*
  - *E = total edges*
  - *each edge is added once to the graph and traversed at most twice during DFS*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(graph + DFS recursion stack) → O(graph + recursion depth) → O((V + E) + V) → O(V + E)
    - *the graph stores all nodes and edges*
    - *recursion depth is V because DFS may visit every node before returning (worst case: the tree is a single chain)*


## Code

```python
class Solution:

    def minTime(self, n: int, edges: List[List[int]], hasApple: List[bool]) -> int:
        adj = defaultdict(list)
        for a, b in edges:
            adj[a].append(b)
            adj[b].append(a)

        def dfs(node, parent):
            time = 0
            for neighbor in adj[node]:
                if neighbor != parent:
                    time += dfs(neighbor, node)
            if (time or hasApple[node]) and node != 0:
                time += 2
            return time
            
        return dfs(0, None)
        
```
