## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given a tree where each node is a city and each edge is a road. Every city has one representative who needs to reach the capital (node 0). Each car has a limited number of seats, and we need to find the minimum fuel cost.

Use DFS to count how many representatives are in each subtree. For every child node, calculate how many cars are needed to move those representatives to its parent. The number of cars required is `ceil(representatives / seats)`.

*`ceil` is used to round a number up to the nearest integer.*  
*Example: representatives = 1, seats = 2*  
*cars required = ceil(1 / 2) = ceil(0.5) = 1*

The total fuel cost is the sum of the cars needed from every child node to its parent.


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = 0 to track the total fuel
- create a hashmap for the adjacency list representation of the graph
    - *key = node*
    - *value = list of neighbor nodes*
- create the graph (adjacency list)
    - for each edge (source node, destination node)
        - add a bidirectional edge from the source to the destination i.e add destination node to the neighbors list of the source node and vice versa
- define a helper function for DFS <br>
*dfs(current node, parent node)*
    - mark res as nonlocal
        - *(this is to access outer function variables in nested functions)*
    - set total = 1 to track the total representatives in the current subtree
        - *(1 because each city has one representative)*
    - for each neighbor (child node) of the current node
        - if the neighbor is not the parent i.e it is a child node, *(this is to make sure we are not moving backward)*
            - do DFS to find the total representatives in the child subtree
            - add the child subtree representatives to the current subtree total
            - find the total cars required for the child node passengers and update res
    - return total
- start DFS from the capital (node 0)
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(graph creation + graph traversal) → O(E + (V + E)) → O(V + E) → O(n)
    - *V = total vertices or nodes = cities*
    - *E = total edges = roads*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(graph + DFS recursion stack) → O(adjacency list + recursion depth) → O((V + E) + V) → O(V + E) → O(n + n) → O(n)


## Code

```python
class Solution:
    
    def minimumFuelCost(self, roads: List[List[int]], seats: int) -> int:
        res = 0
        graph = defaultdict(list)
        for u, v in roads:
            graph[u].append(v)
            graph[v].append(u)
        
        def dfs(node: int, parent: int) -> None:
            nonlocal res
            total = 1
            for neighbor in graph[node]:
                if neighbor != parent:
                    cur = dfs(neighbor, node)
                    total += cur
                    res += ceil(cur / seats)
            return total

        dfs(0, None)
        return res
        
```
