# <center> 1557. Minimum Number of Vertices to Reach All Nodes </center>


## Problem Description
[Click here](https://leetcode.com/problems/minimum-number-of-vertices-to-reach-all-nodes/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find the smallest set of nodes from which we can reach every node.

A node with an incoming edge (in-degree > 0) can already be reached from another node. A node with no incoming edges (in-degree = 0) cannot be reached from any other node.

Find and return all nodes with in-degree 0.


## Approach
<!-- Describe your approach to solving the problem. -->
- set in-degree = a list of size n to count the incoming edges for each node
    - *index = node*
    - *value = number of incoming edges*
- traverse the edges list <br>
*for each edge (source node, destination node)*
    - increment the in-degree of the destination node
- traverse the in-degree list and add every node with in-degree 0 to the result list
- return the result list


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(edges traversal + in-degree traversal) → O(E + V) → O(V + E)
    - *V = total vertices/nodes = n*
    - *E = total directed edges*
    - *the first traversal computes the in-degree of every node, and the second traversal finds all nodes with in-degree 0*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(in-degree list + result list) → O(V + V) → O(V)
    - *the in-degree list stores one count for each node*
    - *the result list stores at most V nodes*


## Code

```python
class Solution:

    def findSmallestSetOfVertices(self, n: int, edges: List[List[int]]) -> List[int]:
        in_degree = [0] * n
        for src, dst in edges:
            in_degree[dst] += 1
        return [node for node, degree in enumerate(in_degree) if not degree]
        
```
