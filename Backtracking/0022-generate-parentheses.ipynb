# <center> 22. Generate Parentheses </center>


## Problem Description
[Click here](https://leetcode.com/problems/generate-parentheses/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to generate all valid combinations of parentheses.

This is a backtracking problem because we need to explore all possible valid choices of `(` and `)` at each step.

Use DFS/backtracking to generate all valid balanced parentheses combinations. At each step:
- add `(` if we still have opening brackets left
- add `)` only if there are more opening brackets than closing brackets


## Approach
<!-- Describe your approach to solving the problem. -->
- set res = a list to store the result i.e valid combinations
- define a helper function for DFS <br>
    - *dfs(left brackets count, right brackets count, current string)*
        - if the string contains n opening and n closing brackets, the combination is complete
            - add current string to the result list
            - return
        - else the current combination is incomplete
          - if left brackets count is less than n
            - increment the left bracket count, add an opening bracket to the current string, and call dfs()
         - if right brackets count is less than left brackets count
           - increment the right bracket count, add a closing bracket to the current string, and call dfs()
- start DFS with the initial left and right bracket counts as 0 and an empty string
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(recursion) → O(calls at each level ^ total levels) → O(total valid combinations ^ total levels) → O($C_n^n$)
    - *total valid combinations = nth Catalan number because:*
        - *n = 1 → 1 combination: ()*
        - *n = 2 → 2 combinations: ()(), (())**
        - *n = 3 → 5 combinations: ((())), (()()), (())(), ()(()), ()()()*
        - *n = 4 → 14 combinations*
        - *1, 2, 5, 14, ... form the Catalan sequence*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(recursion stack + result list) → O(recursion depth/levels + total valid combinations × each combination size) → O(n + Catalan number × n) → O(n + $C_n$ × n) → O($C_n$ × n) → O(n × $C_n$)
    - *recursion depth = n left + n right = 2n = O(n)*


## Code

```python
class Solution:

    def generateParenthesis(self, n: int) -> List[str]:
        res = []
        
        def dfs(left, right, s):
            if left == right == n:
                res.append(s)
                return
            if left < n:
                dfs(left + 1, right, s + '(')
            if right < left:
                dfs(left, right + 1, s + ')')

        dfs(0, 0, '')
        return res
        
```
