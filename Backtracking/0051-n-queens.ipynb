# <center> 51. N-Queens </center>


## Problem Description
[Click here](https://leetcode.com/problems/n-queens/description/)


## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We are given n × n chessboard and n queens. We need to return the valid board configurations. A configuration is valid if the queens are placed in a way that they will not attack each other.

A queen can move in any direction i.e top, bottom, left, right, and across the diagonals. We need to place each queen in a different row, column, and diagonal so they will not be able to attack.

This is a backtracking problem because we place one queen at a time, explore possible positions, and undo the placement when it leads to a conflict.

Use backtracking to explore all possible paths/configurations row by row and check if it is valid
- a path is valid if there is no queen placed in the same row, column, or diagonal
- to speed up the process, don't check every cell, there are n queens and n rows and we need to place each in a row, so for every row, find a position (column) that will not result in a clash
    - for this track the occupied columns and diagonals
    - use hash set to track/search, insert, and remove in O(1) time
    - the left diagonal can be determined by (row - column) 
        - along this diagonal, when we move down, the row and column both increase by 1, so (row - column) remains constant
    - the right diagonal can be determined by (row + column)
        - along this diagonal, when we move down, the row increases by 1 and the column decreases by 1, so (row + column) remains constant


## Approach
<!-- Describe your approach to solving the problem. -->
- create three sets to track the occupied columns, left diagonals, and right diagonals
- initialize a board of size n × n with dots to represent the empty cells
- create a list to store the valid board configurations
- define a helper function for backtracking <br>
*backtrack(row index)*
    - if current row index is equal to n, we have traversed all rows and placed all queens
        - add the current configuration to the result list because it is valid
            - *convert the current n x n board into a list of strings, with each row represented as one string*
        - return
    - for each column in the current row
        - if there is a clash i.e current col or right diagonal or left diagonal is occupied
            - skip because we can't choose it to place the queen
        - else we can place a queen in the current cell
            - add current column, left and right diagonal to the sets to mark them as occupied
            - place queen in the current cell i.e mark current cell as 'Q'
            - backtrack to next row
            - unmark the column and diagonals from the sets
            - remove queen from the current cell
- start backtracking from the first row at index 0
- return the result


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: O(backtracking) → O(total recursive calls × work per call) → O(total states explored × work per call) → O(n! × n) → O(n x n!)
  - *branching factor is variable because valid columns decrease at each level*
  - *as the branching factor is variable, total recursive calls ≈ total states explored*
  - *work per call is O(n) because we check up to n columns in each row*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: O(recursion stack + sets + board + output) → O(recursion depth + sets + board + all valid boards) → O(recursion depth + sets + board + total boards × board size)→ O(n + 3n + n² + k × n²) → O(n² + k × n²)
  - *recursion depth is n because we place one queen per level*
  - *sets store at most n columns, n left diagonals, and n right diagonals*
  - *board uses nxn space*


## Code

```python
class Solution:

    def solveNQueens(self, n: int) -> List[List[str]]:
        cols = set()
        right_diag = set() 
        left_diag = set()
        board = [['.'] * n for _ in range(n)]
        res = []

        def backtrack(r: int) -> None:
            if r == n:
                res.append([''.join(row) for row in board])
                return
            for c in range(n):
                if c in cols or (r + c) in right_diag or (r - c) in left_diag:
                    continue
                cols.add(c)
                right_diag.add(r + c)
                left_diag.add(r - c)
                board[r][c] = 'Q'
                backtrack(r + 1)
                cols.remove(c)
                right_diag.remove(r + c)
                left_diag.remove(r - c)
                board[r][c] = '.'

        backtrack(0)
        return res
        
```
