## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
We need to find up to three products for each prefix of the search word.

- **Approach 1: Two Pointers**
    - sort the products to group similar words (products)
    - use two pointers to find the products matching each prefix
- **Approach 2: Binary Search**
    - sort the products to group similar words (products)
    - use binary search to find the first product that can match each prefix because the next two products may also match


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 1: Two Pointers
- set res = a list to store the suggested products for each search word prefix
- sort products
- set left pointer l = leftmost index to track the start of the valid product range
- set right pointer r = rightmost index to track the end of the valid product range
- for each character c at index i of the search word
    - find the first valid product index l
        - keep moving l until the product at position l is invalid for suggestion i.e either the product word size is too short or the current char is not equal to the corresponding character in the product
    - find the last valid product index r using the same logic
    - if l <= r, we can choose some products
        - choose up to 3 products from the valid range and add them to the result
    - else add an empty list to the result
- return res 

### Approach 2: Binary Search
- set res as a list to store the suggestions
- for each character in searchWord
    - create the current prefix
    - use binary search to find the first product greater than or equal to the prefix
        - set l = 0 and r = len(products) to define the binary search range
        - while l < r
            - find the midpoint m
            - if products[m] < prefix
                - move l to m + 1 to search in the right half
            - else
                - move r to m to search in the left half
    - after binary search, l is the first product that is greater than or equal to the prefix
    - create an empty suggestions list
    - check at most 3 products starting from l
        - if the product starts with the prefix
            - add it to suggestions
    - add suggestions to res
- return res


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Two-Pointers Approach: O(sort + search word traversal + products traversal) → O(nlogn + m + n) → O(nlogn + n)
        - *n = number of products*
        - *m = length of searchWord*
        - *l and r move through the products only once, so both while loops together take O(n)*
    - Binary Search Approach: O(sort + search word traversal × binary search + result creation) → O(nlogn + m × logn + 3m) → O(nlogn + mlogn)


<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Two-Pointers Approach: O(sort + result list) → O(n + search word size) → O(n + m) → O(m + n)
    - Binary Search Approach: O(sort + result list) → O(n + search word size) → O(n + m) → O(m + n)

```python
class Solution:
    
    def suggestedProducts(self, products: List[str], searchWord: str) -> List[List[str]]:
        
        # Approach 1: Two Pointers
        # res = []
        # products.sort()
        # l, r = 0, len(products) - 1
        # for i, c in enumerate(searchWord):
        #     while l <= r and (i >= len(products[l]) or c != products[l][i]): l += 1
        #     while l <= r and (i >= len(products[r]) or c != products[r][i]): r -= 1
        #     if l <= r:
        #         res.append(products[l: min(l + 3, r + 1)])
        #     else:
        #         res.append([])
        # return res

        # Approach 2: Binary Search
        res = []
        products.sort()
        for i in range(1, len(searchWord) + 1):
            prefix = searchWord[:i]
            l, r = 0, len(products)
            while l < r:
                m = (l + r) // 2
                if products[m] < prefix:
                    l = m + 1
                else:
                    r = m
            suggestions = []
            for j in range(l, min(l + 3, len(products))):
                if products[j].startswith(prefix):
                    suggestions.append(products[j])
            res.append(suggestions)
        return res


        
```
