## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
Coins required for k rows = k × (k + 1) / 2

- **Approach 1: Math**
    - we only need to compute (-1 + sqrt(1 + 8 × n)) / 2
    - because:
        - k × (k + 1) / 2 = n
        - k² + k = 2n
        - k² + k - 2n = 0 → quadratic equation (ax² + bx + c = 0)
        - substitute the values (a = 1, b = 1, c = -2n) into the quadratic formula (-b ± √(b² - 4ac)) / 2a
        - (-1 + √(1 + 8n)) / 2  
- **Approach 2: Binary Search**
    - the range is 1 to n, consider it a sorted array and use binary search
    - find coins required for the midpoint (row)
        - if the required coins are less than n, search on the right side to find rows that require more coins
        - else search on the left side


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: Binary Search
- set left pointer l = 1 and right pointer r = n
- while l <= r
    - compute the midpoint
    - compute the coins required to build mid complete rows
    - if required coins are equal to n
        - return mid 
    - if coins < n
        - move l to search on the right for a larger number of rows that require more coins
    - else coins > n
        - move r to search on the left for a smaller number of rows that require fewer coins
- if the loop completes without returning anything, an exact match is not found, return r as the no. of complete rows


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Binary Search Approach: O(loop) → O(each time search space is divided by 2) → O($log_2(\text{search range})$) → O($log_2{n}$)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Binary Search Approach: O(1)


## Code

```python
class Solution:

    def arrangeCoins(self, n: int) -> int:

        # Approach 2: Binary Search
        l, r = 1, n
        while l <= r:
            mid = (l + r) // 2
            coins = (mid * (mid + 1)) // 2
            if coins == n:
                return mid
            elif coins < n:
                l = mid + 1
            else:
                r = mid - 1
        return r
        
```
