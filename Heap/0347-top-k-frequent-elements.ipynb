## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
- **Approach 1: Sorting O(n + mlogm)**
    - use a map to store element-frequency pairs
    - sort the pairs in decreasing order of frequency
    - return the first k pairs (elements)

- **Approach 2: Min Heap O(n + mlogk)**
    - use a map to store element-frequency pairs
    - create a min-heap to store at most k pairs
    - traverse the remaining pairs, and pop the heap if the heap has more than k pairs to remove the lowest-frequency pair
    - at the end, there will be k high-freq pairs left in the heap

- **Approach 3: Max Heap O(n + klogm)**
    - use a map to store element-frequency pairs
    - create a max-heap with all m pairs 
    - pop the heap k times to get the k high-freq pairs

- **Approach 4: Bucket/Index sort O(n)** 
    - bucket sort can take O(n²) in the worst case, but here we do not actually sort the elements
    - create n + 1 buckets and place each element in the bucket corresponding to its frequency


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: Min Heap
- set res = an array to store the k most frequent elements
- set count = a hashmap with a default value of 0 for each key to store the frequency of each element
- create a min-heap to store at most k `(frequency, element)` pairs
- traverse the input array elements and get their frequencies
    - for each number in the input array
        - increment its count in the hashmap
- traverse the count hashmap <br>
*for each element and its count in the hashmap*
    - add the `(count, element)` pair to the heap
    - if the heap has more than k elements
        - pop heap to remove the element with the lowest frequency
- traverse the heap and add the elements (top-k frequent) to the result
- return res

### Approach 3: Max Heap
- set res = an array to store the k most frequent elements
- set count = a hashmap with a default value of 0 for each key to store the frequency of each element
- create a max-heap containing all `(frequency, element)` pairs
    - *Python's heap library doesn't have a max heap class. We can implement the max heap using the min heap class by negating the frequencies, so the largest frequency becomes the smallest value in the min heap and moves to the heap top*
- traverse the input array elements and get their frequencies <br>
*for each number in the input array*
    - increment its count in the hashmap
- traverse the count hashmap and add each `(count, element)` pair to the max-heap
- pop the heap k times to get top-k frequent elements and add them to res
- return res

### Approach 4: Bucket/Index Sort
- set res = an array to store the k most frequent elements
- set count = a hashmap with a default value of 0 for each key to store the frequency of each element
- set freq = a 2D array of size n + 1 to use as frequency buckets
    - *freq[f][n] stores the elements that occur f times, e.g element with frequency 4 is placed at index 4 (bucket 4)*
    - *f represents the frequency*
    - *n represents the list of elements with frequency f*
- traverse the input array elements and get their frequencies <br>
*for each number in the input array*
    - increment its count in the hashmap
- traverse the count hashmap and fill the frequency array
*for each element and its count in the hashmap*
    - add the element to the bucket at the index equal to its frequency
- traverse the frequency array in reverse to get the most frequent element first
    - for each index in the freq array
        - for each element in the list at the current index
            - add the element to the result list
        - return when you get k elements 


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Min Heap Approach: O(input array traversal + hashmap traversal × heap operation) → O(n + m × logk) → O(n + mlogk)
        - *m = total distinct elements*
        - *k < m ≤ n*
    - Max Heap Approach: O(input array traversal + hashmap traversal + k heap operations) → O(n + m + k × logm) → O(n + klogm)
    - Bucket Sort Approach: O(input array traversal to fill hashmap + hashmap traversal to fill frequency array + frequency array traversal for k elements) → O(n + m + k) → O(n)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Min Heap Approach: O(frequency hashmap + heap + result array) → O(m + k + k) → O(m)
    - Max Heap Approach: O(frequency hashmap + heap + result array) → O(m + m + k) → O(m)
    - Bucket Sort Approach: O(frequency hashmap + frequency array + result array) → O(m + n + k) → O(n)
        - *the frequency array has n + 1 buckets, but only n elements are stored across all buckets*
