## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
This is similar to [Kth Largest Element in an Array](https://leetcode.com/problems/kth-largest-element-in-an-array/description/) but here we also need to handle incoming elements.

Use a Min Heap of size k to store k largest elements of the stream
- the min heap top contains the smallest element and can be accessed in O(1) time
- in this case, the top would be the Kth largest element because there are k largest elements in the heap.
- to process a new element of the stream, compare it with the heap top element
    - if it is greater than the top, replace the top with it
- we never store the full stream, only the k largest elements


## Approach
<!-- Describe your approach to solving the problem. -->
**init()**
- set min_heap = an empty list
- set k
- for each number in the input list
    - if there are less than k elements in the heap, add the number to the heap
    - else there are k elements in the heap. If the number is greater than the heap top
        - replace the heap top i.e pop heap to remove the top element and push the current number 
         - *heapreplace() is equivalent to pop first and then push*
- *note: if you heapify the list first and then pop elements until k elements are left, the time complexity will be O(n + (n-k)logn), and the space complexity will be O(n)*

**add()**
- if there are less than k elements in the heap
    - add the number to the heap
- else there are k elements in the heap
    - if the number is greater than the top element
        - replace the top
- return the heap top


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity:
    - init(): O(array traversal × heap pop push) → O(n × (log(heap size) →O(n × logk)
    - add(): O(heap pop push) → O(log(heap size)) → O(logk)

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - init(): O(heap size) → O(k)
    - add(): O(1)


## Code

```python
class KthLargest:

    def __init__(self, k: int, nums: List[int]):
        self.min_heap = []
        self.k = k
        for num in nums:
            if len(self.min_heap) < k:
                heapq.heappush(self.min_heap, num)
            elif num > self.min_heap[0]:
                heapq.heapreplace(self.min_heap, num)

    def add(self, val: int) -> int:
        if len(self.min_heap) < self.k:
            heapq.heappush(self.min_heap, val)
        elif val > self.min_heap[0]:
            heapq.heapreplace(self.min_heap, val)
        return self.min_heap[0]
        


# Your KthLargest object will be instantiated and called as such:
# obj = KthLargest(k, nums)
# param_1 = obj.add(val)
```
