## Intuition
<!-- Describe your first thoughts on how to solve this problem. -->
- **Approach 1: Max heap O(n+k)logn**
  - in the max heap, the largest element is at the top and can be accessed in O(1) time
  - create a max heap with all elements and then pop heap k times to get the kth largest element

- **Approach 2: Min heap O(nlogk)**
  - in the min heap, the smallest element is at the top and can be accessed in O(1) time
  - use a Min Heap of size k to store k largest elements
      - create a min heap with k elements
      - traverse the remaining elements
          - if an element is greater than the heap top element, pop heap to remove it and insert the greater element
      - in the end, the heap will contain k largest elements
          - Kth largest element will be at the heap top because there are k largest elements in the heap


## Approach
<!-- Describe your approach to solving the problem. -->
### Approach 2: Min Heap
- set min_heap = an empty list
- for each number in the input list
      - if there are less than k elements in the heap
          - add the number to the heap
      - else there are k elements in the heap, if the number is greater than the heap top element
            - replace the heap top i.e pop heap to remove the top element and push the current number
                - *heapreplace() is equivalent to pop first and then push*
- *note: if you heapify the list first and then pop elements until k elements are left, the time complexity will be O(n + (n-k)logn), and the space complexity will be O(n)*
- return heap top element


## Complexity
<!-- Add your time complexity here, e.g. $$O(n)$$ -->
- Time complexity: 
    - Min Heap Approach: O(array traversal × replacing heap top) → O(n × log(heap size)) → O(n × logk)
        - *each heap operation takes O(logk) because the heap contains at most k elements*

<!-- Add your space complexity here, e.g. $$O(n)$$ -->
- Space complexity: 
    - Min Heap Approach: O(heap size) → O(k)


## Code

```python
class Solution:
    
    def findKthLargest(self, nums: List[int], k: int) -> int:

        # MinHeap Approach
        min_heap = []
        for num in nums:
            if len(min_heap) < k :
                heapq.heappush(min_heap, num)
            elif num > min_heap[0]:
                heapq.heapreplace(min_heap, num)
        return min_heap[0]
        
```
